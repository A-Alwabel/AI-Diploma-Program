## 📌 Real case: the $1,000,000 matrix

In October 2006 Netflix published a single matrix and offered a million dollars to anyone who could
fill in its blanks. The matrix had **480,189 rows** (customers), **17,770 columns** (films) and
**100,480,507 ratings** — which sounds enormous until you notice that 480,189 × 17,770 is about
8.5 billion cells, so **roughly 99% of the matrix was empty**. The task: predict the missing entries
at least 10% more accurately (RMSE) than Netflix's own Cinematch system. The prize was claimed in 2009
by *BellKor's Pragmatic Chaos* with a 10.06% improvement (test RMSE 0.8567 against the 0.8572 target).
Every serious entry, the winner included, was built on **matrix factorization** — writing that huge
sparse matrix as the product of two thin ones. Netflix later said it never put the winning ensemble
into production: the engineering cost of the full blend was not worth the accuracy it bought
(Netflix Tech Blog, 2012).

**What goes wrong without this.** Treat the problem as 480,189 separate customers and you have
480,189 separate prediction problems, with no way for one customer's taste to inform another's.
The entire value of the matrix view is that a *single* operation — a product of two matrices —
answers all 8.5 billion questions at once, and lets the pattern "people who liked A liked B" flow
between rows that never meet.

**The same reasoning appears elsewhere:** a hospital's patients × lab-tests table, where the missing
cells are tests nobody ordered; a retailer's stores × products sales table; a university's
students × courses grade table. Same shape, same trick.

# Part 2: HOW

## Implementation and Practical Application

### How Linear Algebra Works in Machine Learning

Let's understand **HOW** linear algebra operations work and **WHY** each operation is used:

#### 1. **WHY Vectors and Matrices?**
- **WHY**: ML algorithms process data in batches efficiently
- **HOW**: Data is organized as matrices where:
  - Each **row** = one data point (sample)
  - Each **column** = one feature
  - Matrix operations process all samples simultaneously

#### 2. **WHY Matrix Multiplication?**
- **WHY**: Neural networks need to transform data through layers
- **HOW**: Matrix multiplication `A @ B` computes:
  - Each output element = dot product of row from A and column from B
  - This allows parallel computation of all transformations

#### 3. **WHY Eigenvalues/Eigenvectors?**
- **WHY**: Find the most important directions in data (for PCA in Module 04)
- **HOW**: Eigenvalue decomposition finds:
  - **Eigenvectors**: Directions of maximum variance
  - **Eigenvalues**: How much variance each direction explains

#### 4. **WHY Linear Transformations?**
- **WHY**: Transform features to better representations for ML
- **HOW**: Multiply data matrix by transformation matrix to:
  - Rotate, scale, or project data
  - Create new feature combinations

### Step-by-Step: Working with Linear Algebra in Python

**Step-by-Step Guide to Working with Linear Algebra in Python:**

Here's how to apply linear algebra concepts in practice:

2. **Matrix Operations**: Addition, multiplication, transpose, inverse

### Best Practices

**Best Practices for Linear Algebra in ML:**

- **Vectorization**: Always use NumPy vectorized operations instead of loops for better performance
- **Matrix Shapes**: Always check matrix dimensions before operations to avoid errors
- **Memory Efficiency**: Use appropriate data types (float32 vs float64) based on precision needs
- **Eigenvalue Computation**: Use `np.linalg.eig()` for small matrices, specialized methods for large ones
- **Matrix Conditioning**: Check condition number before computing inverses to avoid numerical instability

### Common Pitfalls and How to Avoid Them

**Common Pitfalls and How to Avoid Them:**

1. **Shape Mismatches**: Always verify matrix dimensions match before multiplication
   - **Solution**: Use `print(matrix.shape)` to debug dimension issues

2. **Broadcasting Confusion**: Understanding when NumPy broadcasts vs when it errors
   - **Solution**: Learn broadcasting rules: dimensions must be compatible (1 or same size)

3. **In-place Operations**: Modifying arrays in place can cause unexpected behavior
   - **Solution**: Use explicit copies (`arr.copy()`) when needed

4. **Eigenvalue Ordering**: Eigenvalues may not be sorted - always sort if order matters
   - **Solution**: Use `np.argsort()` to get sorted indices

### Advanced Topics

**Advanced Topics for Further Learning:**

- **Singular Value Decomposition (SVD)**: More general than eigenvalue decomposition
- **Matrix Factorization**: Techniques like LU, QR, Cholesky decomposition
- **Sparse Matrices**: Efficient representation for matrices with many zeros
- **Tensor Operations**: Extending matrix operations to higher dimensions
- **Numerical Linear Algebra**: Stable algorithms for large-scale computations

```python
# Example: Basic Linear Algebra Operations in Python
# This shows HOW each operation works and WHY it's used in ML

import numpy as np

print("=" * 60)
print("HOW Linear Algebra Operations Work in ML")
print("=" * 60)

# 1. Creating vectors and matrices
# WHY: This is HOW data is represented in ML
# HOW: NumPy arrays organize data efficiently
vector = np.array([1, 2, 3])  # A single data point
matrix = np.array([[1, 2], [3, 4]])  # Multiple data points

print("\n1. Data Representation:")
print(f"   Vector (1 data point, 3 features): {vector}")
print(f"   Matrix (2 data points, 2 features):\n{matrix}")
print("   WHY: ML algorithms process data in this format")
print("   HOW: Each row = one sample, each column = one feature")

# 2. Matrix multiplication
# WHY: Neural networks transform data through layers
# HOW: Matrix multiplication computes transformations
A = np.array([[1, 2], [3, 4]])  # Transformation matrix
B = np.array([[5, 6], [7, 8]])  # Data matrix
result = np.dot(A, B)  # or A @ B

print("\n2. Matrix Multiplication (Neural Network Layers):")
print(f"   A (transformation):\n{A}")
print(f"   B (data):\n{B}")
print(f"   Result (transformed data):\n{result}")
print("   WHY: Neural networks use this to transform data")
print("   HOW: Each element = sum of products (row from A × column from B)")

# 3. Transpose
# WHY: Change data orientation for different operations
# HOW: Swap rows and columns
print("\n3. Transpose (Changing Data Orientation):")
print(f"   Original A:\n{A}")
print(f"   Transposed A.T:\n{A.T}")
print("   WHY: Needed for certain operations (e.g., computing gradients)")
print("   HOW: Rows become columns, columns become rows")

# 4. Dot product (vector multiplication)
# WHY: Measures similarity or computes weighted sums
# HOW: Element-wise multiplication then sum
v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])
dot_product = np.dot(v1, v2)
print("\n4. Dot Product (Similarity/Weighted Sum):")
print(f"   v1: {v1}")
print(f"   v2: {v2}")
print(f"   Dot product: {dot_product}")
print("   WHY: Used in neural networks (weighted sum of inputs)")
print("   HOW: (1×4) + (2×5) + (3×6) = 4 + 10 + 18 = 32")

# 5. Matrix properties for ML
print("\n5. Matrix Properties Used in ML:")
determinant = np.linalg.det(matrix)
print(f"   Determinant of matrix: {determinant:.2f}")
print("   WHY: Tells us if transformation is invertible")
print("   HOW: Computed from matrix elements (ad - bc for 2×2)")

# Additional concepts you can explore:
# - Matrix inverse: Used for solving systems of equations
# - Eigenvalues and eigenvectors: Already covered above - used in PCA (Module 04)
# - Linear transformations: Already demonstrated above - fundamental to neural networks
```

### 🧪 Does "use NumPy, not loops" actually matter?

The Best Practices section above says *vectorize*. Here is the measurement behind that advice.
Baseline: matrix multiplication written out as the definition says, with three nested Python loops.
Change one thing: hand the identical job to NumPy's `@`. Then read the two numbers.

```python
# Comparative experiment: triple loop vs NumPy's @ operator.
# The matrices are REAL: X is sklearn's 1797 x 64 handwritten-digit matrix, so X.T @ X is the
# 64 x 64 pixel co-occurrence matrix that a real image pipeline actually builds.

import time
from sklearn.datasets import load_digits

X = load_digits().data          # 1797 real 8x8 scanned digits, flattened to 64 pixels each
A, B = X.T, X                   # (64 x 1797) @ (1797 x 64)


def matmul_loops(A, B):
    """Matrix product written exactly as the definition states: three nested loops."""
    n, k = A.shape
    _, m = B.shape
    out = np.zeros((n, m))
    for i in range(n):
        for j in range(m):
            total = 0.0
            for t in range(k):          # this inner line runs 64 x 64 x 1797 times
                total += A[i, t] * B[t, j]
            out[i, j] = total
    return out


t0 = time.perf_counter(); slow = matmul_loops(A, B); t_loops = time.perf_counter() - t0
t0 = time.perf_counter(); fast = A @ B;              t_numpy = time.perf_counter() - t0

print("Comparative experiment: triple loop vs NumPy @")
print("=" * 62)
print("Job: (64 x 1797) @ (1797 x 64) on real handwritten-digit data")
print(f"   Python loops : {t_loops * 1000:10.1f} ms")
print(f"   NumPy @      : {t_numpy * 1000:10.1f} ms")
print(f"   Speed-up     : {t_loops / max(t_numpy, 1e-9):10.0f}x")
print(f"   Largest disagreement between the two answers: {np.abs(slow - fast).max():.2e}")
print()
print("Conclusion the numbers support: the two answers agree to floating-point precision, so")
print("'vectorize' is not an approximation - it is the SAME arithmetic dispatched to compiled")
print("code. That speed-up is the whole reason a 60-million-parameter network is trainable.")
```

### 🧩 The Netflix trick, on data we actually have

The case that opened this notebook claimed a huge, mostly-empty matrix can be written as the product of **two thin ones**. That is a claim about pictures, not about prose — so let's draw it. We take the same real 1797 × 64 handwritten-digit matrix used in the timing experiment above, keep only the *k* strongest factors of its SVD (`X ≈ U_k S_k V_kᵀ`), and rebuild the images from those factors alone. If the claim is true, a small *k* should already look like a digit.

```python
# Frame-grid storyboard: rebuild real digit images from a rank-k factorization.
# WHY: this is the Netflix idea in miniature - store two thin matrices instead of one fat one.
# HOW: the SVD X = U S V^T sorts the factors by importance; keeping the first k is the best
#      possible rank-k approximation of X (Eckart-Young theorem).

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits

digits = load_digits()
X = digits.data                       # 1797 real scanned digits x 64 pixels - the SAME matrix as above
U, S, Vt = np.linalg.svd(X, full_matrices=False)

ranks = [1, 2, 4, 8, 16, 64]          # 64 = full rank, i.e. the original data, no approximation
show_rows = [642, 1332]               # two real samples from the dataset (a 0 and a 3)
n_samples, n_pixels = X.shape
full_numbers = n_samples * n_pixels

# Reconstruct once per rank (cheap: the biggest matrix multiply here is 1797x64), then display.
with np.errstate(all="ignore"):   # some BLAS builds raise a harmless FP flag on big matmuls
    recon = {k: (U[:, :k] * S[:k]) @ Vt[:k] for k in ranks}

fig, axes = plt.subplots(len(show_rows), len(ranks), figsize=(11, 4.2))
for r, row_idx in enumerate(show_rows):
    for c, k in enumerate(ranks):
        ax = axes[r, c]
        ax.imshow(recon[k][row_idx].reshape(8, 8), cmap="gray_r", vmin=0, vmax=16)
        ax.set_xticks([]); ax.set_yticks([])
        if r == 0:
            stored = k * (n_samples + n_pixels)
            label = (f"rank {k}\n= the original data" if k == n_pixels
                     else f"rank {k}\n{100 * stored / full_numbers:.0f}% of the numbers")
            ax.set_title(label, fontsize=9)
    axes[r, 0].set_ylabel(f"true digit = {digits.target[row_idx]}", fontsize=10)
fig.suptitle("Two real digits rebuilt from k factors only — left: k = 1, right: the untouched data",
             fontsize=12, fontweight="bold")
plt.tight_layout()
plt.show()

# The step table behind the storyboard: what each frame costs and how wrong it is.
print("rank k | numbers stored | % of the full matrix | relative error ||X - X_k||_F / ||X||_F")
print("-" * 88)
normX = np.linalg.norm(X)
for k in ranks:
    stored = k * (n_samples + n_pixels)
    err = np.linalg.norm(X - recon[k]) / normX
    print(f"{k:6d} | {stored:14,d} | {100 * stored / full_numbers:19.1f}% | {err:38.3f}")
print(f"{'store X':>7} | {full_numbers:13,d} | {100.0:19.1f}% | {0.0:38.3f}")
print()
print("Note the last two rows: a FULL-rank factorization stores 104% of the numbers - more than")
print("just keeping X. Factorization only pays when k is small, which is the entire bet: the")
print("useful structure of a big matrix has to live in a few factors, or the trick buys nothing.")
```

**What to look for.** Read each row left to right. At **rank 1** the two frames are the same vertical smudge — with one factor the reconstruction cannot tell a 0 from a 3. By **rank 4** the 0 has opened its hole and the 3 has grown its two bumps. By **rank 8**, storing about 13% of the original numbers, both digits are unmistakable. The right-hand column is the untouched data itself; rank 16 is visibly smoother than it, because the discarded factors were mostly pixel noise. The table under the figure gives the exact error behind each frame. That is the Netflix bet in one picture: most of what a big matrix contains lives in a few factors, so you can store two thin matrices instead of one fat one — and, if some cells are missing, you can *predict* them from the factors you kept.

```python
# Example: Understanding the Results and Implications
# VISUAL: one pair of axes, so you can SEE the square move — not two pictures side by side.

import numpy as np
import matplotlib.pyplot as plt

# The unit square, as 5 corner points (last = first, so the outline closes).
# Columns are points, rows are (x, y) — the layout a matrix multiply expects.
original_points = np.array([[0, 0], [1, 0], [1, 1], [0, 1], [0, 0]]).T

# The transformation. WHY this one: it rotates AND stretches, so both effects show up at once.
transformation_matrix = np.array([[2, -1], [1, 2]])
transformed_points = transformation_matrix @ original_points

# WHY draw both on ONE axis: a transformation is a comparison. Two separate panels with two
# separate axis ranges hide exactly the thing you want to measure — how much bigger, how far round.
fig, ax = plt.subplots(figsize=(6.4, 6.4))

ax.fill(original_points[0], original_points[1], alpha=0.35, color="tab:blue")
ax.plot(original_points[0], original_points[1], "-o", color="tab:blue", lw=2, ms=7,
        label="before: unit square, area = 1")

ax.fill(transformed_points[0], transformed_points[1], alpha=0.25, color="tab:red")
ax.plot(transformed_points[0], transformed_points[1], "-o", color="tab:red", lw=2, ms=7,
        label="after: A @ square, area = |det A| = 5")

# The two black arrows are the whole secret of a matrix: they show where the basis vectors go,
# and their coordinates ARE the columns of A.
for col, name in zip(range(2), ["e1 = (1,0)", "e2 = (0,1)"]):
    tip = transformation_matrix[:, col]
    ax.annotate("", xy=(tip[0], tip[1]), xytext=(0, 0),
                arrowprops=dict(arrowstyle="-|>", lw=2.2, color="black"))
    ax.text(tip[0] + 0.12, tip[1] + 0.12, f"{name} lands at ({tip[0]}, {tip[1]})",
            fontsize=10, fontweight="bold")

ax.axhline(0, color="grey", lw=0.8)
ax.axvline(0, color="grey", lw=0.8)
ax.set_xlim(-2, 4); ax.set_ylim(-2, 4)
ax.set_aspect("equal")
ax.set_xlabel("x"); ax.set_ylabel("y")
ax.grid(True, alpha=0.3)
ax.legend(loc="lower right", fontsize=10)
ax.set_title("One matrix, one picture: A = [[2, -1], [1, 2]] applied to the unit square",
             fontsize=12, fontweight="bold")
plt.tight_layout()
plt.show()

print("Reading the figure:")
print("=" * 62)
det = np.linalg.det(transformation_matrix)
print(f"   Blue square area                     : 1.00")
print(f"   Red image area (measured by det A)   : {abs(det):.2f}")
print("   The corners of the red shape are the blue corners multiplied by A -")
print("   e.g. corner (1,1) -> A @ (1,1) = "
      f"{tuple(transformation_matrix @ np.array([1, 1]))}, the top-right red dot.")
print("   The shape stayed a square: this matrix rotates and scales, it does not skew.")

# Eigenvalue analysis
print("\n📊 Eigenvalue Analysis:")
eigenvals, eigenvecs = np.linalg.eig(transformation_matrix)
print(f"Eigenvalues: {eigenvals}")
print(f"|λ| (magnitudes): {np.abs(eigenvals)}")
print("\n💡 Reading this result:")
print("   - The eigenvalues are COMPLEX because this matrix rotates as well as scales:")
print("     a rotation leaves no real direction unchanged, so no real eigenvectors exist.")
print("   - That is why the figure draws NO eigenvector arrows: there are none to draw in")
print("     the real plane. The two black arrows are the images of e1 and e2, not eigenvectors.")
print(f"   - The magnitude |λ| = {np.abs(eigenvals[0]):.4f} (= √5) is the factor by which")
print("     every vector gets stretched by this rotation-scaling, and |λ|² = 5 = det A")
print("     is exactly the area factor you can see in the figure.")
print("   - Contrast with PCA (Module 04): covariance matrices are symmetric, so their")
print("     eigenvalues are always REAL and their eigenvectors give the principal directions.")

# You've now seen how linear algebra transforms data visually!
# In real ML applications, these transformations happen in neural network layers.
```

**What to look for in the figure above.** The blue patch is the unit square (area 1); the red patch drawn on the *same* axes is its image after multiplying by `A = [[2,-1],[1,2]]`. Two things are visible at once: the red shape is turned about 27° anticlockwise, and it is **five times larger in area** — which is exactly `det(A) = 5`, printed under the figure. The two black arrows are where the basis vectors land: `e1 → (2,1)` and `e2 → (-1,2)`. Read them off the picture and you have read the columns of `A`. That is the whole content of "a matrix is a transformation": the matrix is nothing but a list of the places the axes go.

## ⚠️ Where this breaks

- **Non-symmetric matrices have no real principal direction.** Look at this notebook's own eigenvalue
  output: for the transformation `[[2,-1],[1,2]]` NumPy returned `2+1j` and `2-1j`. Complex — because
  that matrix *rotates*, and a rotation leaves no real direction unchanged. The comfortable story
  "eigenvectors are the directions of maximum variance" is true only for **symmetric** matrices:
  covariance matrices, Gram matrices, kernel matrices. Elsewhere the answer is not wrong, it is not real.
- **Dense arrays assume a dense problem.** The Netflix matrix above is 99% empty. Held as a dense
  `float32` NumPy array it would need 480,189 × 17,770 × 4 bytes ≈ **34 GB**. The assumption that must
  hold before you type `np.array(...)` is that your data really is dense and rectangular. When it is not,
  you need sparse formats (`scipy.sparse`) or factorization — which is exactly what the prize winners used.
- **Cheaper alternative.** For a few dozen records with mixed types — names, dates, categories, holes —
  a pandas DataFrame and ordinary Python logic is clearer and faster to write than forcing everything
  into a numeric matrix. Linear algebra earns its keep when the same operation must run over thousands
  of rows at once.
