## Setup

Run this cell first. It defines the city map you'll route on (no external data files needed).

```python
# Set up the exercise world: a street-grid graph with some closed roads, plus the delivery jobs to route.
# Why: BFS and A* both need a graph as input - this cell builds the one every task below uses.

from collections import deque
import heapq

# City district: a 4-row x 6-column grid of intersections, named (row, col).
# A few streets are closed for construction, so some neighbors are NOT connected.
CLOSED_STREETS = {
    ((1, 1), (1, 2)),   # street closed between these two intersections
    ((2, 1), (2, 2)),
    ((0, 4), (1, 4)),
    ((2, 4), (3, 4)),
}

def build_city(rows=4, cols=6, closed=CLOSED_STREETS):
    """Build the adjacency dict for the street grid, skipping closed streets."""
    city = {(r, c): [] for r in range(rows) for c in range(cols)}
    for r in range(rows):
        for c in range(cols):
            for nb in [(r, c + 1), (r + 1, c)]:          # east + south neighbors
                if nb in city and ((r, c), nb) not in closed:
                    city[(r, c)].append(nb)
                    city[nb].append((r, c))              # streets are two-way
    return city

CITY = build_city()

# Three delivery jobs to route: (pickup, drop-off)
JOBS = [((0, 0), (3, 3)), ((0, 2), (3, 1)), ((1, 5), (3, 3))]

def manhattan(a, b):
    """Heuristic for A*: Manhattan distance between two intersections."""
    return abs(a[0] - b[0]) + abs(a[1] - b[1])

print(f"City map loaded: {len(CITY)} intersections, "
      f"{sum(len(v) for v in CITY.values()) // 2} streets "
      f"({len(CLOSED_STREETS)} closed for construction)")
print(f"Delivery jobs to route: {JOBS}")
```
