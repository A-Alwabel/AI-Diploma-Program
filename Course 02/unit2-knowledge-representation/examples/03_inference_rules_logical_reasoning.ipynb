## Introduction to Inference Rules

**Inference rules** are logical patterns that allow us to derive valid conclusions from given premises. They are the foundation of logical reasoning in AI systems.


## 🌍 The case: machines that reason, and what they have actually settled

Three inference rules look like a Latin vocabulary test. Here is what applying them mechanically, at scale, has produced.

**10 October 1996, Argonne National Laboratory.** William McCune's equational theorem prover **EQP** found a proof of the **Robbins conjecture** — an algebra question open since the 1930s that Alfred Tarski and others had failed to settle. The successful search ran about **eight days** on an RS/6000 and used roughly 30 MB of memory. The proof object it emitted is **16 lines long** and, by the authors' own account, extremely hard for a human to follow. A machine applying inference rules had closed a sixty-year-old mathematical question.

**2009, NICTA/UNSW.** The **seL4** operating-system microkernel — roughly 8,700 lines of C — was given a machine-checked proof of functional correctness against its specification: not "we tested it thoroughly" but "there is a formal derivation, verified line by line by a proof checker." seL4 now runs in avionics and defence systems where the argument for correctness has to be a proof.

**25 July 2024, Google DeepMind.** **AlphaProof** and **AlphaGeometry 2** solved four of the six problems at the International Mathematical Olympiad, scoring **28 points — the silver-medal threshold**. AlphaProof searched for proofs inside the **Lean** proof assistant, so every step it produced was mechanically checked before it counted. The reasoning engine was a machine; the referee was a program applying inference rules.

### What goes wrong without this

A chain of reasoning is only as good as its weakest step, and a *plausible* step is not a *valid* one. The most common bad step has a name — **affirming the consequent**: *"if it rains, the ground is wet; the ground is wet; therefore it rained."* It feels like modus ponens. It is not, and a sprinkler is enough to make it false.

The trap is that the failure is invisible from the outside. A conclusion reached by four valid steps and one invalid one looks exactly like a conclusion reached by five valid steps — same confidence, same tone, wrong answer. Demo 2 in Part 4 below builds exactly that argument and shows what a checker does with it. That is what these rules buy: not new knowledge, but the ability to *prove* that a conclusion follows.

````python
# Re-define the logical operators from the previous notebook so this notebook runs standalone.
# Why: every inference rule below is built from these four tiny building blocks.
def implies(p, q):
    """IMPLIES operator"""
    return not p or q

def logical_and(p, q):
    return p and q

def logical_or(p, q):
    return p or q

def logical_not(p):
    return not p

print("✅ Logical operators loaded!")
````

## Part 1: Modus Ponens

**Modus Ponens** (Latin: "method of affirming") is one of the most important inference rules:

**Pattern:** If P → Q, and P is true, then Q must be true.

**Formal Representation:**
```
P → Q
P
---
∴ Q
```

**Example:**
- If it rains, then the ground gets wet. (P → Q)
- It is raining. (P)
- Therefore, the ground is wet. (Q)


````python
# Modus Ponens: from 'P implies Q' and 'P is true', conclude 'Q is true'.
# Why: this is the single most-used inference rule - it is how rule-based systems fire a rule.

def modus_ponens(premise1_implies, premise2_p, conclusion_q):
    """
    Validate Modus Ponens inference rule.
    
    Parameters:
    - premise1_implies: Truth value of P → Q
    - premise2_p: Truth value of P
    - conclusion_q: Truth value of Q (to validate)
    
    Returns:
    - True if inference is valid, False otherwise
    """
    # Modus Ponens is valid if:
    # 1. P → Q is True
    # 2. P is True
    # 3. Then Q must be True
    if premise1_implies and premise2_p:
        return conclusion_q == True
    # If P → Q is False or P is False, Modus Ponens doesn't apply
    return True  # No contradiction if premises aren't met

# Test Modus Ponens
print("=" * 60)
print("Testing Modus Ponens:")
print("=" * 60)

# Example 1: Valid Modus Ponens
print("\nExample 1: Valid Modus Ponens")
print("Premise 1: If it rains, then ground is wet (P → Q) = True")
print("Premise 2: It is raining (P) = True")
print("Conclusion: Ground is wet (Q) = True")
valid = modus_ponens(True, True, True)
print(f"Result: {'✅ VALID' if valid else '❌ INVALID'}")

# Example 2: Invalid (contradiction)
print("\nExample 2: Invalid Modus Ponens (contradiction)")
print("Premise 1: If it rains, then ground is wet (P → Q) = True")
print("Premise 2: It is raining (P) = True")
print("Conclusion: Ground is NOT wet (Q) = False")
valid = modus_ponens(True, True, False)
print(f"Result: {'✅ VALID' if valid else '❌ INVALID (CONTRADICTION)'}")
````

## Part 2: Modus Tollens

**Modus Tollens** (Latin: "method of denying") reasons *backwards* from a failed consequence:

**Pattern:** If P → Q, and Q is false, then P must be false.

**Formal Representation:**
```
P → Q
¬Q
---
∴ ¬P
```

**Example:**
- If it rains, then the ground gets wet. (P → Q)
- The ground is NOT wet. (¬Q)
- Therefore, it is NOT raining. (¬P)

This is *contrapositive* reasoning: P → Q is logically equivalent to ¬Q → ¬P (you verified this equivalence with truth tables in the previous notebook).

````python
# Modus Tollens: from 'P implies Q' and 'Q is false', conclude 'P is false'.
# Why: reasoning backwards from a missing effect to rule out a cause - the logic behind diagnosis.

def modus_tollens(premise1_implies, premise2_not_q, conclusion_not_p):
    """
    Validate Modus Tollens inference rule.
    
    Parameters:
    - premise1_implies: Truth value of P → Q
    - premise2_not_q: Truth value of ¬Q (NOT Q)
    - conclusion_not_p: Truth value of ¬P (NOT P) to validate
    
    Returns:
    - True if inference is valid, False otherwise
    """
    # Modus Tollens is valid if:
    # 1. P → Q is True
    # 2. Q is False (¬Q is True)
    # 3. Then P must be False (¬P must be True)
    if premise1_implies and premise2_not_q:
        return conclusion_not_p == True
    return True

# Test Modus Tollens
print("=" * 60)
print("Testing Modus Tollens:")
print("=" * 60)

print("\nExample: Valid Modus Tollens")
print("Premise 1: If it rains, then ground is wet (P → Q) = True")
print("Premise 2: Ground is NOT wet (¬Q) = True")
print("Conclusion: It is NOT raining (¬P) = True")
valid = modus_tollens(True, True, True)
print(f"Result: {'✅ VALID' if valid else '❌ INVALID'}")
````

## Part 3: Disjunctive Syllogism

**Disjunctive Syllogism** deals with OR statements:

**Pattern:** If P OR Q, and P is false, then Q must be true.

**Formal Representation:**
```
P ∨ Q
¬P
---
∴ Q
```

**Example:**
- Either it is sunny OR it is raining. (P ∨ Q)
- It is NOT sunny. (¬P)
- Therefore, it is raining. (Q)


````python
# Disjunctive Syllogism: from 'P or Q' and 'not P', conclude 'Q'.
# Why: elimination reasoning - cross out the option that failed and what remains must hold.

def disjunctive_syllogism(premise1_or, premise2_not_p, conclusion_q):
    """
    Validate Disjunctive Syllogism inference rule.
    
    Parameters:
    - premise1_or: Truth value of P ∨ Q
    - premise2_not_p: Truth value of ¬P
    - conclusion_q: Truth value of Q to validate
    
    Returns:
    - True if inference is valid, False otherwise
    """
    # Disjunctive Syllogism is valid if:
    # 1. P ∨ Q is True
    # 2. P is False (¬P is True)
    # 3. Then Q must be True
    if premise1_or and premise2_not_p:
        return conclusion_q == True
    return True

# Test Disjunctive Syllogism
print("=" * 60)
print("Testing Disjunctive Syllogism:")
print("=" * 60)

print("\nExample: Valid Disjunctive Syllogism")
print("Premise 1: Either sunny OR raining (P ∨ Q) = True")
print("Premise 2: NOT sunny (¬P) = True")
print("Conclusion: It is raining (Q) = True")
valid = disjunctive_syllogism(True, True, True)
print(f"Result: {'✅ VALID' if valid else '❌ INVALID'}")
````

````python
# FIGURE: what makes an inference rule VALID, and what a broken one looks like.
# Why a picture: the functions above check one case at a time, so the thing that
# actually certifies a rule - "in EVERY row where the premises hold, the conclusion
# holds too" - is a claim about the whole table and cannot be seen one case at a time.
# The fourth panel is the classic fallacy, tested by the same code, so its failure is
# measured rather than asserted.
import itertools
import matplotlib.pyplot as plt

# Each entry: name, premise labels, conclusion label, premise values, conclusion value.
RULES = [
    ("Modus Ponens",              ["p → q", "p"],   "∴ q",
     lambda p, q: (implies(p, q), p),                lambda p, q: q),
    ("Modus Tollens",             ["p → q", "¬q"],  "∴ ¬p",
     lambda p, q: (implies(p, q), logical_not(q)),   lambda p, q: logical_not(p)),
    ("Disjunctive Syllogism",     ["p ∨ q", "¬p"],  "∴ q",
     lambda p, q: (logical_or(p, q), logical_not(p)), lambda p, q: q),
    ("Affirming the Consequent",  ["p → q", "q"],   "∴ p",
     lambda p, q: (implies(p, q), q),                lambda p, q: p),
]

rows = list(itertools.product([False, True], repeat=2))   # (p, q): FF, FT, TF, TT
TRUE_C, FALSE_C = "#1b7837", "#e8e8e8"

fig, axes = plt.subplots(2, 2, figsize=(11.4, 7.0))
verdicts = []
for ax, (name, prem_labels, concl_label, prem_fn, concl_fn) in zip(axes.ravel(), RULES):
    cols = prem_labels + [concl_label]
    live, counter = [], []           # rows where both premises hold / of those, the bad ones
    for i, (p, q) in enumerate(rows):
        vals = list(prem_fn(p, q)) + [concl_fn(p, q)]
        premises_hold = all(vals[:-1])
        if premises_hold:
            (counter if not vals[-1] else live).append((p, q))
        for j, v in enumerate(vals):
            # Rows that cannot test the rule (a premise is false) are faded out:
            # an inference rule promises nothing about them.
            alpha = 1.0 if premises_hold else 0.28
            ax.add_patch(plt.Rectangle((j, i), 1, 1, facecolor=TRUE_C if v else FALSE_C,
                                       edgecolor="white", linewidth=2, alpha=alpha))
            ax.text(j + 0.5, i + 0.5, "T" if v else "F", ha="center", va="center",
                    fontsize=12, fontweight="bold",
                    color="white" if v else "#555555", alpha=alpha)
        if premises_hold:
            ok = vals[-1]
            # Black box = this row tests the rule; red box = it BREAKS the rule.
            # (Green-on-green would be invisible against the True cells.)
            ax.add_patch(plt.Rectangle((0.04, i + 0.04), len(cols) - 0.08, 0.92, fill=False,
                                       edgecolor="#111111" if ok else "#c0392b",
                                       linewidth=2.5 if ok else 3.5))

    valid = not counter
    verdicts.append((name, len(live) + len(counter), len(counter), valid))
    note = (f"VALID — premises hold in {len(live)} row; the conclusion holds there too"
            if valid else
            "INVALID — premises hold in "
            f"{len(live) + len(counter)} rows, and in "
            + ", ".join(f"p={'T' if p else 'F'}, q={'T' if q else 'F'}" for p, q in counter)
            + " the conclusion is FALSE")
    ax.text(len(cols) / 2, -0.55, note, ha="center", va="center", fontsize=8.5,
            color="#1b7837" if valid else "#c0392b", fontweight="bold")

    ax.set_xlim(0, len(cols)); ax.set_ylim(-0.95, len(rows))
    ax.set_xticks([j + 0.5 for j in range(len(cols))])
    ax.set_xticklabels(cols, fontsize=11)
    ax.xaxis.set_ticks_position("top")
    ax.set_yticks([i + 0.5 for i in range(len(rows))])
    ax.set_yticklabels([f"p={'T' if p else 'F'}  q={'T' if q else 'F'}" for p, q in rows],
                       fontsize=9)
    ax.invert_yaxis()
    for s in ax.spines.values():
        s.set_visible(False)
    ax.tick_params(length=0)
    ax.set_title(name, fontsize=12, fontweight="bold", pad=26)

fig.suptitle("A rule is valid when every row that satisfies its premises also satisfies its conclusion\n"
             "faded rows cannot test the rule (a premise is False there); boxed rows are the test",
             fontsize=11.5, y=1.005)
plt.tight_layout()
plt.show()

print(f"{'rule':<28}{'rows testing it':>16}{'counterexamples':>18}{'verdict':>10}")
print("-" * 72)
for name, tested, bad, valid in verdicts:
    print(f"{name:<28}{tested:>16}{bad:>18}{'VALID' if valid else 'INVALID':>10}")
````

**Read the four panels.** Each is the same four rows of `p` and `q`; the columns are the rule's two premises and then its conclusion. A row is **faded** when at least one premise is False — an inference rule makes no promise about those rows, so they cannot test it. The rows that survive are **boxed in black**, and those are the entire test.

- **Modus Ponens** has exactly one boxed row, `p=T, q=T`, and its conclusion cell is green. One row, no counterexample: valid.
- **Modus Tollens** boxes `p=F, q=F` — the only row where `p → q` holds *and* `¬q` holds — and the conclusion `¬p` is green there. Valid.
- **Disjunctive Syllogism** boxes `p=F, q=T` and its conclusion `q` is green. Valid.
- **Affirming the Consequent** — the fourth panel — is where the picture earns its place. It boxes **two** rows, because `p → q` and `q` can both hold in two different worlds. In `p=T, q=T` the conclusion `p` is green. In **`p=F, q=T` the conclusion cell is grey**, and the box around it is red. The premises are satisfied and the conclusion is false. One such row is all it takes: the pattern is not a rule at all.

That red box is the answer to Discuss question 1. `modus_ponens(True, True, True)` tells you this *one* case is consistent with the rule; only checking every row tells you the rule itself never fails. The printed table repeats the same verdict as counts, computed by the same loop that drew the picture.

## Part 4: Building and Validating Logical Arguments

Now let's create a function that can validate complete logical arguments using multiple inference rules.


````python
# Chain the three rules into multi-step arguments and validate the WHOLE chain.
# Why: real reasoning is many small steps - one bad step (see Demo 2) invalidates the entire argument.

class LogicalArgument:
    """Represent a multi-step logical argument and validate it with inference rules.

    An argument is a sequence of steps. Each step names an inference rule,
    gives the truth values of its two premises, and proposes a conclusion.
    The argument is VALID only if EVERY step is valid.
    """

    # The three rule validators defined earlier in this notebook
    RULES = {
        "modus_ponens": modus_ponens,
        "modus_tollens": modus_tollens,
        "disjunctive_syllogism": disjunctive_syllogism,
    }

    def __init__(self, description):
        self.description = description
        self.steps = []  # list of (rule_name, premise1, premise2, conclusion, note)

    def add_step(self, rule_name, premise1, premise2, conclusion, note=""):
        """Add one inference step to the argument."""
        if rule_name not in self.RULES:
            raise ValueError(f"Unknown rule: {rule_name}. Choose from {list(self.RULES)}")
        self.steps.append((rule_name, premise1, premise2, conclusion, note))
        return self  # allow chaining

    def validate(self, verbose=True):
        """Check every step with its inference rule. Returns True if all steps are valid."""
        if verbose:
            print(f"Argument: {self.description}")
        all_valid = True
        for i, (rule_name, p1, p2, concl, note) in enumerate(self.steps, 1):
            step_valid = self.RULES[rule_name](p1, p2, concl)
            all_valid = all_valid and step_valid
            if verbose:
                status = "✅ valid" if step_valid else "❌ INVALID"
                label = f" — {note}" if note else ""
                print(f"  Step {i} [{rule_name}]{label}: {status}")
        if verbose:
            verdict = "✅ ARGUMENT IS VALID" if all_valid else "❌ ARGUMENT IS INVALID"
            print(f"  => {verdict}")
        return all_valid


# ---------------------------------------------------------------
# Demo 1: a VALID three-step argument (uses all three rules)
# ---------------------------------------------------------------
print("=" * 60)
print("Validating Complete Logical Arguments:")
print("=" * 60)
print()

argument1 = LogicalArgument("Why I take the train on a rainy match day")
argument1.add_step(
    "modus_ponens", True, True, True,
    note="If it rains (P) the streets flood (Q); it rains; so streets flood"
).add_step(
    "modus_tollens", True, True, True,
    note="If I cycle (R) I stay dry (S); I am NOT dry; so I did NOT cycle"
).add_step(
    "disjunctive_syllogism", True, True, True,
    note="I take the bus (T) or the train (U); NOT the bus; so the train"
)
result1 = argument1.validate()

print()

# ---------------------------------------------------------------
# Demo 2: an INVALID argument (one step contradicts its rule)
# ---------------------------------------------------------------
argument2 = LogicalArgument("A broken argument (step 2 contradicts Modus Ponens)")
argument2.add_step(
    "modus_ponens", True, True, True,
    note="If P then Q; P; so Q — fine"
).add_step(
    "modus_ponens", True, True, False,
    note="If Q then R; Q; but concluding NOT R — contradiction!"
)
result2 = argument2.validate()

print()
print(f"Summary: argument1 valid = {result1}, argument2 valid = {result2}")
````

````python
# FIGURE: an argument is a CHAIN, and a chain is only as good as its worst link.
# Why a picture: the printout above reports the two arguments as two blocks of text,
# so the structural point - the same first step, then one changed conclusion, then a
# different verdict for the whole thing - has to be reconstructed by reading carefully.
import textwrap
import matplotlib.pyplot as plt

def step_status(step):
    """Re-validate one step with the very same rule table the class uses."""
    rule_name, p1, p2, concl, note = step
    return LogicalArgument.RULES[rule_name](p1, p2, concl)

arguments = [argument1, argument2]
n_max = max(len(a.steps) for a in arguments)

fig, axes = plt.subplots(len(arguments), 1, figsize=(12.6, 5.8))
for ax, arg in zip(axes, arguments):
    statuses = [step_status(s) for s in arg.steps]
    for i, (step, ok) in enumerate(zip(arg.steps, statuses)):
        rule_name, p1, p2, concl, note = step
        ax.add_patch(plt.Rectangle((i + 0.11, 0.19), 0.78, 0.62,
                                   facecolor="#dff0d8" if ok else "#f7d6d2",
                                   edgecolor="#1b7837" if ok else "#c0392b",
                                   linewidth=2.5))
        ax.text(i + 0.5, 0.71, f"Step {i + 1}: {rule_name}", ha="center", va="center",
                fontsize=9.5, fontweight="bold", color="#1f3d1a" if ok else "#7b1c12")
        ax.text(i + 0.5, 0.50, textwrap.fill(note, 34), ha="center", va="center",
                fontsize=7.5, color="#333333")
        ax.text(i + 0.5, 0.27, "valid" if ok else "INVALID", ha="center", va="center",
                fontsize=10, fontweight="bold", color="#1b7837" if ok else "#c0392b")
        # arrow into the next box (or into the verdict box for the last step)
        ax.annotate("", xy=(i + 1.11, 0.5), xytext=(i + 0.89, 0.5),
                    arrowprops=dict(arrowstyle="-|>", color="#666666", linewidth=2))

    all_valid = all(statuses)
    k = len(arg.steps)
    ax.add_patch(plt.Rectangle((k + 0.11, 0.19), 0.78, 0.62,
                               facecolor="#1b7837" if all_valid else "#c0392b",
                               edgecolor="#333333", linewidth=2))
    ax.text(k + 0.5, 0.5, "ARGUMENT\nVALID" if all_valid else "ARGUMENT\nINVALID",
            ha="center", va="center", fontsize=12, fontweight="bold", color="white")

    ax.set_xlim(0, n_max + 1.05); ax.set_ylim(0, 1)
    ax.axis("off")
    ax.set_title(arg.description, fontsize=11, fontweight="bold", loc="left")

fig.suptitle("Validity does not average: every link must hold", fontsize=12.5, y=0.99)
plt.tight_layout()
plt.show()

for arg in arguments:
    statuses = [step_status(s) for s in arg.steps]
    print(f"{arg.description}: {sum(statuses)}/{len(statuses)} steps valid "
          f"-> argument {'VALID' if all(statuses) else 'INVALID'}")
````

**Read the two rows.** Each box is one inference step, re-validated by the same `LogicalArgument.RULES` table the class uses, and the block at the right of each row is the verdict for the argument as a whole.

- **Top row** — the rainy-match-day argument: three green boxes, one per rule, then a green **ARGUMENT VALID**. Three different inference rules chained into one conclusion, every link holding.
- **Bottom row** — the broken argument: step 1 is *green and still valid*, exactly as before. Only step 2 changed, and only in its conclusion. It is red, and the verdict block flipped to **ARGUMENT INVALID**.

The two rows differ in one box, and yet the whole right-hand verdict differs. That is what "valid" means for a chain: not a score, not a majority — a conjunction. Two-thirds of a proof proves nothing, which is why the summary line under the figure prints `3/3 steps valid -> argument VALID` for the first and `1/2 steps valid -> argument INVALID` for the second.

### Read Part 4 as an experiment, not as two demos

The cell above is one controlled experiment with a single deliberate change.

- **Baseline** — *"Why I take the train on a rainy match day"*: three steps, one per rule (modus ponens, modus tollens, disjunctive syllogism). Every step prints ✅ valid, and the argument as a whole prints ✅ ARGUMENT IS VALID.
- **Change exactly one thing** — *"A broken argument"* keeps the rule (modus ponens) and keeps the premises true, and changes only the **conclusion of step 2** from `True` to `False`.
- **Re-run and read the output** — step 1 is still ✅ valid; step 2 is ❌ INVALID; the whole argument is ❌ INVALID. The final summary line prints `argument1 valid = True, argument2 valid = False`.
- **The conclusion those two runs support** — a single bad step poisons the entire chain, and it does so *invisibly from the outside*. Argument 2's conclusion is stated with exactly the same confidence as Argument 1's. Nothing in the conclusion itself carries the flaw; only checking every step does.

That is the whole professional case for mechanical proof checking, and it is why seL4's proof is longer than seL4.

## ⚠️ Where this breaks

These rules are exactly as strong as their guarantee, and no stronger.

- **Validity is not truth.** Every function above answers "does the conclusion follow?", never "is the conclusion true?". Feed it *"if the moon is cheese then Riyadh floods; the moon is cheese"* and it certifies `Riyadh floods` as a valid inference — correctly. Garbage premises, valid reasoning, false conclusion. In deployment, most wrong answers from a rule system come from wrong premises, not from broken logic.
- **Adding knowledge can never remove a conclusion.** Classical inference is *monotonic*. Once you have derived something, no new fact can retract it. Human reasoning is not like that: you conclude the flight is on time, then learn there is fog. Handling that needs non-monotonic or default logic, which is a different and harder machinery.
- **The rules are safe; a rule *base* written by users is not.** Nothing stops someone encoding affirming-the-consequent as a "rule" in an expert system, and once it is in the base, every conclusion downstream of it inherits the flaw — while still being reported with full confidence. Rule review is a code review, and it needs the same discipline.
- **Certainty is the wrong model for evidence.** These rules take premises that are *true* or *false*. A test result, a witness statement, a sensor reading and a fraud signal are none of those — they are evidence with a strength. Forcing them into true/false is where reasoning about statistics goes wrong. Unit 3 replaces "P is true" with "P has probability 0.83", and modus ponens becomes Bayes' theorem.
- **The cheaper alternative.** For an argument with a handful of propositions, the truth-table builder from notebook 02 settles validity exhaustively with no rule names to remember. Inference rules earn their keep only when the formula has too many variables to enumerate — which is exactly the case where you can no longer check the answer by eye.

## Summary

### Key Inference Rules Learned:

1. **Modus Ponens**
   - If P → Q and P, then Q
   - Most common inference rule

2. **Modus Tollens**
   - If P → Q and ¬Q, then ¬P
   - Contrapositive reasoning

3. **Disjunctive Syllogism**
   - If P ∨ Q and ¬P, then Q
   - Eliminates one option
