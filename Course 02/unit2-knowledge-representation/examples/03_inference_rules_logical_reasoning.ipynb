## Introduction to Inference Rules

**Inference rules** are logical patterns that allow us to derive valid conclusions from given premises. They are the foundation of logical reasoning in AI systems.

## Part 1: Modus Ponens

**Modus Ponens** (Latin: "method of affirming") is one of the most important inference rules:

**Pattern:** If P → Q, and P is true, then Q must be true.

**Formal Representation:**
```
P → Q
P
---
∴ Q
```

**Example:**
- If it rains, then the ground gets wet. (P → Q)
- It is raining. (P)
- Therefore, the ground is wet. (Q)


````python
def modus_ponens(premise1_implies, premise2_p, conclusion_q):
    """
    Validate Modus Ponens inference rule.
    
    Parameters:
    - premise1_implies: Truth value of P → Q
    - premise2_p: Truth value of P
    - conclusion_q: Truth value of Q (to validate)
    
    Returns:
    - True if inference is valid, False otherwise
    """
    # Modus Ponens is valid if:
    # 1. P → Q is True
    # 2. P is True
    # 3. Then Q must be True
    if premise1_implies and premise2_p:
        return conclusion_q == True
    # If P → Q is False or P is False, Modus Ponens doesn't apply
    return True  # No contradiction if premises aren't met

# Test Modus Ponens
print("=" * 60)
print("Testing Modus Ponens:")
print("=" * 60)

# Example 1: Valid Modus Ponens
print("\nExample 1: Valid Modus Ponens")
print("Premise 1: If it rains, then ground is wet (P → Q) = True")
print("Premise 2: It is raining (P) = True")
print("Conclusion: Ground is wet (Q) = True")
valid = modus_ponens(True, True, True)
print(f"Result: {'✅ VALID' if valid else '❌ INVALID'}")

# Example 2: Invalid (contradiction)
print("\nExample 2: Invalid Modus Ponens (contradiction)")
print("Premise 1: If it rains, then ground is wet (P → Q) = True")
print("Premise 2: It is raining (P) = True")
print("Conclusion: Ground is NOT wet (Q) = False")
valid = modus_ponens(True, True, False)
print(f"Result: {'✅ VALID' if valid else '❌ INVALID (CONTRADICTION)'}")
````

## Part 2: Modus Tollens

**Modus Tollens** (Latin: "method of denying") reasons *backwards* from a failed consequence:

**Pattern:** If P → Q, and Q is false, then P must be false.

**Formal Representation:**
```
P → Q
¬Q
---
∴ ¬P
```

**Example:**
- If it rains, then the ground gets wet. (P → Q)
- The ground is NOT wet. (¬Q)
- Therefore, it is NOT raining. (¬P)

This is *contrapositive* reasoning: P → Q is logically equivalent to ¬Q → ¬P (you verified this equivalence with truth tables in the previous notebook).

````python
def modus_tollens(premise1_implies, premise2_not_q, conclusion_not_p):
    """
    Validate Modus Tollens inference rule.
    
    Parameters:
    - premise1_implies: Truth value of P → Q
    - premise2_not_q: Truth value of ¬Q (NOT Q)
    - conclusion_not_p: Truth value of ¬P (NOT P) to validate
    
    Returns:
    - True if inference is valid, False otherwise
    """
    # Modus Tollens is valid if:
    # 1. P → Q is True
    # 2. Q is False (¬Q is True)
    # 3. Then P must be False (¬P must be True)
    if premise1_implies and premise2_not_q:
        return conclusion_not_p == True
    return True

# Test Modus Tollens
print("=" * 60)
print("Testing Modus Tollens:")
print("=" * 60)

print("\nExample: Valid Modus Tollens")
print("Premise 1: If it rains, then ground is wet (P → Q) = True")
print("Premise 2: Ground is NOT wet (¬Q) = True")
print("Conclusion: It is NOT raining (¬P) = True")
valid = modus_tollens(True, True, True)
print(f"Result: {'✅ VALID' if valid else '❌ INVALID'}")
````

## Part 3: Disjunctive Syllogism

**Disjunctive Syllogism** deals with OR statements:

**Pattern:** If P OR Q, and P is false, then Q must be true.

**Formal Representation:**
```
P ∨ Q
¬P
---
∴ Q
```

**Example:**
- Either it is sunny OR it is raining. (P ∨ Q)
- It is NOT sunny. (¬P)
- Therefore, it is raining. (Q)


````python
def disjunctive_syllogism(premise1_or, premise2_not_p, conclusion_q):
    """
    Validate Disjunctive Syllogism inference rule.
    
    Parameters:
    - premise1_or: Truth value of P ∨ Q
    - premise2_not_p: Truth value of ¬P
    - conclusion_q: Truth value of Q to validate
    
    Returns:
    - True if inference is valid, False otherwise
    """
    # Disjunctive Syllogism is valid if:
    # 1. P ∨ Q is True
    # 2. P is False (¬P is True)
    # 3. Then Q must be True
    if premise1_or and premise2_not_p:
        return conclusion_q == True
    return True

# Test Disjunctive Syllogism
print("=" * 60)
print("Testing Disjunctive Syllogism:")
print("=" * 60)

print("\nExample: Valid Disjunctive Syllogism")
print("Premise 1: Either sunny OR raining (P ∨ Q) = True")
print("Premise 2: NOT sunny (¬P) = True")
print("Conclusion: It is raining (Q) = True")
valid = disjunctive_syllogism(True, True, True)
print(f"Result: {'✅ VALID' if valid else '❌ INVALID'}")
````

## Part 4: Building and Validating Logical Arguments

Now let's create a function that can validate complete logical arguments using multiple inference rules.


````python
class LogicalArgument:
    """Represent a multi-step logical argument and validate it with inference rules.

    An argument is a sequence of steps. Each step names an inference rule,
    gives the truth values of its two premises, and proposes a conclusion.
    The argument is VALID only if EVERY step is valid.
    """

    # The three rule validators defined earlier in this notebook
    RULES = {
        "modus_ponens": modus_ponens,
        "modus_tollens": modus_tollens,
        "disjunctive_syllogism": disjunctive_syllogism,
    }

    def __init__(self, description):
        self.description = description
        self.steps = []  # list of (rule_name, premise1, premise2, conclusion, note)

    def add_step(self, rule_name, premise1, premise2, conclusion, note=""):
        """Add one inference step to the argument."""
        if rule_name not in self.RULES:
            raise ValueError(f"Unknown rule: {rule_name}. Choose from {list(self.RULES)}")
        self.steps.append((rule_name, premise1, premise2, conclusion, note))
        return self  # allow chaining

    def validate(self, verbose=True):
        """Check every step with its inference rule. Returns True if all steps are valid."""
        if verbose:
            print(f"Argument: {self.description}")
        all_valid = True
        for i, (rule_name, p1, p2, concl, note) in enumerate(self.steps, 1):
            step_valid = self.RULES[rule_name](p1, p2, concl)
            all_valid = all_valid and step_valid
            if verbose:
                status = "✅ valid" if step_valid else "❌ INVALID"
                label = f" — {note}" if note else ""
                print(f"  Step {i} [{rule_name}]{label}: {status}")
        if verbose:
            verdict = "✅ ARGUMENT IS VALID" if all_valid else "❌ ARGUMENT IS INVALID"
            print(f"  => {verdict}")
        return all_valid


# ---------------------------------------------------------------
# Demo 1: a VALID three-step argument (uses all three rules)
# ---------------------------------------------------------------
print("=" * 60)
print("Validating Complete Logical Arguments:")
print("=" * 60)
print()

argument1 = LogicalArgument("Why I take the train on a rainy match day")
argument1.add_step(
    "modus_ponens", True, True, True,
    note="If it rains (P) the streets flood (Q); it rains; so streets flood"
).add_step(
    "modus_tollens", True, True, True,
    note="If I cycle (R) I stay dry (S); I am NOT dry; so I did NOT cycle"
).add_step(
    "disjunctive_syllogism", True, True, True,
    note="I take the bus (T) or the train (U); NOT the bus; so the train"
)
result1 = argument1.validate()

print()

# ---------------------------------------------------------------
# Demo 2: an INVALID argument (one step contradicts its rule)
# ---------------------------------------------------------------
argument2 = LogicalArgument("A broken argument (step 2 contradicts Modus Ponens)")
argument2.add_step(
    "modus_ponens", True, True, True,
    note="If P then Q; P; so Q — fine"
).add_step(
    "modus_ponens", True, True, False,
    note="If Q then R; Q; but concluding NOT R — contradiction!"
)
result2 = argument2.validate()

print()
print(f"Summary: argument1 valid = {result1}, argument2 valid = {result2}")
````
