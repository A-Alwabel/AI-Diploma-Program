## 🌍 Real-World Worked Example — Inventory Management with Q-Learning

**Industry context:**
- Amazon uses RL to optimise warehouse robot routing
- Google DeepMind reduced data centre cooling costs by 40% using RL
- Uber uses RL for dynamic surge pricing

We model a simple **inventory replenishment** problem as an MDP and solve it with Q-learning.

```python
import numpy as np, matplotlib.pyplot as plt

np.random.seed(42)

# ── Environment: Inventory Management ─────────────────────────────────────
# State: stock level 0..10. Action: 0=don't restock, 1=restock (+5 units).
# Each day: sell 0-3 units (random). Reward: +1 per unit sold, -2 if out of stock.
MAX_STOCK = 10
n_states, n_actions = MAX_STOCK+1, 2

def step(state, action):
    new_s = min(state + action*5, MAX_STOCK)
    demand = np.random.randint(0,4)
    sold   = min(new_s, demand)
    reward = sold - (2 if demand > new_s else 0) - (0.5 * action)  # restocking has cost
    new_s  = max(new_s - demand, 0)
    return new_s, reward

# ── Q-Learning ─────────────────────────────────────────────────────────────
Q      = np.zeros((n_states, n_actions))
alpha  = 0.1; gamma = 0.95
eps    = 1.0; eps_min = 0.05; eps_decay = 0.995
rewards_history = []

for episode in range(3000):
    state = np.random.randint(0, MAX_STOCK+1)
    total_r = 0
    for _ in range(30):  # 30 days per episode
        action = np.random.randint(n_actions) if np.random.rand()<eps else Q[state].argmax()
        next_state, reward = step(state, action)
        Q[state,action] += alpha*(reward + gamma*Q[next_state].max() - Q[state,action])
        state = next_state; total_r += reward
    rewards_history.append(total_r)
    eps = max(eps*eps_decay, eps_min)

# ── Results ──────────────────────────────────────────────────────────────
print("Optimal policy (restock when stock is low):")
for s in range(11):
    action = ['Do NOT restock', 'RESTOCK'][Q[s].argmax()]
    print(f"  Stock={s:2d} → {action}")

# Moving average reward
ma = np.convolve(rewards_history, np.ones(100)/100, mode='valid')
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.plot(ma); plt.title("Q-Learning: Avg Reward over Training"); plt.xlabel("Episode")
plt.subplot(1,2,2)
plt.imshow(Q, aspect='auto', cmap='RdYlGn')
plt.colorbar(label="Q-value"); plt.xlabel("Action (0=hold, 1=restock)")
plt.ylabel("Stock level"); plt.title("Learned Q-table")
plt.tight_layout(); plt.show()
```
