## 🎯 The reflex agent that sold tens of millions of units

iRobot launched the **Roomba** in **September 2002**. For its first thirteen years
it had no map of your home and no idea where it had been. It used a
sensor-driven pseudorandom coverage strategy: drive, bump into something, turn a
semi-random amount, drive again — and rely on statistics and time to cover the
floor, passing over the average patch several times from several angles. That is a
**simple reflex agent**, and it is the design you are about to build below.

It was only with the **Roomba 980**, announced on **16 September 2015**, that
iRobot put its visual SLAM (simultaneous localisation and mapping) into a consumer
product: a camera-based system that builds a map as it cleans and tracks where it
is, so it can cover a floor once, systematically, and resume after recharging.

Thirteen years is the interesting number. A reflex agent shipped, sold, and did
the job well enough for over a decade, because the *performance measure* was
"floor is clean", not "floor was cleaned efficiently". The upgrade to a
goal-based, map-building agent happened when customers began to measure the second
thing. **Which agent type you need is decided by the performance measure, not by
which one is more sophisticated** — and that is why PEAS starts with P.

### What goes wrong without PEAS

Skip the specification and you will build a vacuum that maximises something nobody
asked for. If "performance" means dirt collected, an agent can learn to spread
dirt so there is more to collect. If it means area covered, it will cover the same
easy corridor forever. Writing P, E, A and S down before you code forces the
argument about what "good" means to happen while it is still cheap.

```python
# Case study: describing a robot vacuum with the PEAS framework (Performance measure,
# Environment, Actuators, Sensors). PEAS is how you specify ANY agent problem before coding it.
# Case Study 1: Autonomous Vacuum Cleaner Agent
# PEAS Framework Analysis

print("=== PEAS Framework Analysis ===")
print("\nPerformance Measure:")
print("- Cleanliness of floors")
print("- Time to complete cleaning")
print("- Energy efficiency")
print("- Battery life")

print("\nEnvironment:")
print("- Type: Discrete, Partially Observable, Sequential")
print("- Characteristics: Grid-based floor layout, dynamic (people, pets)")
print("- Agents: Single agent (vacuum cleaner)")

print("\nActuators:")
print("- Movement motors (forward, backward, turn)")
print("- Vacuum motor (on/off)")
print("- Brush mechanism")

print("\nSensors:")
print("- Dirt sensors")
print("- Bump sensors (collision detection)")
print("- Battery level sensor")
print("- Camera/LIDAR for mapping (in advanced models)")
```

```python
# From PEAS description to working code: a simple REFLEX agent — it maps the current
# percept straight to an action with no memory or planning, the simplest agent design.
# Simple Reflex Vacuum Agent Implementation
class SimpleVacuumAgent:
    """Simple reflex-based vacuum cleaner agent"""
    def __init__(self):
        self.location = (0, 0)
        self.battery = 100
        self.cleaned = set()
    
    def sense(self, environment):
        """Sense the environment"""
        x, y = self.location
        return {
            'dirt': environment.get((x, y), False),
            'battery': self.battery,
            'location': (x, y)
        }
    
    def act(self, perception, environment):
        """Simple reflex: if dirty, clean; else move"""
        if perception['battery'] <= 10:
            return 'return_to_charger'
        elif perception['dirt']:
            environment[self.location] = False   # cleaning removes the dirt
            self.cleaned.add(self.location)
            self.battery -= 1
            return 'clean'
        else:
            x, y = self.location
            self.location = (x + 1, y)           # simple sweep: move one tile right
            self.battery -= 1
            return 'move_forward'

# Example usage
# Simulate six steps in a tiny 1-row world and watch the percept -> action pairs.
agent = SimpleVacuumAgent()
environment = {(0, 0): True, (1, 0): False, (2, 0): True}

print("=== Simple Vacuum Agent Decision-Making ===")
for step in range(6):
    perception = agent.sense(environment)
    action = agent.act(perception, environment)
    print(f"Step {step + 1}: Location {perception['location']}, Dirt: {perception['dirt']}, Action: {action}")

print(f"\nTiles cleaned: {sorted(agent.cleaned)}  |  Battery left: {agent.battery}%")
```

```python
# ── Storyboard: the six-step vacuum run, drawn tile by tile ───────────────────
# WHAT: replay the EXACT six steps printed above with a fresh agent and a fresh room,
#       recording the room state at every percept, then draw one panel per step + a GIF.
# WHY:  the trace above ends "Step 6: Location (3, 0)". Nothing in that line tells a
#       student that tile (3,0) is OUTSIDE the room. Drawn, the mistake is unmissable.
import io

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Rectangle

# Replay from scratch so the storyboard does not depend on the mutated objects above.
replay_agent = SimpleVacuumAgent()
replay_room = {(0, 0): True, (1, 0): False, (2, 0): True}   # the same tiny 3-tile world
ROOM_TILES = [0, 1, 2]                                       # tiles that actually exist

frames_state = []
for step in range(6):
    perception = replay_agent.sense(replay_room)
    dirt_now = {x: replay_room.get((x, 0), False) for x in ROOM_TILES}   # room BEFORE acting
    action = replay_agent.act(perception, replay_room)
    frames_state.append({'step': step + 1, 'x': perception['location'][0],
                         'dirt_here': perception['dirt'], 'dirt': dirt_now,
                         'action': action, 'battery': replay_agent.battery})

# Honesty check: this replay must be the run the cell above printed.
assert [f['action'] for f in frames_state] == ['clean', 'move_forward', 'move_forward',
                                               'clean', 'move_forward', 'move_forward']
assert [f['x'] for f in frames_state] == [0, 0, 1, 2, 2, 3]
assert replay_agent.battery == 94 and sorted(replay_agent.cleaned) == [(0, 0), (2, 0)]

TILES = [0, 1, 2, 3]        # we draw one tile beyond the room, because the agent goes there


def draw_step(ax, f):
    """Draw the one-row room: dirty tiles brown, clean tiles pale, outside tiles hatched."""
    for x in TILES:
        inside = x in ROOM_TILES
        if not inside:
            colour, hatch = '#f5f5f5', '///'          # this tile is not part of the world
        elif f['dirt'][x]:
            colour, hatch = '#8d6e63', None           # dirty
        else:
            colour, hatch = '#e8f5e9', None           # already clean
        ax.add_patch(Rectangle((x, 0), 1, 1, facecolor=colour, hatch=hatch,
                               edgecolor='#9e9e9e' if inside else '#e57373',
                               linestyle='-' if inside else '--', linewidth=1.6))
        ax.text(x + 0.5, 0.08, f"({x},0)", ha='center', fontsize=6.5, color='#616161')
    ax.text(3.5, 1.16, 'outside the room', ha='center', fontsize=6.5, color='#e57373',
            bbox=dict(boxstyle='round,pad=0.15', facecolor='white', edgecolor='none'))
    # the agent itself
    ax.plot([f['x'] + 0.5], [0.55], marker='o', markersize=17,
            markerfacecolor='#1e88e5', markeredgecolor='#0d47a1')
    ax.text(f['x'] + 0.5, 0.55, 'R', ha='center', va='center', color='white',
            fontsize=8, fontweight='bold')
    if f['action'] == 'clean':                        # show the action about to happen
        ax.annotate('clean!', (f['x'] + 0.5, 0.55), xytext=(f['x'] + 0.5, 1.12),
                    ha='center', fontsize=8, color='#2e7d32',
                    arrowprops=dict(arrowstyle='->', color='#2e7d32'))
    else:
        ax.annotate('', (f['x'] + 1.35, 0.55), xytext=(f['x'] + 0.6, 0.55),
                    arrowprops=dict(arrowstyle='->', color='#1e88e5', lw=1.8))
    ax.set_xlim(-0.15, 4.15)
    ax.set_ylim(-0.05, 1.45)
    ax.set_axis_off()
    ax.set_title(f"step {f['step']}: at ({f['x']},0), dirt={f['dirt_here']}\n"
                 f"{f['action']}  |  battery after: {f['battery']}%", fontsize=8)


fig, axes = plt.subplots(3, 2, figsize=(9.0, 5.8))
for ax, f in zip(axes.ravel(), frames_state):
    draw_step(ax, f)
fig.suptitle('Simple reflex vacuum — six steps in a three-tile room\n'
             'each panel shows the room AS THE AGENT FINDS IT, before the action in its title\n'
             '(brown = dirty, pale green = clean, red dashed + hatched = outside the room)',
             fontsize=10.5)
fig.tight_layout(rect=[0, 0, 1, 0.88])
plt.show()

# The same storyboard as a table you can read out loud in class.
print(f"{'step':>4}  {'tile':^6} {'dirt here':^10} {'action':<16} {'battery after':>13}  room as the agent finds it")
for f in frames_state:
    room_txt = ' '.join('DIRTY' if f['dirt'][x] else 'clean' for x in ROOM_TILES)
    print(f"{f['step']:>4}  {'(' + str(f['x']) + ',0)':^6} {str(f['dirt_here']):^10} "
          f"{f['action']:<16} {f['battery']:>12}%  {room_txt}")
print(f"\nRoom finished after step 4. Steps 5 and 6 spent battery going nowhere,")
print(f"and step 6 put the agent on (3,0) — a tile that is not in the room at all.")

# Animated version of the very same frames (the repo's Course 09 step-viz idiom:
# render frames -> encode a GIF -> display it). The grid above is the fallback.
try:
    from IPython.display import Image as IPyImage, display
    from PIL import Image

    gif_frames = []
    for f in frames_state:
        f2, ax2 = plt.subplots(figsize=(4.4, 1.9), dpi=100)
        draw_step(ax2, f)
        f2.tight_layout()
        f2.canvas.draw()
        gif_frames.append(Image.fromarray(np.asarray(f2.canvas.buffer_rgba())[:, :, :3]))
        plt.close(f2)
    buf = io.BytesIO()
    gif_frames[0].save(buf, format='GIF', save_all=True, append_images=gif_frames[1:],
                       duration=1000, loop=0)
    display(IPyImage(data=buf.getvalue()))
    print(f"\nAnimated GIF above: {len(gif_frames)} frames, {len(buf.getvalue()) / 1024:.0f} KB")
except Exception as exc:                        # pragma: no cover - display convenience only
    print(f"\n(GIF skipped: {exc}) — the frame grid above carries the same information.")
```

### 👁️ Read the storyboard

Six panels, one per step of the trace printed above. The blue **R** is the agent,
brown tiles are dirty, pale green tiles are clean, and the hatched tile with the red
dashed border is **not part of the room** — it exists in the drawing only because the
agent goes there. Each panel shows the room *as the agent finds it*, before the action
named in the panel title.

**What the picture proves:**

- **The job is finished by step 4.** Panel 4 is the last one with any brown in it:
  the agent is standing on the dirt at `(2,0)` and is about to clean it. From panel 5
  on, the room is entirely pale green — nothing is left to do.
- **The agent keeps going anyway.** Steps 5 and 6 are two more `move_forward`
  actions, and panel 6 puts R on the hatched tile outside the room — with **94%
  battery** still in hand. A reflex agent has no memory and no goal, so it has no
  concept of "done" and no concept of "edge of the room".
- **Every panel is decided by one tile.** The agent's percept is the square it is
  standing on. In panel 1 it cannot see the dirt on `(2,0)` at all, which is why it
  sweeps blindly rightward instead of going straight to the mess.

That is the whole argument for goal-based and map-building agents — and the reason
iRobot eventually shipped SLAM in the Roomba 980. Read the panels once more as a
performance measure: "the floor is clean" is satisfied after step 4; "the floor was
cleaned efficiently" never is.

## The Four Classic Agent Types — Where Our Vacuum Fits

| Agent type | How it decides | Our vacuum? |
|---|---|---|
| **Simple reflex** | Condition → action rules on the current percept only | ✅ this is what we built: *if dirt → clean, else → move* |
| **Goal-based** | Chooses actions that lead toward an explicit goal state | A vacuum that plans a route to cover every tile |
| **Utility-based** | Maximizes a utility score (speed vs battery vs coverage) | A vacuum that weighs finishing fast against saving battery |
| **Learning** | Improves its policy from experience | A vacuum that learns which rooms get dirty fastest |

Notice the limitation the demo exposes: our reflex agent only sweeps right and never plans a route — exactly why goal-based agents were invented. You will meet learning agents properly in the Reinforcement Learning course (AIAT 123).

## ⚠️ Where this breaks

**The demo's own failure is the lesson.** Look at the trace: the agent cleaned
`(0,0)`, moved right, cleaned `(2,0)`, and then kept moving right to `(3,0)` —
outside its environment — with 94% battery still to burn. It did that because a
reflex agent has **no memory and no goal**. It cannot know it is finished, cannot
know it has left the room, and cannot plan a route back.

- **No memory means no coverage guarantee.** With one sweep direction and no map,
  any tile the agent starts to the right of is never cleaned. The real Roomba
  survived this only by relying on randomness plus a great deal of time.
- **No model of the environment means no recovery.** Battery ≤ 10 returns
  `return_to_charger` as a *string*. There is no route home, because the agent has
  never represented where home is.
- **Partial observability is the normal case.** The agent senses dirt only on the
  tile it occupies. It cannot see a spill one metre away, so its "environment" is
  a one-tile window that it mistakes for the world.
- **PEAS specifies, it does not verify.** Writing "Performance measure:
  cleanliness of floors" commits you to nothing measurable. Until you say
  *cleanliness measured how, over what area, within what time and energy budget*,
  two engineers can both claim to have met the spec.
- **The environment description here is a static list.** Real homes are
  **dynamic** — people, pets and new dirt arrive while the agent is running — so
  a plan computed once is stale by the time it executes. Dynamic environments are
  why learning agents (Course 09) exist.

**Cheaper alternative:** before adding SLAM, a map and a planner, check whether a
better *reflex rule* plus more running time meets the performance measure. iRobot
shipped exactly that trade for thirteen years, and the extra sensor only earned
its cost when the measure changed.

### 🔁 The same reasoning shows up here too

- **Thermostats and lighting controllers:** reflex rules that were good enough
  until "comfort" was replaced by "kWh per comfortable hour" as the measure.
- **Warehouse robots:** reflex obstacle avoidance underneath a goal-based route
  planner — the same two layers, with the reflex layer as the safety net.
- **Email spam filtering:** a reflex rule ("contains this phrase → junk") that was
  replaced by a learning agent only once spammers made the environment dynamic.


## 🔭 State of the field (2026)

Everything you just did to the Roomba — PEAS, agent type, environment properties — is still the first
move a practitioner makes in 2026, but the systems being analysed have changed. The case studies that
dominate now are software agents: a language model in a loop with tools, a file system or a browser,
where the live design questions are about the **harness** (tool surface, memory, recovery path) rather
than about choosing between reflex and goal-based (Guo et al., 2026). The second 2026 lesson is one
this notebook is already built to teach: a single successful trace proves very little, because the
same agent run again on the same task often fails. Current benchmark work therefore runs stateful
workflows repeatedly and reports the spread, not the best run (Li et al., 2026). Read our six-step
vacuum trace in that light — it is one run, of one policy, in one tiny world, and the claims it
supports are correspondingly narrow.

PEAS stays on the syllabus because it is what turns a vague product idea into a specification people
can argue about before any code exists — and because the official plan examines exactly this analysis.
