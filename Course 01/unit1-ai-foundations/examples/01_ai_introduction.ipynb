```python
# Guided tour of what 'artificial intelligence' means: definitions, capability tiers,
# a tiny rule-based decision maker, and real-world applications.
# This vocabulary is reused in every later notebook of the diploma, so read the output carefully.
"""
Unit 1 - Example 1: Introduction to Artificial Intelligence

This example demonstrates:
1. Basic AI concepts and definitions
2. Types of AI systems
3. Real-world AI applications
4. Simple AI decision-making example
"""

print("=" * 60)
print("Example 1: Introduction to Artificial Intelligence")
print("=" * 60)

# 1. What is AI?
print("\n1. What is Artificial Intelligence?")
print("-" * 60)

# A one-sentence working definition — we will keep refining it throughout the course.
ai_definition = """
Artificial Intelligence (AI) is the simulation of human intelligence 
in machines that are programmed to think and learn like humans.

"""

print(ai_definition)

# 2. Types of AI Systems
print("\n2. Types of AI Systems")
print("-" * 60)

# The three classic capability tiers; every technique you will learn sits in the 'Narrow AI' tier.
ai_types = {
    "Narrow AI (Weak AI)": {
        "description": "AI designed for specific tasks", "examples": ["Siri", "Google Translate", "Image Recognition"],
    },
    "General AI (Strong AI)": {
        "description": "AI with human-level intelligence across all domains",
        "examples": ["Not yet achieved", "Theoretical"],
    },
    "Superintelligent AI": {
        "description": "AI that surpasses human intelligence",
        "examples": ["Future concept", "Theoretical"],
    }
}

for ai_type, info in ai_types.items():
    print(f"\n{ai_type}:")
    print(f"  Description: {info['description']}")
    print(f"  Examples: {', '.join(info['examples'])}")

# 3. Simple AI Decision-Making Example
print("\n" + "=" * 60)
print("3. Simple AI Decision-Making Example")
print("=" * 60)

# A rule-based (symbolic) agent: hand-written if/else rules stand in for intelligence — no learning happens here.
def simple_ai_decision(temperature, humidity, time_of_day):
    """
    Simple rule-based AI system for weather-based activity recommendation.
    
    Args:
        temperature: Temperature in Celsius
        humidity: Humidity percentage (0-100)
        time_of_day: 'morning', 'afternoon', or 'evening'
    
    Returns:
        Recommended activity
    """
    if temperature > 25 and humidity < 60:
        if time_of_day == 'morning':
            return "Go for a jog in the park"
        elif time_of_day == 'afternoon':
            return "Go swimming"
        else:
            return "Take an evening walk"
    
    elif temperature < 15:
        return "Stay indoors, read a book"
    
    elif humidity > 80:
        return "Stay indoors, watch a movie"
    
    else:
        return "Moderate weather, any outdoor activity is fine"

# Test the AI system
print("\nTesting AI Decision System:")

# Run several weather situations through the same rules to watch the recommendation change with the inputs.
test_cases = [
    (28, 50, 'morning'),
    (30, 45, 'afternoon'),
    (12, 70, 'evening'),
    (20, 85, 'afternoon')
]

for temp, hum, time in test_cases:
    decision = simple_ai_decision(temp, hum, time)
    print(f"\nTemperature: {temp}°C, Humidity: {hum}%, Time: {time}")
    print(f"Recommendation: {decision}")

# 4. AI Applications in Real World
print("\n" + "=" * 60)
print("4. AI Applications in Real World")
print("=" * 60)

# A quick survey of domains where narrow AI already creates value.
applications = {
    "Healthcare": [
        "Medical image analysis", "Drug discovery",
        "Personalized treatment plans"
    ],
    "Transportation": [
        "Autonomous vehicles",
        "Traffic optimization",
        "Route planning"
    ],
    "Finance": [
        "Fraud detection",
        "Algorithmic trading",
        "Credit scoring"
    ],
    "Education": [
        "Personalized learning",
        "Automated grading",
        "Intelligent tutoring systems"
    ]
}

for domain, apps in applications.items():
    print(f"\n{domain}:")
    for app in apps:
        print(f"  - {app}")

print("\n" + "=" * 60)
print("Example completed successfully!")
print("=" * 60)
```
