## 🎯 Start with a real system, not a definition

In May 2016 ProPublica published *Machine Bias*, an audit of **COMPAS** — a
commercial risk-scoring tool used by courts in Broward County, Florida to advise
judges on bail and sentencing. The journalists collected the scores of more than
7,000 people arrested in that county and checked, two years later, who had
actually been rearrested. Among the defendants who did **not** go on to reoffend,
Black defendants had been labelled high-risk about **45%** of the time against
about **24%** of white defendants — roughly double the rate. The tool's maker
rejected the analysis and pointed to a different fairness measure, on which the
scores did behave equally across groups. Both sides were arithmetically right,
and the argument about which measure a court should use is still running.

That is what "AI in the real world" looks like in this diploma: a program that
produced a number, a person whose bail depended on it, and a disagreement about
whether the number was fair.

### Why the vocabulary below is not filler

Every term in this notebook — *narrow AI*, *rule-based*, *decision system* — is a
label someone will use in a meeting to approve or block a system like COMPAS. If
you cannot say whether a tool learns from data or applies hand-written rules,
what exactly its output claims, and what its error rate is, you cannot challenge
it, buy it, or defend it. Vague vocabulary is how bad systems get signed off.

```python
# Guided tour of what 'artificial intelligence' means: definitions, capability tiers,
# a tiny rule-based decision maker, and real-world applications.
# This vocabulary is reused in every later notebook of the diploma, so read the output carefully.
"""
Unit 1 - Example 1: Introduction to Artificial Intelligence

This example demonstrates:
1. Basic AI concepts and definitions
2. Types of AI systems
3. Real-world AI applications
4. Simple AI decision-making example
"""

print("=" * 60)
print("Example 1: Introduction to Artificial Intelligence")
print("=" * 60)

# 1. What is AI?
print("\n1. What is Artificial Intelligence?")
print("-" * 60)

# A one-sentence working definition — we will keep refining it throughout the course.
ai_definition = """
Artificial Intelligence (AI) is the simulation of human intelligence 
in machines that are programmed to think and learn like humans.

"""

print(ai_definition)

# 2. Types of AI Systems
print("\n2. Types of AI Systems")
print("-" * 60)

# The three classic capability tiers; every technique you will learn sits in the 'Narrow AI' tier.
ai_types = {
    "Narrow AI (Weak AI)": {
        "description": "AI designed for specific tasks", "examples": ["Siri", "Google Translate", "Image Recognition"],
    },
    "General AI (Strong AI)": {
        "description": "AI with human-level intelligence across all domains",
        "examples": ["Not yet achieved", "Theoretical"],
    },
    "Superintelligent AI": {
        "description": "AI that surpasses human intelligence",
        "examples": ["Future concept", "Theoretical"],
    }
}

for ai_type, info in ai_types.items():
    print(f"\n{ai_type}:")
    print(f"  Description: {info['description']}")
    print(f"  Examples: {', '.join(info['examples'])}")

# 3. Simple AI Decision-Making Example
print("\n" + "=" * 60)
print("3. Simple AI Decision-Making Example")
print("=" * 60)

# A rule-based (symbolic) agent: hand-written if/else rules stand in for intelligence — no learning happens here.
def simple_ai_decision(temperature, humidity, time_of_day):
    """
    Simple rule-based AI system for weather-based activity recommendation.
    
    Args:
        temperature: Temperature in Celsius
        humidity: Humidity percentage (0-100)
        time_of_day: 'morning', 'afternoon', or 'evening'
    
    Returns:
        Recommended activity
    """
    if temperature > 25 and humidity < 60:
        if time_of_day == 'morning':
            return "Go for a jog in the park"
        elif time_of_day == 'afternoon':
            return "Go swimming"
        else:
            return "Take an evening walk"
    
    elif temperature < 15:
        return "Stay indoors, read a book"
    
    elif humidity > 80:
        return "Stay indoors, watch a movie"
    
    else:
        return "Moderate weather, any outdoor activity is fine"

# Test the AI system
print("\nTesting AI Decision System:")

# Run several weather situations through the same rules to watch the recommendation change with the inputs.
test_cases = [
    (28, 50, 'morning'),
    (30, 45, 'afternoon'),
    (12, 70, 'evening'),
    (20, 85, 'afternoon')
]

for temp, hum, time in test_cases:
    decision = simple_ai_decision(temp, hum, time)
    print(f"\nTemperature: {temp}°C, Humidity: {hum}%, Time: {time}")
    print(f"Recommendation: {decision}")

# 4. AI Applications in Real World
print("\n" + "=" * 60)
print("4. AI Applications in Real World")
print("=" * 60)

# A quick survey of domains where narrow AI already creates value.
applications = {
    "Healthcare": [
        "Medical image analysis", "Drug discovery",
        "Personalized treatment plans"
    ],
    "Transportation": [
        "Autonomous vehicles",
        "Traffic optimization",
        "Route planning"
    ],
    "Finance": [
        "Fraud detection",
        "Algorithmic trading",
        "Credit scoring"
    ],
    "Education": [
        "Personalized learning",
        "Automated grading",
        "Intelligent tutoring systems"
    ]
}

for domain, apps in applications.items():
    print(f"\n{domain}:")
    for app in apps:
        print(f"  - {app}")

print("\n" + "=" * 60)
print("Example completed successfully!")
print("=" * 60)
```

```python
# ── The AI / ML / DL map this notebook's first objective asks for ───────────
# WHAT: draw the three terms as NESTED regions and drop every system named in the
#       cell above (and the COMPAS story at the top) into the ring it belongs to.
# WHY : "distinguish AI, ML and DL" is objective #1 of this notebook, and the three
#       words are constantly used as if they were alternatives. They are not — each
#       is a subset of the one outside it, and a diagram settles the argument that
#       three paragraphs of prose does not.
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

fig, ax = plt.subplots(figsize=(11.2, 5.4))

RINGS = [   # outermost first; each box is strictly inside the previous one
    dict(xy=(0.15, 0.15), w=10.7, h=6.1, fc="#e8f1fb", ec="#1565c0",
         label="ARTIFICIAL INTELLIGENCE", lx=0.45, ly=5.95, col="#1565c0"),
    dict(xy=(3.05, 0.60), w=7.55, h=5.2, fc="#dcecdc", ec="#2e7d32",
         label="MACHINE LEARNING", lx=3.35, ly=5.52, col="#2e7d32"),
    dict(xy=(6.35, 1.05), w=4.05, h=4.3, fc="#f6e4c9", ec="#e07b00",
         label="DEEP LEARNING", lx=6.62, ly=5.07, col="#b35c00"),
]
for r in RINGS:
    ax.add_patch(FancyBboxPatch(r["xy"], r["w"], r["h"],
                                boxstyle="round,pad=0.02,rounding_size=0.25",
                                fc=r["fc"], ec=r["ec"], lw=2.0))
    ax.text(r["lx"], r["ly"], r["label"], fontsize=10.5, fontweight="bold",
            color=r["col"], va="center")

BANDS = [   # (x of the column, text colour, what defines the band, the examples)
    (0.45, "#0d47a1", "rules written by a person\n(no data, no training)", [
        "• `simple_ai_decision()` — the\n   recommender in THIS notebook",
        "• A* route search (notebook 08)",
        "• Expert systems (Unit 2)",
    ]),
    (3.35, "#1b5e20", "behaviour fitted to data,\nno deep network", [
        "• Spam filter trained on\n   labelled email",
        "• Logistic regression,\n   decision trees (Unit 3)",
        "• k-means clustering (Unit 2)",
    ]),
    (6.62, "#8a4700", "many-layer neural networks", [
        "• Image recognition (Unit 4)",
        "• Google Translate\n   (neural since 2016)",
        "• Siri speech recognition",
        "• ChatGPT (notebook 11)",
    ]),
]
for x, col, subtitle, items in BANDS:
    ax.text(x, 4.62, subtitle, fontsize=8.2, color=col, style="italic", va="top")
    y = 3.95
    for item in items:
        ax.text(x, y, item, fontsize=8.4, va="top")
        y -= 0.52 * (1 + item.count("\n"))

# The honest case: a system we cannot place precisely, and why.
ax.annotate(
    "COMPAS — the tool at the top of this notebook — is built by\n"
    "statistical modelling of historical records, so it sits somewhere\n"
    "in here. Its exact algorithm is a trade secret, so nobody outside\n"
    "the vendor can say where. That is the point of the story.",
    xy=(5.00, 1.02), xytext=(1.05, -1.15), fontsize=8.6, color="#b71c1c",
    arrowprops=dict(arrowstyle="->", color="#b71c1c", lw=1.3),
    bbox=dict(boxstyle="round,pad=0.35", fc="#fdecea", ec="#e57373"))
ax.plot(5.00, 1.02, marker="X", markersize=13, color="#b71c1c", zorder=5)

ax.set_xlim(-0.2, 11.1)
ax.set_ylim(-2.3, 6.6)
ax.axis("off")
ax.set_title("AI, machine learning and deep learning are nested, not alternatives\n"
             "every system named in this notebook, placed in the right ring",
             fontsize=12)
fig.tight_layout()
plt.show()
```

### 👁️ Read the AI / ML / DL map

**Three boxes, each strictly inside the one outside it.** That containment is the
whole claim: every deep-learning system is a machine-learning system, and every
machine-learning system is an AI system. The reverse is false, and the left-hand
column is the proof.

- **In the blue ring but outside green:** the `simple_ai_decision()` recommender you
  are about to map in the next cell. It is AI by the definition printed above, and it
  contains **no machine learning at all** — no data, no training, just four thresholds
  a person typed. A* search (notebook 08) and Unit 2's expert systems sit there too.
  When a vendor says "our product uses AI", this ring is where it may well live.
- **In green but outside orange:** systems whose behaviour was *fitted to data* without
  a deep network — the spam filter, and the regressions, trees and clusterings you meet
  in Units 2 and 3. Most working machine learning in industry is in this ring.
- **In orange:** many-layer neural networks. Google Translate has been in this ring
  since 2016; ChatGPT and the image recognisers of Unit 4 are here.

**Now the red X.** COMPAS is built by statistical modelling of historical records, so
it belongs somewhere in the green ring — but the vendor has never published the
algorithm, so no one outside can say where, or on what it was fitted. Read that back
against the opening story: a court was told the score was reliable, and the box it
came out of could not be pointed at on a diagram this simple. **"What ring is it in,
and who verified that?" is a question you can now ask in a procurement meeting.**


```python
# ── See the whole rule system at once: its decision map ──────────────────────
# WHAT: call simple_ai_decision() on a dense grid of temperatures and humidities — every
#       input the recommender could ever receive — and paint each point with its answer.
# WHY:  four printed test cases show four answers. The map shows ALL of them, and it
#       makes the shape of a hand-written rule system visible: flat blocks, hard edges.
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import ListedColormap
from matplotlib.patches import Patch

temperatures = np.arange(0, 48.5, 0.5)     # °C, from cold to Saudi-summer hot
humidities = np.arange(0, 100.5, 1.0)      # %

# The six answers the function can return — nothing else is reachable.
ANSWERS = ["Go for a jog in the park", "Go swimming", "Take an evening walk",
           "Stay indoors, read a book", "Stay indoors, watch a movie",
           "Moderate weather, any outdoor activity is fine"]
answer_index = {answer: i for i, answer in enumerate(ANSWERS)}
palette = ListedColormap(['#66bb6a', '#29b6f6', '#9575cd', '#ef9a9a', '#ffb74d', '#e0e0e0'])

fig, axes = plt.subplots(1, 3, figsize=(12, 3.6), sharey=True)
for ax, moment in zip(axes, ['morning', 'afternoon', 'evening']):
    # one call per grid point: this is the agent's actual output, not a sketch of it
    grid = np.array([[answer_index[simple_ai_decision(t, h, moment)] for t in temperatures]
                     for h in humidities])
    ax.imshow(grid, origin='lower', aspect='auto', cmap=palette, vmin=0, vmax=5,
              extent=[temperatures[0], temperatures[-1], humidities[0], humidities[-1]])
    for t, h, moment_of_case in test_cases:          # the four cases printed above
        if moment_of_case == moment:
            ax.plot(t, h, marker='o', markersize=9, markerfacecolor='white',
                    markeredgecolor='black', markeredgewidth=1.6)
            ax.annotate(f"({t}°C, {h}%)", (t, h), xytext=(5, 8),
                        textcoords='offset points', fontsize=8)
    ax.set_title(f"time_of_day = '{moment}'", fontsize=10)
    ax.set_xlabel('temperature (°C)')
axes[0].set_ylabel('humidity (%)')

# An input nobody tested: 45 °C in the afternoon. The rule answers anyway.
extreme = simple_ai_decision(45, 30, 'afternoon')
axes[1].plot(45, 30, marker='*', markersize=17, color='crimson')
axes[1].annotate(f'45 °C, 30% → "{extreme}"', (45, 30), xytext=(-150, -26),
                 textcoords='offset points', fontsize=8, color='crimson',
                 arrowprops=dict(arrowstyle='->', color='crimson'))

fig.legend(handles=[Patch(facecolor=palette(i), edgecolor='#999', label=a)
                    for i, a in enumerate(ANSWERS)],
           loc='lower center', ncol=3, frameon=False, fontsize=8)
fig.suptitle('Every answer the rule-based recommender can give, for every possible input',
             fontsize=12)
fig.tight_layout(rect=[0, 0.17, 1, 0.93])
plt.show()

# The boundary in numbers: one degree and two percentage points flip the advice.
print("Two inputs a person would call identical, two different answers:")
for t, h in [(26, 59), (26, 61), (25, 50), (26, 50)]:
    print(f"  {t} °C, {h:>2}% humidity, morning -> {simple_ai_decision(t, h, 'morning')}")
print(f"\nAn input nobody tested — 45 °C, 30% humidity, afternoon -> {extreme}")
```

### 👁️ Read the map

Each panel is the *whole input space* of the recommender you just ran: temperature
across the bottom, humidity up the side, and the colour is the sentence the function
returns at that point. The white circles are the four test cases printed above; the
red star is an input nobody tested.

**What the picture proves:**

- **A rule system is a set of rectangles.** There are no curves and no gradients —
  just flat blocks with hard edges at exactly the numbers in the code:
  `temperature > 25`, `temperature < 15`, `humidity < 60`, `humidity > 80`. Those four
  thresholds *are* the system. Everything it will ever say is decided by where you
  land relative to them.
- **The edges are cliffs, not slopes.** The printed lines under the figure show it:
  26 °C at 59% humidity is a jog in the park, 26 °C at 61% is "moderate weather" —
  two percentage points of humidity, a completely different recommendation, and no
  hint anywhere that the call was close. Every model in this diploma has to answer
  this question somehow; a rule answers it with a cliff.
- **The system never says "I don't know".** The map has no white space. At **45 °C**
  in the afternoon — dangerous heat in a Saudi summer — the red star still lands in
  the blue block and the machine recommends going swimming, with exactly the same
  confidence as for a mild spring morning. A hand-written rule cannot report that it
  has left the conditions it was written for.
- **Time of day changes surprisingly little.** Compare the three panels: only the
  hot-and-dry block changes colour. The red, orange and grey regions are identical,
  because those branches never look at `time_of_day` at all.

Hold this picture next to the COMPAS story at the top of the notebook. That system,
too, turned continuous facts about a person into a block with a hard edge — and the
argument was about where the edge should be and who was allowed to see it.

## ⚠️ Where this breaks

The definition at the top of this notebook — "the simulation of human
intelligence in machines" — is the one most textbooks use, and it is the weakest
thing on this page.

- **It is not testable.** Nothing about the `if/else` recommender you just ran
  "simulates human intelligence", yet it has exactly the shape of systems sold as
  AI. A definition that admits everything decides nothing.
- **The three-tier ladder is not a roadmap.** Narrow → general → superintelligent
  reads like progress with a scale bar. It is not: every technique in this
  twelve-course diploma sits in the narrow tier, and no result says that stacking
  narrow systems produces a general one. Read the top two rows as *categories that
  are currently empty*, not as next year's release.
- **"Human-level" is the wrong yardstick for buying software.** COMPAS was never
  meant to think like a judge — it was meant to predict rearrest. The questions
  that decide whether it should be used are the ones this course keeps returning
  to: predicting what exactly, measured against which baseline, on whose data,
  and what happens to the person when it is wrong.
- **The rule-based demo has a limit you can see.** Feed it a temperature of 45 °C
  and it still answers confidently, because a hand-written rule never says "I have
  not seen this before". That silence is the failure mode of every rule system in
  Unit 2 — and, in a different costume, of every model in Units 3-5.

**Use instead:** for the rest of this diploma, describe a system by its *inputs,
its output, its training data and its error rate* — never by how human-like it seems.
