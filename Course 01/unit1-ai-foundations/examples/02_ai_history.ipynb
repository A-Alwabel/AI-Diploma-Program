```python
"""
Unit 1 - Example 2: History of Artificial Intelligence
الوحدة 1 - مثال 2: تاريخ الذكاء الاصطناعي

This example demonstrates:
1. Key milestones in AI history
2. AI Winter periods
3. Modern AI renaissance
4. Timeline visualization
"""

# Note: Visualization libraries would be used for timeline plots
# ملاحظة: ستستخدم مكتبات التصور لرسم الخطوط الزمنية
# Note: Visualization libraries would be used for timeline plots
# ملاحظة: ستستخدم مكتبات التصور لرسم الخطوط الزمنية

print("=" * 60)
print("Example 2: History of Artificial Intelligence")
print("مثال 2: تاريخ الذكاء الاصطناعي")
print("=" * 60)

# AI History Timeline
# الخط الزمني لتاريخ الذكاء الاصطناعي
ai_timeline = [
    {
        "year": 1950, "event": "Turing Test proposed",
        "arabic": "اقتراح اختبار تورنغ",
        "significance": "Foundation of AI philosophy"
    },
    {
        "year": 1956,
        "event": "Dartmouth Conference - AI term coined",
        "arabic": "مؤتمر دارتموث - صياغة مصطلح الذكاء الاصطناعي",
        "significance": "Birth of AI as a field"
    },
    {
        "year": 1997,
        "event": "Deep Blue defeats world chess champion",
        "arabic": "ديب بلو يهزم بطل العالم في الشطرنج",
        "significance": "First major AI victory"
    },
    {
        "year": 2011,
        "event": "IBM Watson wins Jeopardy!",
        "arabic": "IBM Watson يفوز في Jeopardy!",
        "significance": "Natural language processing breakthrough"
    },
    {
        "year": 2016,
        "event": "AlphaGo defeats Go world champion",
        "arabic": "AlphaGo يهزم بطل العالم في Go",
        "significance": "Deep learning milestone"
    },
    {
        "year": 2022,
        "event": "ChatGPT released",
        "arabic": "إطلاق ChatGPT",
        "significance": "Generative AI revolution"
    }
]

print("\nAI History Timeline:")
print("الخط الزمني لتاريخ الذكاء الاصطناعي:")
print("-" * 60)

for milestone in ai_timeline:
    print(f"\n{milestone['year']}: {milestone['event']}")
    print(f"  {milestone['arabic']}")
    print(f"  Significance: {milestone['significance']}")

# Visualization: AI History Timeline
# تصور: الخط الزمني لتاريخ الذكاء الاصطناعي
try:
    import matplotlib.pyplot as plt
    import matplotlib.dates as mdates
    from datetime import datetime
    
    # Extract years and events
    years = [milestone['year'] for milestone in ai_timeline]
    events = [milestone['event'] for milestone in ai_timeline]
    
    # Create timeline plot
    fig, ax = plt.subplots(figsize=(14, 8))
    
    # Plot timeline
    y_pos = range(len(years))
    colors = plt.cm.viridis([y/2022 for y in years])
    
    # Draw timeline line
    ax.plot(years, y_pos, 'o-', linewidth=3, markersize=12, color='steelblue', alpha=0.7)
    
    # Add event labels
    for i, (year, event) in enumerate(zip(years, events)):
        ax.annotate(f'{year}\n{event[:30]}...', xy=(year, i), xytext=(10, 0),
                   textcoords='offset points', fontsize=9,
                   bbox=dict(boxstyle='round,pad=0.5', facecolor='lightyellow', alpha=0.7),
                   arrowprops=dict(arrowstyle='->', connectionstyle='arc3,rad=0'))
    
    ax.set_yticks(y_pos)
    ax.set_yticklabels([f'{y}' for y in years])
    ax.set_xlabel('Year | السنة', fontsize=12, fontweight='bold')
    ax.set_ylabel('Milestones | المعالم', fontsize=12, fontweight='bold')
    ax.set_title('AI History Timeline | الخط الزمني لتاريخ الذكاء الاصطناعي', fontsize=16, pad=20, fontweight='bold')
    ax.grid(True, alpha=0.3, linestyle='--')
    ax.set_xlim(1945, 2025)
    plt.tight_layout()
    plt.show()
    print("\n✅ Timeline visualization displayed")
    print("تم عرض التصور الزمني بنجاح")
except ImportError:
    print("\n⚠️  Install matplotlib for timeline visualization:")
    print("   pip install matplotlib")
    print("ملاحظة: قم بتثبيت matplotlib لتصور الخط الزمني")

# AI Winter Periods
# فترات شتاء الذكاء الاصطناعي
print("\n" + "=" * 60)
print("AI Winter Periods")
print("فترات شتاء الذكاء الاصطناعي")
print("=" * 60)

ai_winters = [
    {
        "period": "1974-1980", "reason": "Overpromising and underdelivering",
        "arabic": "الوعود المفرطة وعدم الوفاء بها"
    },
    {
        "period": "1987-1993",
        "reason": "Expert systems limitations",
        "arabic": "قيود الأنظمة الخبيرة"
    }
]

for winter in ai_winters:
    print(f"\n{winter['period']}: {winter['reason']}")
    print(f"  {winter['arabic']}")

# Modern AI Renaissance
# نهضة الذكاء الاصطناعي الحديثة
print("\n" + "=" * 60)
print("Modern AI Renaissance (2010s-present)")
print("نهضة الذكاء الاصطناعي الحديثة (2010 حتى الآن)")
print("=" * 60)

renaissance_factors = [
    "Big Data availability", "Computational power increase (GPUs)",
    "Improved algorithms (Deep Learning)",
    "Open source frameworks (TensorFlow, PyTorch)",
    "Cloud computing accessibility"
]

print("\nKey Factors:")
print("العوامل الرئيسية:")
for i, factor in enumerate(renaissance_factors, 1):
    print(f"  {i}. {factor}")

print("\n" + "=" * 60)
print("Example completed successfully!")
print("تم إكمال المثال بنجاح!")
print("=" * 60)


# Visualize AI history timeline
# تصور الخط الزمني لتاريخ الذكاء الاصطناعي
try:
    import matplotlib.pyplot as plt
    
    # Extract years and events from timeline
    years = [milestone['year'] for milestone in ai_timeline]
    events = [milestone['event'] for milestone in ai_timeline]
    
    plt.figure(figsize=(14, 8))
    plt.plot(years, range(len(years)), 'o-', linewidth=2.5, markersize=12, 
             color='steelblue', markerfacecolor='lightblue', markeredgewidth=2)
    
    # Add event labels
    for i, (year, event) in enumerate(zip(years, events)):
        plt.annotate(f'{year}\n{event[:30]}...', (year, i), 
                    textcoords="offset points", 
                    xytext=(0, 20 if i % 2 == 0 else -30),
                    ha='center', fontsize=9, 
                    bbox=dict(boxstyle='round,pad=0.5', facecolor='yellow', alpha=0.7))
    
    plt.xlabel('Year | السنة', fontsize=13, fontweight='bold')
    plt.ylabel('Milestone | المعلم', fontsize=13, fontweight='bold')
    plt.title('AI History Timeline | الخط الزمني لتاريخ الذكاء الاصطناعي', fontsize=15, pad=20, fontweight='bold')
    plt.grid(True, alpha=0.3, linestyle='--')
    plt.tight_layout()
    plt.show()
    print("\n✅ AI history timeline visualization displayed!")
    print("تم عرض الخط الزمني لتاريخ الذكاء الاصطناعي بنجاح!")
except ImportError:
    print("\n📦 To see timeline visualization, install:")
    print("   pip install matplotlib")
    print("\nملاحظة: لرؤية التصور، قم بتثبيت:")
    print("   pip install matplotlib")
```
