## 🎯 The report that froze a field

The timeline below marks two shaded gaps labelled "AI winter". The first one has
a named cause you can go and read. In **1973** the British mathematician **Sir
James Lighthill** delivered *Artificial Intelligence: A General Survey* to the UK
Science Research Council. His verdict was blunt — "in no part of the field have
the discoveries made so far produced the major impact that was then promised" —
and his central technical objection was **combinatorial explosion**: the methods
that worked on toy problems needed exponentially more search as the problem grew,
so they would never reach useful scale. The UK government cut AI funding at all
but a handful of universities; researchers left the field or left the country. A
televised BBC debate the same year, where Lighthill argued against John McCarthy,
Donald Michie and Richard Gregory, is still on the record.

The second winter, in the late 1980s, killed a *commercial* success rather than a
research programme. You will meet it by name in Unit 2, when you build an expert
system of the kind that collapsed.

### Why a history lesson is a technical lesson

Both winters followed the same three steps: a real result on a small problem, a
public promise that it would scale, and a bill that came due when it did not.
You will feel the pull of that sequence yourself in this very course — the XOR
network in Unit 3, the CNN in Unit 4 and the GAN in Unit 5 all work beautifully
on small data, and each of them has a scale at which it stops working. Knowing
that the field has twice mistaken a working demo for a working product is the
cheapest protection you will ever get against doing it a third time.

```python
# Seven decades of AI history in one cell: milestones, the two 'AI winters', and the
# modern renaissance — then the same timeline drawn as a plot.
# Knowing why the field boomed and crashed before helps you judge today's AI claims critically.
"""
Unit 1 - Example 2: History of Artificial Intelligence

This example demonstrates:
1. Key milestones in AI history
2. AI Winter periods
3. Modern AI renaissance
4. Timeline visualization
"""


print("=" * 60)
print("Example 2: History of Artificial Intelligence")
print("=" * 60)

# AI History Timeline
# Each milestone records the year, the event, and why it mattered to the field.
ai_timeline = [
    {
        "year": 1950, "event": "Turing Test proposed",
        "significance": "Foundation of AI philosophy"
    },
    {
        "year": 1956,
        "event": "Dartmouth Conference - AI term coined",
        "significance": "Birth of AI as a field"
    },
    {
        "year": 1997,
        "event": "Deep Blue defeats world chess champion",
        "significance": "First major AI victory"
    },
    {
        "year": 2011,
        "event": "IBM Watson wins Jeopardy!",
        "significance": "Natural language processing breakthrough"
    },
    {
        "year": 2016,
        "event": "AlphaGo defeats Go world champion",
        "significance": "Deep learning milestone"
    },
    {
        "year": 2022,
        "event": "ChatGPT released",
        "significance": "Generative AI revolution"
    }
]

print("\nAI History Timeline:")
print("-" * 60)

for milestone in ai_timeline:
    print(f"\n{milestone['year']}: {milestone['event']}")
    print(f"  Significance: {milestone['significance']}")

# Visualization: AI History Timeline
try:
    print("\n✅ Timeline visualization displayed")
except ImportError:
    print("\n⚠️  Install matplotlib for timeline visualization:")
    print("   pip install matplotlib")

# AI Winter Periods
print("\n" + "=" * 60)
print("AI Winter Periods")
print("=" * 60)

# The two 'AI winters': funding collapses that followed periods of over-promising.
ai_winters = [
    {
        "period": "1974-1980", "reason": "Overpromising and underdelivering",
    },
    {
        "period": "1987-1993",
        "reason": "Expert systems limitations",
    }
]

for winter in ai_winters:
    print(f"\n{winter['period']}: {winter['reason']}")

# Modern AI Renaissance
print("\n" + "=" * 60)
print("Modern AI Renaissance (2010s-present)")
print("=" * 60)

# The ingredients that ended the last winter — data, compute, better algorithms, open tooling.
renaissance_factors = [
    "Big Data availability", "Computational power increase (GPUs)",
    "Improved algorithms (Deep Learning)",
    "Open source frameworks (TensorFlow, PyTorch)",
    "Cloud computing accessibility"
]

print("\nKey Factors:")
for i, factor in enumerate(renaissance_factors, 1):
    print(f"  {i}. {factor}")

print("\n" + "=" * 60)
print("Example completed successfully!")
print("=" * 60)
```

```python
# ── The timeline the opening paragraph promised: milestones AND the two winters ──
# WHAT: put the same ai_timeline entries on a single year axis and shade the two funding
#       collapses from ai_winters on top of them, so the gaps are visible, not implied.
# WHY:  the plots above draw six wins on a rising line. A field that boomed and froze
#       twice does not look like a rising line, and a student should see the difference.
import matplotlib.pyplot as plt

years = [m['year'] for m in ai_timeline]
events = [m['event'] for m in ai_timeline]
# Read the winter spans straight out of the ai_winters list defined above ("1974-1980").
winter_spans = [tuple(int(y) for y in w['period'].split('-')) for w in ai_winters]

fig, ax = plt.subplots(figsize=(12, 3.4))

# 1) the two winters, as shaded bands across the whole plot
for start, end in winter_spans:
    ax.axvspan(start, end, color='#90a4ae', alpha=0.35, zorder=0)
    ax.text((start + end) / 2, 0.78, f"AI winter\n{start}–{end}", ha='center', va='center',
            fontsize=8.5, color='#37474f', fontweight='bold')

# 2) the milestones, alternating above and below the axis line so labels do not collide
ax.axhline(0, color='#455a64', linewidth=1.6, zorder=1)
for i, (year, event) in enumerate(zip(years, events)):
    up = i % 2 == 0
    y_text = 0.42 if up else -0.42
    ax.plot([year, year], [0, y_text * 0.55], color='#455a64', linewidth=1.0, zorder=2)
    ax.plot(year, 0, marker='o', markersize=9, color='#1565c0', zorder=3)
    ax.text(year, y_text, f"{year}\n{event}", ha='center', va='center', fontsize=8,
            bbox=dict(boxstyle='round,pad=0.3', facecolor='#e3f2fd', edgecolor='#90caf9'))

ax.set_xlim(1945, 2028)
ax.set_ylim(-0.95, 0.95)
ax.set_yticks([])
ax.set_xlabel('year')
ax.set_title('AI history: six milestones and the two funding winters, on one axis', fontsize=12)
for side in ['left', 'right', 'top']:
    ax.spines[side].set_visible(False)
fig.tight_layout()
plt.show()

# The gaps between milestones — the part a "list of wins" never shows you.
print("Years between consecutive milestones on this timeline:")
for (y0, e0), (y1, e1) in zip(list(zip(years, events)), list(zip(years, events))[1:]):
    print(f"  {y0} -> {y1}: {y1 - y0:>2} years   ({e0[:28]:<28} -> {e1[:28]})")
winter_years = sum(end - start for start, end in winter_spans)
print(f"\nSpan covered by the chart : {years[-1] - years[0]} years")
print(f"Years inside a winter     : {winter_years}")
print(f"Longest gap with no entry : {max(y1 - y0 for y0, y1 in zip(years, years[1:]))} years "
      f"(1956 -> 1997) — and both winters fall inside it.")
```

### 👁️ Read the timeline

One axis, 1945 to 2028. Blue dots are the six milestones the cell above printed; the
grey bands are the two funding collapses listed in `ai_winters`. This is the picture
the first paragraph of this notebook promised you.

**What the picture proves:**

- **The milestones are not evenly spread.** Two at the beginning (1950, 1956), then
  nothing for **41 years** until Deep Blue in 1997, then four in the last 26 years.
  Progress that looks steady on a rising line is, in real time, long silence
  punctuated by news.
- **The silence is where the winters are.** Both grey bands sit inside that 41-year
  gap. The chart's emptiest stretch is the stretch where the field ran out of money
  — and a timeline of wins renders that as *nothing at all*.
- **The bands carry dates and no names.** Look at what is written inside them: a
  period, and the word "winter". The 1973 Lighthill report that caused the first one
  is not on this chart, because it was not a win, and lists of wins do not record the
  documents that stopped things. You learned about it from a paragraph of text, not
  from a figure — which is exactly the criticism this notebook makes of its own timeline.

Both winters began after a public promise that a working demo would scale. When you
next see a chart like this in a vendor deck, the question to ask is not "what is on
it" but "what would have to have happened for something to be left off".

## ⚠️ Where this breaks

This timeline is a *selection*, and every selection argues for a conclusion.

- **It is a list of wins.** Six entries, six triumphs. The winters appear as
  labelled gaps with no names attached — yet the Lighthill report and the
  late-1980s collapse of the LISP-machine and expert-system market shaped the
  field at least as much as Deep Blue did. A timeline of successes teaches you
  that progress is monotonic. It is not.
- **The dates belong to the winners.** "2016: AlphaGo" is here because it made
  headlines. The things that made it possible — GPUs, ImageNet-scale labelled
  data, and years of approaches that failed — have no dates on this chart.
  Milestones are visible; the conditions that produce them are not.
- **A milestone is not a deployment.** Every entry is a demonstration under
  contest conditions, with unlimited engineering attention and a fixed opponent.
  None of them tells you whether the same technique worked in a hospital, a court
  or a factory — the only question an applied AI diploma can afford to care about.
- **The chart cannot show what stayed hard.** Robotic manipulation, common-sense
  reasoning and reliable long-horizon planning have no milestone rows because they
  produced no single dramatic day. Absence from a timeline is not absence of
  importance.

**Use instead:** when you meet an AI claim, ask for the three things this
timeline hides — which task exactly, measured against which baseline, and did it
survive contact with a real deployment.
