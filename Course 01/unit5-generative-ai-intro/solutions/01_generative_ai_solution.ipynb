```python
"""
Unit 5 - Solution 1: Generative AI Concepts
الوحدة 5 - حل 1: مفاهيم الذكاء الاصطناعي التوليدي

Complete solutions to Exercise 1.
"""

import numpy as np
import matplotlib.pyplot as plt
print("=" * 70)
print("Solution 1: Generative AI Concepts")
print("=" * 70)

# Solution 1: Identify Model Types
print("\nSolution 1: Identify Model Types")
print("-" * 70)

models = {
    "Logistic Regression": "D",  # Discriminative - learns P(Y|X)
    "GAN (Generative Adversarial Network)": "G",  # Generative - creates new data
    "Support Vector Machine": "D",  # Discriminative - learns decision boundary
    "VAE (Variational Autoencoder)": "G",  # Generative - learns data distribution
    "Neural Network Classifier": "D",  # Discriminative - learns P(Y|X)
    "Naive Bayes": "G"  # Generative - learns P(X|Y) and P(Y)
}

for model, answer in models.items():
    print(f"{model}: {answer}")

# Solution 2: Generate Synthetic Data
print("\nSolution 2: Generate Synthetic Data")
print("-" * 70)

def generate_square_pattern(n_points=100):
    """
    Generate synthetic data points forming a square pattern.
    """
    side_length = 4
    points_per_side = n_points // 4
    # Top side
    x_top = np.linspace(-side_length/2, side_length/2, points_per_side)
    y_top = np.full(points_per_side, side_length/2)
    # Right side
    x_right = np.full(points_per_side, side_length/2)
    y_right = np.linspace(side_length/2, -side_length/2, points_per_side)
    # Bottom side
    x_bottom = np.linspace(side_length/2, -side_length/2, points_per_side)
    y_bottom = np.full(points_per_side, -side_length/2)
    # Left side
    x_left = np.full(points_per_side, -side_length/2)
    y_left = np.linspace(-side_length/2, side_length/2, points_per_side)
    # Combine and add noise
    x = np.concatenate([x_top, x_right, x_bottom, x_left])
    y = np.concatenate([y_top, y_right, y_bottom, y_left])
    x += np.random.normal(0, 0.1, len(x))
    y += np.random.normal(0, 0.1, len(y))
    return x, y

# Test and visualize
x_square, y_square = generate_square_pattern(200)
plt.figure(figsize=(8, 8))
plt.scatter(x_square, y_square, alpha=0.6, s=30)
plt.title('Generated Square Pattern\n(Solution)', fontsize=12)
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(True, alpha=0.3)
plt.axis('equal')
plt.show()

print("✅ Square pattern generated successfully!")

# Solution 3: Compare Approaches
print("\nSolution 3: Compare Generative vs Discriminative")
print("-" * 70)

comparison = {
    "What does it learn?": {
        "Generative": "P(X|Y) and P(Y) - Joint probability distribution", "Discriminative": "P(Y|X) - Conditional probability"
    },
    "Can it generate new data?": {
        "Generative": True,
        "Discriminative": False
    },
    "Best use case": {
        "Generative": "Image generation, data augmentation, anomaly detection",
        "Discriminative": "Classification, prediction, spam detection"
    }
}

for question, answers in comparison.items():
    print(f"\n{question}")
    for model_type, answer in answers.items():
        print(f"  {model_type}: {answer}")

print("\n" + "=" * 70)
print("✅ All exercises completed!")
print("=" * 70)
```
