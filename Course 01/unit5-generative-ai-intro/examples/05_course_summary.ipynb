## 🎯 Where this course actually lands: an Arabic-first model, built here

In 2024 a team at Saudi Arabia's **SDAIA** published *ALLaM: Large Language Models
for Arabic and English* — a family of Arabic-first models trained with a
deliberate second-language-acquisition recipe rather than by bolting Arabic onto
an English model. A year later the **Absher** benchmark (Al-Monef et al., 2025)
put a number on the gap that generic models leave behind, with over 18,000
questions spanning Saudi dialects, and a 2026 study (*Are Arabic Benchmarks
Reliable?*) went further and validated the benchmarks themselves before ranking
any model.

Read that sequence as a map of this course. ALLaM is Units 4 and 5 — neural
networks and generative models — at national scale. Absher is Unit 3: you cannot
claim performance without a held-out measurement on the population you actually
serve. And the benchmark-validation paper is the habit this course has repeated in
every notebook: **check the measuring instrument before you trust the measurement.**

The skills in this diploma are not preparation for someone else's field. The gap
Absher measures — models that work in Modern Standard Arabic and fail in the
dialect a citizen actually types — is a gap in *your* region, and closing it needs
exactly what the five units below trained you to do.

### What goes wrong without connecting the units

Each unit taught a tool, and a tool used without the others produces the failures
you met in this course: a search that ignores the opponent (Unit 1), a rule base
nobody can maintain (Unit 2), a model with no baseline (Unit 3), a network that
memorised 300 passengers (Unit 4), a generator with no coverage metric (Unit 5).
The through-line is not the algorithms. It is one habit: **state what you are
measuring, against what baseline, on data the model has not seen — then say what
would make you wrong.**

## ⚠️ Where this breaks — what this course did not teach you

An honest summary has to include the gaps. None of these is a criticism of the
plan; all of them are things a client will assume you covered.

- **Nothing here was deployed.** Every model in this course ran once, in a
  notebook, on a laptop, and was then abandoned. Serving, latency, versioning,
  monitoring, retraining and rollback — the majority of the work in a real ML
  system — are Course 05 and beyond.
- **Every dataset was small enough to be convenient.** 442 patients, 569 biopsies,
  600 Titanic passengers, a 6,000-image MNIST subset, 2,000 hours of call volume.
  Nothing about data pipelines, distributed training, or the failure modes that
  only appear at scale appears anywhere in this course.
- **No result here was repeated.** Almost every number was produced by a single
  run with a single seed on a single split. You have seen how large the noise can
  be — the 50% XOR run in Unit 3, the 2-image difference in Unit 4, notebook 04.
  Cross-validation and repeated runs arrive in Course 04.
- **We measured accuracy, not cost.** Not once did we attach a currency value, a
  clinical harm, or a legal consequence to a false positive against a false
  negative — even though every real case in this course (COMPAS, Epic sepsis,
  Zillow, Air Canada) failed on exactly that asymmetry. Course 06 makes it a
  requirement.
- **Fairness was discussed and never measured.** You read about COMPAS, the Amazon
  résumé screener and the Dutch benefits scandal. You did not compute a single
  group-wise error rate. That is Course 06, and it is a skill, not an opinion.
- **The frontier moved while this plan was written.** GANs have largely given way
  to diffusion for images; RNNs to transformers for sequences; "prompt
  engineering" to retrieval and context engineering; and the 2026 security
  question is less about the model than about an agent that read an untrusted
  document and then used a real tool. What survives is the reasoning you practised
  here — baseline, change one thing, re-measure, state the limit — and it survives
  precisely because it does not depend on which architecture is current.

**The single sentence to leave with:** a model that cannot be beaten by a baseline,
explained to the person it affects, and described honestly in terms of what would
break it, is not finished — however good its accuracy looks.
