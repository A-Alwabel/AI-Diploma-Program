## 🎯 One in five genetics papers, corrupted by a spreadsheet

In 2016 Mark Ziemann, Yotam Eren and Assam El-Osta published *Gene name errors
are widespread in the scientific literature* in **Genome Biology**. They scanned
the supplementary Excel files of leading genomics journals and found that roughly
**one paper in five** contained gene names silently converted into something else:
Excel's default settings turn the gene symbol `SEPT2` into the date *2-Sep*, and
`MARCH1` into *1-Mar*. Nobody noticed, because nothing failed — the file opened,
the column looked full, the analysis ran.

The fix, four years later, is the part that should stay with you. In **August
2020** the HUGO Gene Nomenclature Committee — the body that names human genes —
gave up and renamed about **27 genes** so that a spreadsheet would stop
misreading them. `MARCH1` became `MARCHF1`; `SEPT1` became `SEPTIN1`. Human
biology was renamed to accommodate a data-handling default.

### Why this is a lesson about lists, dictionaries and files

Every step in that failure is a step in this notebook. A value was read from a
file. Its **type** was guessed instead of declared. It was stored in a container
that did not care what it held. Nothing raised an error. When you write
`data['grades']` or `float(line.split(':')[1])` below, you are making the same
three decisions — about types, containers and parsing — that decided whether a
genomics paper was correct. Doing that in code, where the parsing rule is visible
and reviewable, is the whole argument for this notebook.

```python
# Python Lists - Essential for AI data processing
print("=" * 60)
print("PYTHON LISTS REVIEW")
print("=" * 60)

# Creating lists
data_points = [1, 2, 3, 4, 5]
features = ['age', 'income', 'education']
print(f"\nData points: {data_points}")
print(f"Features: {features}")

# List operations
data_points.append(6)
print(f"\nAfter append: {data_points}")

# List comprehension (common in AI)
squared = [x**2 for x in data_points]
print(f"Squared values: {squared}")

# Slicing
print(f"First 3 elements: {data_points[:3]}")
print(f"Last 2 elements: {data_points[-2:]}")

print("\n✅ Lists reviewed!")
```

```python
# Python Dictionaries - Key-value pairs for AI data
print("=" * 60)
print("PYTHON DICTIONARIES REVIEW")
print("=" * 60)

# Creating dictionaries
student_data = {
    'name': 'Alice', 'age': 20,
    'grades': [85, 90, 88]
}

model_params = {
    'learning_rate': 0.01,
    'epochs': 100,
    'batch_size': 32
}

print(f"\nStudent data: {student_data}")
print(f"Model parameters: {model_params}")

# Dictionary operations
print(f"\nStudent name: {student_data['name']}")
print(f"Learning rate: {model_params.get('learning_rate', 'Not set')}")

# Adding/updating
student_data['major'] = 'AI'
print(f"\nUpdated student data: {student_data}")

# Iterating
print("\nModel parameters:")
for key, value in model_params.items():
    print(f"  {key}: {value}")

print("\n✅ Dictionaries reviewed!")
```

```python
# File Handling - Reading and writing data
print("=" * 60)
print("FILE HANDLING REVIEW")
print("=" * 60)

# Writing to a file
data_to_save = {
    'dataset': 'iris', 'samples': 150,
    'features': 4
}

# Write JSON-like data
with open('temp_data.txt', 'w') as f:
    for key, value in data_to_save.items():
        f.write(f"{key}: {value}\n")

print("✅ Data written to file")

# Reading from a file
with open('temp_data.txt', 'r') as f:
    content = f.read()
    print(f"\nFile content:\n{content}")

# Reading line by line
print("\nReading line by line:")
with open('temp_data.txt', 'r') as f:
    for line in f:
        print(f"  {line.strip()}")

# Cleanup
import os
if os.path.exists('temp_data.txt'):
    os.remove('temp_data.txt')
    print("\n✅ Temporary file cleaned up")

print("\n✅ File handling reviewed!")
```

```python
# 📊 FIGURE — what a text file does to your data, measured on the very file this
# notebook just wrote and read back.
# WHAT: we write eight records with the same `f"{key}: {value}"` line format used above,
#       read them back with the parser the Discuss box describes — `line.split(':')[1]` —
#       and compare each recovered value against the value that went in. The third column
#       repeats the round trip through the `json` module for contrast.
# WHY: "splitting text by hand is not parsing" is a sentence. Drawn, it becomes a count:
#       how many of eight values came back as what you stored, and how many errors fired.
import json
import os
import matplotlib.pyplot as plt

# --- WHAT: eight records chosen because each one is an ordinary thing to store -----
records = {
    "dataset":       "iris",     # plain text
    "samples":       150,        # an integer
    "features":      4,          # an integer
    "test_fraction": 0.2,        # a float
    "shuffled":      True,       # a boolean
    "collected_at":  "09:30",    # text that contains a colon
    "class_ratio":   "3:1",      # text that contains a colon
    "gene":          "SEPT2",    # the real gene symbol from the story above
}

# --- Round trip 1: the hand-written format and the hand-written parser ------------
with open("roundtrip_demo.txt", "w") as f:
    for k, v in records.items():
        f.write(f"{k}: {v}\n")

naive, errors_raised = {}, 0
with open("roundtrip_demo.txt") as f:
    for line in f:
        try:
            key = line.split(":")[0]
            naive[key] = line.split(":")[1].strip()   # "first colon, throw the rest away"
        except Exception:
            errors_raised += 1

# --- Round trip 2: the same records through a real serialiser --------------------
with open("roundtrip_demo.json", "w") as f:
    json.dump(records, f)
with open("roundtrip_demo.json") as f:
    parsed = json.load(f)

def same(a, b):
    """Identical value AND identical type — the only definition of 'it came back'."""
    return type(a) is type(b) and a == b

ok_naive = [k for k in records if same(records[k], naive[k])]
ok_json = [k for k in records if same(records[k], parsed[k])]

# ---------------------------------------------------------------------------------
fig, (axT, axB) = plt.subplots(1, 2, figsize=(13.0, 4.6),
                               gridspec_kw={"width_ratios": [2.35, 1]})

GREEN, RED, GREY = "#DCEEDC", "#F8D7D7", "#EFEFEF"
keys = list(records)
cols = ["written to the file", "read back with split(':')[1]", "read back with json"]
for c, title in enumerate(cols):
    axT.text(c + 0.5, len(keys) + 0.30, title, ha="center", va="bottom",
             fontsize=9, fontweight="bold")
for r, k in enumerate(keys):
    y = len(keys) - 1 - r
    axT.text(-0.12, y + 0.5, k, ha="right", va="center", fontsize=9, family="monospace")
    trio = [(records[k], GREY), (naive[k], GREEN if same(records[k], naive[k]) else RED),
            (parsed[k], GREEN if same(records[k], parsed[k]) else RED)]
    for c, (val, colour) in enumerate(trio):
        axT.add_patch(plt.Rectangle((c, y), 1, 1, facecolor=colour, edgecolor="white", lw=2))
        axT.text(c + 0.5, y + 0.60, repr(val), ha="center", va="center", fontsize=8.5,
                 family="monospace")
        axT.text(c + 0.5, y + 0.26, type(val).__name__, ha="center", va="center",
                 fontsize=7.5, color="#666", style="italic")
axT.set_xlim(-1.50, 3.05); axT.set_ylim(0, len(keys) + 0.85)
axT.set_axis_off()
axT.set_title(f"Eight values went into a text file. {len(ok_naive)} of {len(records)} came back unchanged.\n"
              "green = same value AND same type   ·   red = something changed",
              fontsize=10.5, color="#222", y=1.10)

bars = axB.bar(["split(':')[1]", "json"], [len(ok_naive), len(ok_json)],
               color=["#E45756", "#54A24B"], width=0.55)
for b, n in zip(bars, [len(ok_naive), len(ok_json)]):
    axB.text(b.get_x() + b.get_width() / 2, n + 0.15, f"{n} of {len(records)}",
             ha="center", fontweight="bold")
axB.set_ylim(0, len(records) + 1.4)
axB.set_ylabel("records recovered intact (value + type)")
axB.set_title(f"Errors raised by either parser: {errors_raised}", fontsize=10)
axB.grid(axis="y", alpha=0.25)

plt.tight_layout()
plt.show()

for path in ("roundtrip_demo.txt", "roundtrip_demo.json"):
    if os.path.exists(path):
        os.remove(path)

print(f"Records written            : {len(records)}")
print(f"Recovered intact by split(): {len(ok_naive)} -> {ok_naive}")
print(f"Recovered intact by json   : {len(ok_json)} -> {ok_json}")
print(f"Exceptions raised          : {errors_raised}")
print("\nfield            written        -> split(':')[1]   -> json")
for k in records:
    print(f"  {k:14s} {repr(records[k]):>13s}  -> {repr(naive[k]):>14s}  -> {repr(parsed[k])}")
```

### 👁️ Read the figure

**The middle column is the story.** Six of its eight cells are red. Read down them:

- `samples` went in as the integer **150** and came back as the string **`'150'`**.
  So did `features`, `test_fraction` and `shuffled` — an `int`, a `float` and a
  `bool` all arrive back as `str`, because a text file stores characters and
  nothing else. The grey column and the red column hold the *same characters*;
  only the type under each value changed.
- `collected_at` went in as **`'09:30'`** and came back as **`'09'`**. `class_ratio`
  went in as **`'3:1'`** and came back as **`'3'`**. Splitting on the first colon and
  taking element `[1]` threw away everything after the second colon. The value did
  not fail to load — it loaded, shortened, and looked perfectly reasonable.
- Only `dataset` and `gene` are green, and only because they were plain text with
  no colon in them. **Two of eight.**

**Right panel — and not one exception was raised.** The count printed in that title
is **0**. Every corruption above happened on a parser that ran to completion and
reported success. `json`, given exactly the same eight records, returns all eight
with their original types, which is why the whole right-hand column is green.

**What this proves:** `line.split(':')` is not a parser, it is a coincidence that
usually works. The failure mode is silence — the exact failure mode of the
genomics files at the top of this notebook, where nothing crashed and one paper in
five was wrong. **Where the analogy stops:** `'SEPT2'` came back intact here,
because a plain text file does not try to be helpful. It took a program that
*guesses* types — a spreadsheet reading a cell — to turn it into a date. Every
tool in this diploma that reads a file makes that same guess somewhere; your job
is to know where, and to declare the type before the tool invents one.

## ⚠️ Where this breaks

Everything here works on five clean records. Here is where each piece stops working.

- **Splitting text by hand is not parsing.** `line.split(': ')` has no concept of
  quoting, escaping, embedded separators, or character encoding, and it fails
  *silently* — you get a wrong value, not an exception. **Use instead:** `csv` for
  delimited files, `json` for structured records, and pandas for tables.
- **Types are guessed, never declared.** Read a file and everything is a string
  until someone converts it. That conversion is where `'150'` becomes 150 and
  where `'2-Sep'` should have raised an alarm and did not. Declare and check types
  at the boundary of your program, not deep inside it.
- **Dictionaries do not enforce a schema.** `student_data['major'] = 'AI'` adds a
  key with no complaint, and a typo — `'grads'` for `'grades'` — creates a second
  key rather than an error. Fine for five records; unmanageable for five thousand.
  **Use instead:** a dataclass, a `TypedDict`, or a DataFrame with named columns.
- **`model_params` as a plain dict has no validation.** Nothing stops
  `learning_rate = -0.01` or `epochs = 'many'`. Unit 3 will show you what a bad
  learning rate does to training; the container will not.
- **Reading a whole file with `f.read()` assumes it fits in memory.** For a log or
  a dataset larger than RAM, iterate line by line — or use a library that streams.
- **The assumption that must hold for this whole notebook:** the data is small,
  clean, complete and produced by you. The Titanic file two notebooks from now
  is none of those things: 177 of its 891 ages were never written down.
