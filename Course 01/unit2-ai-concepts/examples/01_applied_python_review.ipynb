## 🎯 One in five genetics papers, corrupted by a spreadsheet

In 2016 Mark Ziemann, Yotam Eren and Assam El-Osta published *Gene name errors
are widespread in the scientific literature* in **Genome Biology**. They scanned
the supplementary Excel files of leading genomics journals and found that roughly
**one paper in five** contained gene names silently converted into something else:
Excel's default settings turn the gene symbol `SEPT2` into the date *2-Sep*, and
`MARCH1` into *1-Mar*. Nobody noticed, because nothing failed — the file opened,
the column looked full, the analysis ran.

The fix, four years later, is the part that should stay with you. In **August
2020** the HUGO Gene Nomenclature Committee — the body that names human genes —
gave up and renamed about **27 genes** so that a spreadsheet would stop
misreading them. `MARCH1` became `MARCHF1`; `SEPT1` became `SEPTIN1`. Human
biology was renamed to accommodate a data-handling default.

### Why this is a lesson about lists, dictionaries and files

Every step in that failure is a step in this notebook. A value was read from a
file. Its **type** was guessed instead of declared. It was stored in a container
that did not care what it held. Nothing raised an error. When you write
`data['grades']` or `float(line.split(':')[1])` below, you are making the same
three decisions — about types, containers and parsing — that decided whether a
genomics paper was correct. Doing that in code, where the parsing rule is visible
and reviewable, is the whole argument for this notebook.

```python
# Python Lists - Essential for AI data processing
print("=" * 60)
print("PYTHON LISTS REVIEW")
print("=" * 60)

# Creating lists
data_points = [1, 2, 3, 4, 5]
features = ['age', 'income', 'education']
print(f"\nData points: {data_points}")
print(f"Features: {features}")

# List operations
data_points.append(6)
print(f"\nAfter append: {data_points}")

# List comprehension (common in AI)
squared = [x**2 for x in data_points]
print(f"Squared values: {squared}")

# Slicing
print(f"First 3 elements: {data_points[:3]}")
print(f"Last 2 elements: {data_points[-2:]}")

print("\n✅ Lists reviewed!")
```

```python
# Python Dictionaries - Key-value pairs for AI data
print("=" * 60)
print("PYTHON DICTIONARIES REVIEW")
print("=" * 60)

# Creating dictionaries
student_data = {
    'name': 'Alice', 'age': 20,
    'grades': [85, 90, 88]
}

model_params = {
    'learning_rate': 0.01,
    'epochs': 100,
    'batch_size': 32
}

print(f"\nStudent data: {student_data}")
print(f"Model parameters: {model_params}")

# Dictionary operations
print(f"\nStudent name: {student_data['name']}")
print(f"Learning rate: {model_params.get('learning_rate', 'Not set')}")

# Adding/updating
student_data['major'] = 'AI'
print(f"\nUpdated student data: {student_data}")

# Iterating
print("\nModel parameters:")
for key, value in model_params.items():
    print(f"  {key}: {value}")

print("\n✅ Dictionaries reviewed!")
```

```python
# File Handling - Reading and writing data
print("=" * 60)
print("FILE HANDLING REVIEW")
print("=" * 60)

# Writing to a file
data_to_save = {
    'dataset': 'iris', 'samples': 150,
    'features': 4
}

# Write JSON-like data
with open('temp_data.txt', 'w') as f:
    for key, value in data_to_save.items():
        f.write(f"{key}: {value}\n")

print("✅ Data written to file")

# Reading from a file
with open('temp_data.txt', 'r') as f:
    content = f.read()
    print(f"\nFile content:\n{content}")

# Reading line by line
print("\nReading line by line:")
with open('temp_data.txt', 'r') as f:
    for line in f:
        print(f"  {line.strip()}")

# Cleanup
import os
if os.path.exists('temp_data.txt'):
    os.remove('temp_data.txt')
    print("\n✅ Temporary file cleaned up")

print("\n✅ File handling reviewed!")
```

## ⚠️ Where this breaks

Everything here works on five clean records. Here is where each piece stops working.

- **Splitting text by hand is not parsing.** `line.split(': ')` has no concept of
  quoting, escaping, embedded separators, or character encoding, and it fails
  *silently* — you get a wrong value, not an exception. **Use instead:** `csv` for
  delimited files, `json` for structured records, and pandas for tables.
- **Types are guessed, never declared.** Read a file and everything is a string
  until someone converts it. That conversion is where `'150'` becomes 150 and
  where `'2-Sep'` should have raised an alarm and did not. Declare and check types
  at the boundary of your program, not deep inside it.
- **Dictionaries do not enforce a schema.** `student_data['major'] = 'AI'` adds a
  key with no complaint, and a typo — `'grads'` for `'grades'` — creates a second
  key rather than an error. Fine for five records; unmanageable for five thousand.
  **Use instead:** a dataclass, a `TypedDict`, or a DataFrame with named columns.
- **`model_params` as a plain dict has no validation.** Nothing stops
  `learning_rate = -0.01` or `epochs = 'many'`. Unit 3 will show you what a bad
  learning rate does to training; the container will not.
- **Reading a whole file with `f.read()` assumes it fits in memory.** For a log or
  a dataset larger than RAM, iterate line by line — or use a library that streams.
- **The assumption that must hold for this whole notebook:** the data is small,
  clean, complete and produced by you. The Titanic file two notebooks from now
  is none of those things: 177 of its 891 ages were never written down.
