## 📌 The case: a governance document that actually killed a shipping product

Most corporate AI principles cost nothing, because nothing is ever refused under them. Here is
a documented exception.

In **June 2022** Microsoft published version 2 of its **Responsible AI Standard**, an internal
document with binding requirements rather than aspirations. On the same day it announced what
the standard required of an existing, revenue-generating product: **Azure Face would retire
its capabilities for inferring emotional states and identity attributes** — emotion, gender,
age, smile, facial hair, hair and makeup. New customers lost access to those attributes on
**21 June 2022**; existing customers had until **30 June 2023**.

The reasons Microsoft gave map onto this entire course: **no scientific consensus on the
definition of "emotions"**, no evidence that the inferences **generalise across use cases,
regions and demographics** (Unit 2), and heightened **privacy** concerns (Unit 3). The
capabilities were retained only for controlled accessibility scenarios such as Seeing AI.

That is what a governance framework looks like when it has teeth: a written standard, applied
to a live product, with a published retirement date and a customer migration cost the company
absorbed.

**The WHY — what goes wrong without governance.** Units 1–4 gave you techniques. Governance
decides **who has to use them, before what gate, and who can say no**. Without it, every
fairness audit is voluntary, every model card is somebody's side project, and every risky
launch is decided by whoever is most senior in the room that afternoon. You are about to score
eight concrete practices from 0 to 3 — because "we have governance" is untestable, and a
number is not.

## Part 1: Governance You Can Draw and Score

An AI governance framework = **structures** (who decides), **processes** (how projects get
reviewed), and **evidence** (what gets documented). We encode a reference structure, then
run a *maturity self-assessment* an organization could actually use.

```python
# Why structure first: governance is bodies with authority, not documents -
# each body here exists to make a specific decision nobody else can make.

# Step 1: A reference governance structure, as data
import pandas as pd

# One entry per governance body: who sits on it, what it alone decides, and how often.
governance_structure = {
    'AI Ethics Board': {
        'members': 'Legal, engineering, domain experts, external advisor',
        'decides': 'Go/no-go for high-risk AI projects; policy exceptions',
        'meets': 'Monthly + on-call for escalations',
    },
    'Model Review Committee': {
        'members': 'ML leads, QA, security',
        'decides': 'Technical sign-off: fairness/robustness evidence adequate?',
        'meets': 'Per release',
    },
    'Data Protection Officer (DPO)': {
        'members': 'Privacy specialist (GDPR-mandated role)',
        'decides': 'DPIA outcomes; data-processing approvals',
        'meets': 'Continuous',
    },
    'Incident Response Team': {
        'members': 'On-call engineers + comms + legal',
        'decides': 'Rollbacks, disclosures after AI incidents',
        'meets': 'On incident',
    },
}
print("REFERENCE AI GOVERNANCE STRUCTURE")
print("=" * 90)
for body, info in governance_structure.items():
    print(f"\n{body}")
    for k, v in info.items():
        print(f"   {k:>8}: {v}")
```

```python
# WHAT this cell draws: the same four bodies you just printed, but placed on the
# path a project actually travels - so you can see WHERE each one can say no.
# WHY: the dictionary above lists bodies; it does not show the order, and order is
# the whole of governance. A body that sits after deployment cannot stop a launch.
# All the text inside the boxes is read straight out of governance_structure, so
# the picture cannot drift away from the data.
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch
import textwrap

def sub(body, w=27):
    """Wrap a body's 'decides' and 'meets' entries to fit inside a box."""
    info = governance_structure[body]
    return (textwrap.fill(info['decides'], w) + "\n"
            + textwrap.fill(f"({info['meets'].lower()})", w))

fig, ax = plt.subplots(figsize=(12.8, 4.7))

def box(x, y, w, h, title, s, fc, ec, tfs=9.4):
    ax.add_patch(FancyBboxPatch((x, y), w, h,
                                boxstyle="round,pad=0.02,rounding_size=0.06",
                                facecolor=fc, edgecolor=ec, lw=1.4))
    ax.text(x + w / 2, y + h - 0.10, title, ha="center", va="top",
            fontsize=tfs, fontweight="bold", color="0.12")
    ax.text(x + w / 2, y + h - 0.36, s, ha="center", va="top",
            fontsize=7.6, color="0.25")

def arrow(x0, y0, x1, y1, color="0.3", lw=1.6):
    ax.annotate("", xy=(x1, y1), xytext=(x0, y0),
                arrowprops=dict(arrowstyle="-|>", color=color, lw=lw,
                                shrinkA=0, shrinkB=0))

W, H, Y = 2.25, 1.30, 2.05
box(0.05, Y, 1.6, H, "New AI project", "risk tier assigned\n(Notebook 01)", "#EDEDED", "0.6")
box(2.05, Y, W, H, "DPO", sub('Data Protection Officer (DPO)'), "#DCE6F2", "#4C72B0")
box(4.70, Y, W, H, "Model Review Committee", sub('Model Review Committee'),
    "#DCEDDC", "#55A868", tfs=8.8)
box(7.35, Y, W, H, "AI Ethics Board", sub('AI Ethics Board'), "#FAE9C8", "#D9A441")
box(10.00, Y, 1.6, H, "Deployed", "in production", "#EDEDED", "0.6")
for x0, x1 in [(1.65, 2.05), (4.30, 4.70), (6.95, 7.35), (9.60, 10.00)]:
    arrow(x0, Y + H / 2, x1, Y + H / 2)

# The one body that acts AFTER deployment.
box(7.35, 0.15, W, 1.10, "Incident Response Team", sub('Incident Response Team'),
    "#F6DDDD", "#C44E52", tfs=8.8)
arrow(10.80, Y, 9.60, 1.25, color="#C44E52")
ax.text(10.45, 1.68, "incident", fontsize=8.2, color="#C44E52", ha="left")

# The arrow that is DASHED because it does not exist in this structure.
ax.annotate("", xy=(8.45, Y + H + 0.02), xytext=(10.80, Y + H + 0.02),
            arrowprops=dict(arrowstyle="-|>", color="#B0656A", ls="--", lw=1.5,
                            connectionstyle="arc3,rad=0.32", shrinkA=0, shrinkB=0))
ax.text(9.62, Y + H + 0.75,
        "retire a shipped, revenue-earning feature?\n"
        "no body in this structure is assigned that decision",
        fontsize=8.6, color="#7A2E2E", ha="center", va="bottom", style="italic")
ax.text(0.05, 1.62,
        "Every arrow is a gate: the project cannot pass without that body's decision.",
        fontsize=9, color="0.35")
ax.set_xlim(-0.1, 11.9)
ax.set_ylim(0.05, 4.75)
ax.axis("off")
ax.set_title("Governance is bodies with authority, placed on the release path",
             fontsize=12.5)
fig.tight_layout()
plt.show()
```

**What to see in this figure.** Four solid arrows, left to right, each one a gate: a project
goes **DPO → Model Review Committee → AI Ethics Board → Deployed**, and it cannot pass a gate
without that body's decision. The text inside each box is the body's own `decides` and `meets`
entry from the dictionary above, so the picture and the data say the same thing.

Two arrows are worth arguing about. The red one runs *out* of "Deployed" to the **Incident
Response Team** — the only body in this structure that acts after launch, and only once
something has already gone wrong. The **dashed** arrow runs backwards from "Deployed" to the
Ethics Board and is dashed because **it is not in the structure**: the board's remit is
"go/no-go for high-risk AI projects", which is a decision taken *before* release. Microsoft
retiring Azure Face's emotion inference was exactly that missing arrow being drawn. When you
design your own structure, decide who owns it — or accept that nothing shipped will ever be
withdrawn on ethical grounds.



```python
# Why maturity scoring: 'we have governance' is untestable - scoring eight
# concrete practices 0-3 turns governance into something you can measure and improve.

# Step 2: Governance maturity self-assessment (score 0-3 per practice)

# The eight practices scored here map to earlier units: fairness (U2),
# audit trails and model cards (U4), incident playbooks (this unit).
practices = {
    'AI project inventory exists and is current': 3,
    'Risk classification applied to every project': 2,
    'Fairness evaluation required before deployment': 2,
    'Model cards / documentation for every model': 1,
    'Audit trails for automated decisions': 1,
    'HITL defined for high-stakes decisions': 2,
    'Incident response playbook for AI failures': 0,
    'Regular ethics training for AI staff': 1,
}
LEVELS = {0: 'Absent', 1: 'Ad-hoc', 2: 'Defined', 3: 'Managed/measured'}

print("GOVERNANCE MATURITY SELF-ASSESSMENT (example organization)")
print("=" * 80)
for p, s in practices.items():
    print(f"  [{s}] {LEVELS[s]:<16} {p}")
# Aggregate into an overall maturity score, then translate it into a verdict.
total = sum(practices.values())
max_total = 3 * len(practices)
pct = total / max_total
print(f"\nOverall maturity: {total}/{max_total} ({pct:.0%})")
if pct < 0.4:
    verdict = 'Initial - governance is mostly aspirational'
elif pct < 0.7:
    verdict = 'Developing - structures exist, evidence is inconsistent'
else:
    verdict = 'Established - governance operates and is measured'
print(f"Verdict: {verdict}")
# Rank the weakest practices - the actionable output of the whole assessment.
worst = sorted(practices, key=practices.get)[:2]
print("\nPriority gaps to fix first:")
for p in worst:
    print(f"  - {p} (score {practices[p]})")
print("\nThis is how governance frameworks become actionable: score, rank gaps,")
print("assign owners (RACI - Unit 4), re-score next quarter.")
```

```python
# WHAT this cell draws: the eight scores you just printed, as one bar chart, with
# the two priority gaps flagged.
# WHY: printed rows all look equally important. Bars show at a glance that the
# organisation is stuck between "ad-hoc" and "defined" - and that the practice
# with the most to prove (incident response) is the one at zero.
import matplotlib.pyplot as plt

COLORS = {0: '#C44E52', 1: '#DD8452', 2: '#E8C15C', 3: '#55A868'}

# Reuse the SAME dict and the SAME "worst two" rule as the cell above.
items = sorted(practices.items(), key=lambda kv: kv[1])   # weakest at the bottom
names = [k for k, _ in items]
scores = [v for _, v in items]

fig, ax = plt.subplots(figsize=(11.6, 4.6))
y = list(range(len(items)))
ax.barh(y, [3] * len(items), color='0.92', height=0.6)     # the distance still to go
ax.barh(y, scores, color=[COLORS[s] for s in scores], height=0.6)
for i, (n, s) in enumerate(items):
    ax.text(s + 0.06, i, LEVELS[s], va='center', fontsize=8.6, color='0.25')
    if n in worst:                       # 'worst' was computed in the cell above
        ax.text(3.06, i, 'PRIORITY GAP', va='center', fontsize=8.4,
                color='#8B1A1E', fontweight='bold')
ax.set_yticks(y)
ax.set_yticklabels(names, fontsize=9)
ax.set_xlim(0, 3.75)
ax.set_ylim(-0.6, len(items) - 0.4)
ax.set_xticks([0, 1, 2, 3])
ax.set_xticklabels(['0\nAbsent', '1\nAd-hoc', '2\nDefined', '3\nManaged/\nmeasured'],
                   fontsize=8.5)
ax.set_xlabel("Maturity score for the practice")
ax.set_title(f"Governance maturity: {total}/{max_total} = {pct:.0%}  -  '{verdict}'",
             fontsize=11.5)
ax.grid(axis='x', alpha=0.3)
ax.set_axisbelow(True)
for s in ('top', 'right'):
    ax.spines[s].set_visible(False)
fig.tight_layout()
plt.show()
```

**What to see in this figure.** Each bar is one practice; the pale grey strip behind it is the
distance still to travel to "managed/measured". Only **one** bar reaches 3 (the project
inventory). Three sit at 2, three at 1, and one — **incident response playbook** — has no
coloured bar at all, because it scores 0. That is the picture of "structures exist, evidence is
inconsistent": the organisation can tell you what it has, and cannot tell you how well it works.

One honest wrinkle you should notice and the print-out hides: **three practices tie at score 1**
(model cards, audit trails, ethics training), but only one of them is flagged PRIORITY GAP. The
ranking code takes the first two after sorting, and ties are broken by the order the dictionary
happens to be written in — not by importance. If a real budget decision hung on this chart, that
tie-break would need a reason, and the chart is where you can see that it does not have one.

## ⚠️ Where this breaks

- **A maturity score measures artefacts, not authority.** Every practice here asks whether a
  document or process exists. None asks whether anyone has ever used them to stop something.
  The Dutch tax administration (Unit 4) would have scored respectably on most of these
  practices while wrongly accusing about 26,000 families.
- **Self-assessment is scored by the people being scored.** The person filling this in has an
  interest in the result, and no definition of "Defined" versus "Managed/measured" survives
  that pressure unaided. If the score is used for anything consequential, it needs external
  validation and evidence attached to each item — which is what the AI Act's conformity
  assessment regime is for.
- **Governance bodies without a budget or a veto are advisory committees.** An ethics board
  that meets monthly, has no engineering time allocated to it and cannot delay a release will
  produce minutes and nothing else. Before you count a body, ask: what has it refused, and
  what happened next?
- **Process scales badly and blocks badly.** Governance that requires board review of every
  model either gets bypassed by teams under deadline pressure, or slows the organisation until
  it is dismantled. Tier the process to the risk (Notebook 01) or it will not survive contact
  with delivery.
- **These frameworks are voluntary until they are not.** NIST's AI RMF is explicitly voluntary;
  the OECD Principles are non-binding. Their value is as a **structure** you can point to, and
  as preparation for the binding regimes — not as compliance in themselves.

**The cheaper alternative for a small team:** skip the four-body structure. A **single named
owner per model** plus a **one-page risk classification at project start** plus a **standing
rule that high-risk systems require documented per-group evaluation before launch** delivers
most of the protection. Add bodies when the volume of decisions, not the size of the
organisation, makes one person impossible.


## 🔭 State of the field (2026)

Your eight-practice self-assessment measures whether artefacts exist. What moved between 2024 and
2026 is *what counts as an artefact*. NIST's AI RMF — the voluntary framework your maturity items
lean on — gained a cross-sectoral companion profile for generative systems, **NIST AI 600-1**,
published 26 July 2024, so "we follow the AI RMF" now has specific generative-AI subject matter
attached to it; NIST is separately drafting SP 800-53 **Control Overlays for Securing AI Systems**
(COSAiS: concept paper 14 August 2025, annotated outline 8 January 2026) to bind concrete security
controls to AI components such as training data and model weights. At the frontier, risk-management
frameworks are now published as *versioned technical reports* with named risk dimensions and
experimental evaluations attached rather than as principles — Liu et al. (2026) is v1.5 of one, and
it is a document a student can read and criticise, which is roughly what the top of your 0–3 scale
looks like when it is done in public.

**Why keep scoring practices.** None of those documents answers the question your four-body
structure exists to ask: *who here has the authority to refuse a launch, and what happens when they
use it?* Profiles, overlays and published frameworks are content you pour into governance bodies;
they are not a substitute for having bodies, owners and a gate. The Azure Face retirement above was
not caused by a standard — it was caused by an internal standard that somebody was empowered to
enforce against a product that was already earning money.
