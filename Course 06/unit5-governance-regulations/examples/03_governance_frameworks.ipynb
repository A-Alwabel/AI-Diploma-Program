## 📌 The case: a governance document that actually killed a shipping product

Most corporate AI principles cost nothing, because nothing is ever refused under them. Here is
a documented exception.

In **June 2022** Microsoft published version 2 of its **Responsible AI Standard**, an internal
document with binding requirements rather than aspirations. On the same day it announced what
the standard required of an existing, revenue-generating product: **Azure Face would retire
its capabilities for inferring emotional states and identity attributes** — emotion, gender,
age, smile, facial hair, hair and makeup. New customers lost access to those attributes on
**21 June 2022**; existing customers had until **30 June 2023**.

The reasons Microsoft gave map onto this entire course: **no scientific consensus on the
definition of "emotions"**, no evidence that the inferences **generalise across use cases,
regions and demographics** (Unit 2), and heightened **privacy** concerns (Unit 3). The
capabilities were retained only for controlled accessibility scenarios such as Seeing AI.

That is what a governance framework looks like when it has teeth: a written standard, applied
to a live product, with a published retirement date and a customer migration cost the company
absorbed.

**The WHY — what goes wrong without governance.** Units 1–4 gave you techniques. Governance
decides **who has to use them, before what gate, and who can say no**. Without it, every
fairness audit is voluntary, every model card is somebody's side project, and every risky
launch is decided by whoever is most senior in the room that afternoon. You are about to score
eight concrete practices from 0 to 3 — because "we have governance" is untestable, and a
number is not.

## Part 1: Governance You Can Draw and Score

An AI governance framework = **structures** (who decides), **processes** (how projects get
reviewed), and **evidence** (what gets documented). We encode a reference structure, then
run a *maturity self-assessment* an organization could actually use.

```python
# Why structure first: governance is bodies with authority, not documents -
# each body here exists to make a specific decision nobody else can make.

# Step 1: A reference governance structure, as data
import pandas as pd

# One entry per governance body: who sits on it, what it alone decides, and how often.
governance_structure = {
    'AI Ethics Board': {
        'members': 'Legal, engineering, domain experts, external advisor',
        'decides': 'Go/no-go for high-risk AI projects; policy exceptions',
        'meets': 'Monthly + on-call for escalations',
    },
    'Model Review Committee': {
        'members': 'ML leads, QA, security',
        'decides': 'Technical sign-off: fairness/robustness evidence adequate?',
        'meets': 'Per release',
    },
    'Data Protection Officer (DPO)': {
        'members': 'Privacy specialist (GDPR-mandated role)',
        'decides': 'DPIA outcomes; data-processing approvals',
        'meets': 'Continuous',
    },
    'Incident Response Team': {
        'members': 'On-call engineers + comms + legal',
        'decides': 'Rollbacks, disclosures after AI incidents',
        'meets': 'On incident',
    },
}
print("REFERENCE AI GOVERNANCE STRUCTURE")
print("=" * 90)
for body, info in governance_structure.items():
    print(f"\n{body}")
    for k, v in info.items():
        print(f"   {k:>8}: {v}")
```

```python
# Why maturity scoring: 'we have governance' is untestable - scoring eight
# concrete practices 0-3 turns governance into something you can measure and improve.

# Step 2: Governance maturity self-assessment (score 0-3 per practice)

# The eight practices scored here map to earlier units: fairness (U2),
# audit trails and model cards (U4), incident playbooks (this unit).
practices = {
    'AI project inventory exists and is current': 3,
    'Risk classification applied to every project': 2,
    'Fairness evaluation required before deployment': 2,
    'Model cards / documentation for every model': 1,
    'Audit trails for automated decisions': 1,
    'HITL defined for high-stakes decisions': 2,
    'Incident response playbook for AI failures': 0,
    'Regular ethics training for AI staff': 1,
}
LEVELS = {0: 'Absent', 1: 'Ad-hoc', 2: 'Defined', 3: 'Managed/measured'}

print("GOVERNANCE MATURITY SELF-ASSESSMENT (example organization)")
print("=" * 80)
for p, s in practices.items():
    print(f"  [{s}] {LEVELS[s]:<16} {p}")
# Aggregate into an overall maturity score, then translate it into a verdict.
total = sum(practices.values())
max_total = 3 * len(practices)
pct = total / max_total
print(f"\nOverall maturity: {total}/{max_total} ({pct:.0%})")
if pct < 0.4:
    verdict = 'Initial - governance is mostly aspirational'
elif pct < 0.7:
    verdict = 'Developing - structures exist, evidence is inconsistent'
else:
    verdict = 'Established - governance operates and is measured'
print(f"Verdict: {verdict}")
# Rank the weakest practices - the actionable output of the whole assessment.
worst = sorted(practices, key=practices.get)[:2]
print("\nPriority gaps to fix first:")
for p in worst:
    print(f"  - {p} (score {practices[p]})")
print("\nThis is how governance frameworks become actionable: score, rank gaps,")
print("assign owners (RACI - Unit 4), re-score next quarter.")
```

## 💬 Discuss

Your self-assessment scored the example organisation **12 / 24 (50 %) — "Developing:
structures exist, evidence is inconsistent"**, with two priority gaps: **incident response
playbook for AI failures (score 0)** and **model cards for every model (score 1)**.

1. **Fix one gap. Which?** You have budget for exactly one. The incident playbook scores 0 and
   matters only when something has already gone wrong; model cards score 1 and are a legal
   obligation for high-risk systems from 2 August 2026. Choose, and justify it to a board that
   is measured on shipping. Then say what evidence would prove, twelve months later, that you
   chose correctly.
2. **Would your board have retired Azure Face?** Microsoft's standard forced the removal of
   revenue-generating features from a live product. Look at your own reference structure: the
   AI Ethics Board "decides go/no-go for high-risk AI projects". Does it have the authority to
   cancel something already **shipped and earning**? Who does it report to, and what stops
   that person overruling it? If nothing does, what is the score really measuring?
3. **Does a maturity score measure governance or paperwork?** All eight practices are
   artefacts: an inventory exists, a classification is applied, documentation is current. An
   organisation could score 24/24 and never have refused a single launch. Design the ninth
   practice — the one that measures whether governance ever changed an outcome — and be honest
   about whether you would want it scored publicly.

## ⚠️ Where this breaks

- **A maturity score measures artefacts, not authority.** Every practice here asks whether a
  document or process exists. None asks whether anyone has ever used them to stop something.
  The Dutch tax administration (Unit 4) would have scored respectably on most of these
  practices while wrongly accusing about 26,000 families.
- **Self-assessment is scored by the people being scored.** The person filling this in has an
  interest in the result, and no definition of "Defined" versus "Managed/measured" survives
  that pressure unaided. If the score is used for anything consequential, it needs external
  validation and evidence attached to each item — which is what the AI Act's conformity
  assessment regime is for.
- **Governance bodies without a budget or a veto are advisory committees.** An ethics board
  that meets monthly, has no engineering time allocated to it and cannot delay a release will
  produce minutes and nothing else. Before you count a body, ask: what has it refused, and
  what happened next?
- **Process scales badly and blocks badly.** Governance that requires board review of every
  model either gets bypassed by teams under deadline pressure, or slows the organisation until
  it is dismantled. Tier the process to the risk (Notebook 01) or it will not survive contact
  with delivery.
- **These frameworks are voluntary until they are not.** NIST's AI RMF is explicitly voluntary;
  the OECD Principles are non-binding. Their value is as a **structure** you can point to, and
  as preparation for the binding regimes — not as compliance in themselves.

**The cheaper alternative for a small team:** skip the four-body structure. A **single named
owner per model** plus a **one-page risk classification at project start** plus a **standing
rule that high-risk systems require documented per-group evaluation before launch** delivers
most of the protection. Add bodies when the volume of decisions, not the size of the
organisation, makes one person impossible.
