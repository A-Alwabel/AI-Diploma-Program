## Part 1: Working Through Legal Scenarios

Legal challenges become concrete when you analyze scenarios: who could be liable, under
what theory, and what evidence decides it. (Teaching analysis - not legal advice.)

```python
# Why scenarios: AI law is unsettled - working through concrete disputes shows
# WHO can be liable and WHAT evidence decides it, even before courts settle the rules.

# Step 1: Scenario analysis - liability, IP, and cross-border questions

# Four recurring dispute types: each records the facts, the liability candidates,
# the deciding evidence, and the question courts have not resolved yet.
scenarios = [
    {'name': 'Autonomous vehicle accident',
     'facts': 'AV in autopilot mode injures a pedestrian',
     'candidate_liable': ['Manufacturer (product liability)',
                          'Software supplier (component defect)',
                          'Owner/operator (negligent supervision)'],
     'key_evidence': 'Logs: what did the system perceive and decide? (audit trail!)',
     'open_question': 'Is a learning system a "defective product" when it errs?'},
    {'name': 'Discriminatory hiring algorithm',
     'facts': 'Qualified applicants of one group systematically rejected',
     'candidate_liable': ['Employer (disparate impact under employment law)',
                          'Vendor (misrepresentation of fairness testing)'],
     'key_evidence': 'Fairness metrics by group (Unit 2!), validation records',
     'open_question': 'How much vendor due diligence shields the employer?'},
    {'name': 'GenAI trained on copyrighted works',
     'facts': 'Model reproduces protected content in outputs',
     'candidate_liable': ['Model developer (training-data infringement)',
                          'Deploying company (distribution of outputs)',
                          'User (commercial use of infringing output)'],
     'key_evidence': 'Training-data provenance records',
     'open_question': 'Is training fair use? Jurisdictions are split (litigation ongoing)'},
    {'name': 'Cross-border data processing',
     'facts': 'EU user data processed by non-EU AI service',
     'candidate_liable': ['Data controller (GDPR transfer rules)',
                          'Processor (contract violations)'],
     'key_evidence': 'Transfer mechanism (SCCs/adequacy), processing records',
     'open_question': 'Which law wins when requirements directly conflict?'},
]

# Walk through each scenario - watch how often 'key evidence' is an artifact
# this course taught you to produce.
print("LEGAL CHALLENGE SCENARIOS")
print("=" * 90)
for s in scenarios:
    print(f"\n📌 {s['name']}")
    print(f"   Facts: {s['facts']}")
    print("   Who could be liable:")
    for p in s['candidate_liable']:
        print(f"     - {p}")
    print(f"   Key evidence: {s['key_evidence']}")
    print(f"   Open legal question: {s['open_question']}")

print("\n" + "=" * 90)
print("The recurring lesson: in EVERY scenario, the evidence that decides the")
print("case is exactly what this course taught you to produce - audit trails,")
print("fairness metrics, documentation, and provenance records.")
```
