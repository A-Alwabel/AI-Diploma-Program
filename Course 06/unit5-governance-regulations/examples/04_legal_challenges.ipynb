## 📌 The case: a court held the *vendor* liable as an agent of its customers

One of the scenarios you are about to analyse asks: **how much vendor due diligence shields
the employer?** A US court has now begun answering the reverse question — whether the vendor
can hide behind the employer.

**Mobley v. Workday, Inc.** (N.D. Cal., case 3:23-cv-00770). The plaintiff alleged that
Workday's applicant-screening tools rejected him repeatedly on grounds including age. Workday
argued the obvious defence: it is a **software provider**, not an employer, and employment
discrimination law binds employers.

The court rejected that framing, holding that a vendor performing core hiring functions —
screening, scoring, ranking and recommending candidates — on the employer's behalf can be
treated as the employer's **agent**, and therefore reached directly by anti-discrimination
statutes. On **16 May 2025** the court granted **preliminary certification of a nationwide
collective action** under the Age Discrimination in Employment Act, covering applicants aged
40 and over screened through the platform since **September 2020**. Workday's own filings
indicated that around **1.1 billion applications** were rejected using its tools in the
relevant period, so the potential collective is enormous.

Whatever the final outcome, the doctrinal move has already happened: "we only supply the
model" is no longer a complete answer, and the discovery process now reaches the vendor's
training data, validation records and fairness testing.

**The WHY — what goes wrong without thinking about liability.** Every artefact this course
taught you to build — per-group metrics (Unit 2), audit trails and explanations (Unit 4),
model cards (Unit 4) — is **evidence**. It is what you will have, or conspicuously lack, when
somebody asks a court what your system did and why. Deciding what to record is therefore not
only an engineering choice; it is the choice of what you will be able to prove.

---
> **⚖️ A word-sense warning.** "Agent" in this case means the **legal** sense — one party acting on
> another's behalf, so the principal is answerable for what the agent does. It is unrelated to the
> AI sense you learned in Course 01 (AIAT 111), Unit 1, lesson 04, where an agent is anything that
> perceives an environment through sensors and acts on it through actuators (Russell & Norvig, *AIMA*
> 4e, Ch. 2), and unrelated again to the reinforcement-learning agent of Course 09 (AIAT 123). The
> Mobley court is not making a claim about software architecture; it is making a claim about who
> answers for a decision. See `docs/GLOSSARY.md`.

## Part 1: Working Through Legal Scenarios

Legal challenges become concrete when you analyze scenarios: who could be liable, under
what theory, and what evidence decides it. (Teaching analysis - not legal advice.)

```python
# Why scenarios: AI law is unsettled - working through concrete disputes shows
# WHO can be liable and WHAT evidence decides it, even before courts settle the rules.

# Step 1: Scenario analysis - liability, IP, and cross-border questions

# Four recurring dispute types: each records the facts, the liability candidates,
# the deciding evidence, and the question courts have not resolved yet.
scenarios = [
    {'name': 'Autonomous vehicle accident',
     'facts': 'AV in autopilot mode injures a pedestrian',
     'candidate_liable': ['Manufacturer (product liability)',
                          'Software supplier (component defect)',
                          'Owner/operator (negligent supervision)'],
     'key_evidence': 'Logs: what did the system perceive and decide? (audit trail!)',
     'open_question': 'Is a learning system a "defective product" when it errs?'},
    {'name': 'Discriminatory hiring algorithm',
     'facts': 'Qualified applicants of one group systematically rejected',
     'candidate_liable': ['Employer (disparate impact under employment law)',
                          'Vendor (misrepresentation of fairness testing)'],
     'key_evidence': 'Fairness metrics by group (Unit 2!), validation records',
     'open_question': 'How much vendor due diligence shields the employer?'},
    {'name': 'GenAI trained on copyrighted works',
     'facts': 'Model reproduces protected content in outputs',
     'candidate_liable': ['Model developer (training-data infringement)',
                          'Deploying company (distribution of outputs)',
                          'User (commercial use of infringing output)'],
     'key_evidence': 'Training-data provenance records',
     'open_question': 'Is training fair use? Jurisdictions are split (litigation ongoing)'},
    {'name': 'Cross-border data processing',
     'facts': 'EU user data processed by non-EU AI service',
     'candidate_liable': ['Data controller (GDPR transfer rules)',
                          'Processor (contract violations)'],
     'key_evidence': 'Transfer mechanism (SCCs/adequacy), processing records',
     'open_question': 'Which law wins when requirements directly conflict?'},
]

# Walk through each scenario - watch how often 'key evidence' is an artifact
# this course taught you to produce.
print("LEGAL CHALLENGE SCENARIOS")
print("=" * 90)
for s in scenarios:
    print(f"\n📌 {s['name']}")
    print(f"   Facts: {s['facts']}")
    print("   Who could be liable:")
    for p in s['candidate_liable']:
        print(f"     - {p}")
    print(f"   Key evidence: {s['key_evidence']}")
    print(f"   Open legal question: {s['open_question']}")

print("\n" + "=" * 90)
print("The recurring lesson: in EVERY scenario, the evidence that decides the")
print("case is exactly what this course taught you to produce - audit trails,")
print("fairness metrics, documentation, and provenance records.")
```

```python
# WHAT this cell draws: the four scenarios you just printed, joined by arrows to
# the evidence that decides each one, with the unit that taught you to produce it.
# WHY: the closing line above claims the deciding evidence is "exactly what this
# course taught you to produce". Arrows make that checkable - if an arrow landed
# on something with no unit tag, the claim would be visibly false.
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch
import textwrap
import collections

# Left column: the dispute. Right column: the artefact, and where you built it.
SC = [s['name'] for s in scenarios]
EV = [("Decision logs / audit trail", "Unit 4, Nb. 04"),
      ("Fairness metrics by group", "Unit 2"),
      ("Validation records + model card", "Unit 4, Nb. 06"),
      ("Training-data provenance records", "Unit 3, Nb. 05"),
      ("Processing + transfer records (SCCs)", "Unit 3, Nb. 04")]
# (scenario index, evidence index) - read off the 'key_evidence' field above.
LINKS = [(0, 0), (1, 1), (1, 2), (2, 3), (3, 4)]

sy = [8.9, 6.4, 3.9, 1.4]          # vertical position of each dispute box
ey = [9.4, 7.4, 5.4, 3.4, 1.4]     # vertical position of each evidence box

fig, ax = plt.subplots(figsize=(12.2, 4.9))
for i, s in enumerate(SC):
    ax.add_patch(FancyBboxPatch((0.1, sy[i] - 0.55), 3.5, 1.1,
                                boxstyle="round,pad=0.02,rounding_size=0.15",
                                facecolor="#EDEDED", edgecolor="0.55", lw=1.2))
    ax.text(1.85, sy[i], textwrap.fill(s, 26), ha="center", va="center", fontsize=9)
for j, (e, u) in enumerate(EV):
    ax.add_patch(FancyBboxPatch((6.0, ey[j] - 0.55), 5.6, 1.1,
                                boxstyle="round,pad=0.02,rounding_size=0.15",
                                facecolor="#DCE6F2", edgecolor="#4C72B0", lw=1.3))
    ax.text(8.8, ey[j] + 0.16, e, ha="center", va="center",
            fontsize=9.3, fontweight="bold")
    ax.text(8.8, ey[j] - 0.24, f"built in {u}", ha="center", va="center",
            fontsize=8.2, color="#1F3B63")
for i, j in LINKS:
    ax.annotate("", xy=(5.95, ey[j]), xytext=(3.65, sy[i]),
                arrowprops=dict(arrowstyle="-|>", color="#8B6C42", lw=1.5,
                                connectionstyle="arc3,rad=0.06"))

ax.text(1.85, 10.35, "Dispute", ha="center", fontsize=10.5, fontweight="bold")
ax.text(8.8, 10.35, "The evidence that decides it", ha="center",
        fontsize=10.5, fontweight="bold")
cnt = collections.Counter(EV[j][1].split(',')[0] for _, j in LINKS)
ax.text(8.8, 0.15, "  ".join(f"{k} x{v}" for k, v in sorted(cnt.items()))
        + "   - every arrow lands on an artefact this course built",
        ha="center", fontsize=8.8, color="#1F3B63")
ax.set_xlim(0, 11.8)
ax.set_ylim(-0.2, 10.9)
ax.axis("off")
ax.set_title("Four different disputes, one short list of evidence", fontsize=12.5)
fig.tight_layout()
plt.show()
```

**What to see in this figure.** Four unrelated disputes on the left — a crash, a hiring
screen, a copyright claim, a data transfer. Five arrows. **Every arrow lands in a blue box, and
every blue box carries a unit tag**: Unit 2 once, Unit 3 twice, Unit 4 twice. Nothing on the
right is a legal artefact you would have to commission specially; they are the audit trail, the
per-group metrics, the model card, the provenance record and the processing record you already
know how to produce.

The hiring row is the one with two arrows, because that dispute is decided by *two* artefacts
together — the fairness numbers **and** the validation record showing what you checked before

instruction: these five artefacts are what you will be asked for, so decide now which of them
your system actually produces.



```python
# WHAT this cell draws: the autonomous-vehicle row of the table above, with real
# numbers attached - the Tempe 2018 Uber ATG fatality, as the logs recorded it.
# WHY: "key evidence: logs" is abstract until you see the window the logs cover.
# Only the three facts stated in the Discuss prompt below are drawn here; this is
# not a frame-by-frame reconstruction of the NTSB report.
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(11.6, 3.7))

# (label, start s, end s, colour, note, hatched?)  - times are seconds vs impact.
rows = [("Volvo factory automatic\nemergency braking", -6.5, 0.0, "#BBBBBB",
         "DISABLED for the whole window", True),
        ("Uber software:\nbraking suppressed", -1.0, 0.0, "#C44E52",
         "1 second", False),
        ("Object detected by\nthe system", -5.6, 0.0, "#4C72B0",
         "5.6 seconds of data", False)]
for i, (lab, x0, x1, col, note, hatch) in enumerate(rows):
    ax.barh(i, x1 - x0, left=x0, height=0.5, color=col, alpha=0.85,
            hatch="//" if hatch else None, edgecolor="white" if hatch else col)
    ax.text(x0 + 0.08, i + 0.42, note, fontsize=8.6, color="0.2", va="bottom")

ax.axvline(0, color="#8B1A1E", lw=2.2)
ax.text(0.06, 2.75, "IMPACT", color="#8B1A1E", fontsize=10, fontweight="bold", va="top")
ax.scatter([-5.6], [2], s=70, color="#4C72B0", zorder=5)   # first detection
ax.set_yticks(range(len(rows)))
ax.set_yticklabels([r[0] for r in rows], fontsize=9)
ax.set_xlim(-6.6, 0.75)
ax.set_ylim(-0.6, 2.95)
ax.set_xticks([-6, -5, -4, -3, -2, -1, 0])
ax.set_xlabel("Time relative to impact (seconds; 0 = impact)")
ax.grid(axis="x", alpha=0.3)
ax.set_axisbelow(True)
for s in ("top", "right"):
    ax.spines[s].set_visible(False)
ax.set_title("Uber ATG, Tempe 2018: what the logs recorded before impact", fontsize=12)
fig.tight_layout()
plt.show()
```

## ⚠️ Where this breaks

- **This is teaching analysis, not legal advice, and the law here is genuinely unsettled.**
  Mobley is a preliminary certification, not a final judgment on the merits. Doctrines like
  product liability for a learning system, and the agency theory applied to AI vendors, are
  being worked out case by case right now. Anything in this notebook could be overtaken.
- **Liability is jurisdictional all the way down.** The scenarios use US-style employment and
  product-liability framing. Saudi, EU and UK law allocate responsibility differently — the EU
  AI Act, for instance, imposes duties on *providers* and *deployers* as defined roles that do
  not map cleanly onto "vendor" and "employer".
- **Scenario analysis rewards the cases you thought of.** Four dispute types cover the ones
  currently in the news. The disputes that will matter to your career are the ones nobody has
  litigated yet — which is precisely why the artefacts in Unit 4 matter more than the
  taxonomy in this notebook.
- **Evidence has a retention clock and a privacy cost.** The audit trail that proves you acted
  reasonably is personal data with storage-limitation duties (Unit 3, Notebook 04). "Keep
  everything forever" is not available, and the retention period you choose is itself a
  decision a court may examine.
- **Winning the legal argument is not the goal.** iTutorGroup paid $365,000; the ~200
  applicants rejected in 2020 were invited to reapply in 2023. Liability allocates cost after
  the harm. Everything in Units 2–4 exists to prevent it.

**The cheaper alternative to legal risk analysis:** for most teams the highest-value legal work
is not analysis at all. It is **contractual** — require your vendor to supply disaggregated
performance evidence, an independent audit, and a right to test the model yourself, and put
the indemnity where the evidence is. That single clause does more than any amount of scenario
mapping after the fact.
