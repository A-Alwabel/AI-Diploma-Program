## 📌 The case: the first AI hiring case a regulator actually won

**EEOC v. iTutorGroup**, settled **August 2023** — the US Equal Employment Opportunity
Commission's first resolved lawsuit over discriminatory use of AI in hiring.

There was no mysterious model. The company's application-review software had been
**programmed to automatically reject female applicants aged 55 or older and male applicants
aged 60 or older**. The EEOC alleged this violated the Age Discrimination in Employment Act;
iTutorGroup agreed to pay **$365,000** to a class of more than 200 rejected applicants, adopt
anti-discrimination policies, invite the rejected applicants to reapply, and stop asking
applicants for their date of birth.

Two things make this the right case for this notebook. First, **no new AI law was needed** —
a 1967 employment statute did the work. Sector regulators already have the authority and the
appetite. Second, a **sector-specific rule** now goes further: **New York City Local Law 144**
requires any employer using an automated employment decision tool on NYC candidates to
commission an **independent annual bias audit**, **publish a summary of the results**, and give
candidates at least **10 business days' notice**. Enforcement began **5 July 2023**, with
penalties from $500 for a first violation up to $1,500 per day thereafter.

**The WHY — what goes wrong without sector knowledge.** Horizontal laws set the floor. The
regulator who actually shows up is the sector one — the FDA for a diagnostic model, banking
supervisors for a credit model, an employment agency for a screening tool — and they arrive
with decades of doctrine, existing case law and a much shorter route to enforcement than any
new AI statute.

## Part 1: Sector-Specific Rules on Top of the Global Baseline

Industries with existing regulators (medicine, finance, transport) apply their own
frameworks to AI. Snapshot below is a **teaching summary as of mid-2025** - always check
the regulator's current guidance.

```python
# Why sector rules matter: horizontal laws (GDPR, AI Act) set the floor -
# but healthcare, finance, and vehicles each add their own regulator with teeth.

# Step 1: Industry regulation map + a requirements lookup
import pandas as pd

# One row per regulator/rule: its scope and the requirement AI teams actually feel.
industry_regs = pd.DataFrame([
    ['Healthcare', 'FDA (USA)', 'AI/ML-enabled medical device pathway (SaMD)',
     'Clinical validation; change-control plans for learning systems'],
    ['Healthcare', 'HIPAA (USA)', 'Health data privacy',
     'Safeguards for patient data used in training'],
    ['Finance', 'Federal Reserve SR 11-7 (USA)', 'Model risk management',
     'Model validation, documentation, ongoing monitoring'],
    ['Finance', 'ECOA / fair lending (USA)', 'Anti-discrimination in credit',
     'Adverse-action notices - i.e., EXPLANATIONS for denials (Unit 4!)'],
    ['Autonomous vehicles', 'NHTSA (USA)', 'Vehicle safety standards',
     'Crash reporting for automated driving systems'],
    ['Autonomous vehicles', 'UNECE WP.29 (international)', 'Type approval',
     'Requirements for automated lane keeping and software updates'],
], columns=['Industry', 'Regulator / rule', 'Scope', 'Key AI-relevant requirement'])

print("INDUSTRY-SPECIFIC AI REGULATION MAP (teaching summary, mid-2025)")
print("=" * 110)
print(industry_regs.to_string(index=False))

# A lookup helper: given an industry, list what an AI product must satisfy there.
def requirements_for(industry):
    rows = industry_regs[industry_regs['Industry'].str.lower() == industry.lower()]
    return list(rows['Key AI-relevant requirement'])

print("\nLookup demo - what must an AI product satisfy per industry?")
for ind in ['Healthcare', 'Finance', 'Autonomous vehicles']:
    print(f"\n  {ind}:")
    for req in requirements_for(ind):
        print(f"    - {req}")
print("\nPattern: every sector demands (1) validation evidence, (2) documentation,")
print("(3) monitoring, and often (4) explanations - the exact skills of Units 2-4.")
```

## ⚠️ Where this breaks

- **This map is a teaching summary and it expires.** Sector guidance changes more often than
  primary legislation — regulator circulars, updated device pathways, new state statutes. Read
  the current guidance from the regulator itself before relying on anything here.
- **Sector rules stack; they do not replace each other.** A clinical AI product in the EU can
  be simultaneously a medical device, a high-risk AI system under the AI Act, and a personal
  data processing operation under GDPR — three conformity regimes, three sets of documentation,
  possibly three auditors. Budget for the intersection, not for the maximum of the three.
- **A mandated audit is only as good as the auditor and the scope.** Local Law 144 requires an
  independent bias audit and publication; the employer still selects the auditor and, in
  practice, influences the scope and the data supplied. Unit 4 Notebook 07's lesson applies
  directly here — an evaluation the subject controls is weak evidence.
- **"Human review" is the standard escape hatch, and Unit 4 measured why it fails.** Several
  of these regimes soften their requirements where a human makes the final decision. Automation
  bias (AI Act Art. 14(4)(b)), the Horizon prosecutions and the SCHUFA reasoning all say that
  a human who follows the score has not made the decision.
- **Sector regulators are slow, and AI deployment is not.** iTutorGroup's rejections happened
  in 2020 and were resolved in 2023. Enforcement after the fact does not help the 200 people
  who were rejected, and it is not a substitute for testing before deployment.

**The cheaper alternative when you cannot map every regime:** identify the **single most
stringent obligation** across the sectors you touch — usually documented pre-deployment
validation plus disaggregated performance evidence — and build to that. It satisfies most of
the others as a by-product and it is exactly the evidence Units 2 and 4 taught you to produce.
