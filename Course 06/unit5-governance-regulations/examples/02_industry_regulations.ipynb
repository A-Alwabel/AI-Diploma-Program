## 📌 The case: the first AI hiring case a regulator actually won

**EEOC v. iTutorGroup**, settled in **2023** — the EEOC announced the consent decree on
**11 September 2023** — the US Equal Employment Opportunity Commission's first resolved
lawsuit over discriminatory use of AI in hiring.

There was no mysterious model. The company's application-review software had been
**programmed to automatically reject female applicants aged 55 or older and male applicants
aged 60 or older**. The EEOC alleged this violated the Age Discrimination in Employment Act;
iTutorGroup agreed to pay **$365,000** to a class of more than 200 rejected applicants, adopt
anti-discrimination policies, invite the rejected applicants to reapply, and stop asking
applicants for their date of birth.

Two things make this the right case for this notebook. First, **no new AI law was needed** —
a 1967 employment statute did the work. Sector regulators already have the authority and the
appetite. Second, a **sector-specific rule** now goes further: **New York City Local Law 144**
requires any employer using an automated employment decision tool on NYC candidates to
commission an **independent annual bias audit**, **publish a summary of the results**, and give
candidates at least **10 business days' notice**. Enforcement began **5 July 2023**, with
penalties from $500 for a first violation up to $1,500 per day thereafter.

**The WHY — what goes wrong without sector knowledge.** Horizontal laws set the floor. The
regulator who actually shows up is the sector one — the FDA for a diagnostic model, banking
supervisors for a credit model, an employment agency for a screening tool — and they arrive
with decades of doctrine, existing case law and a much shorter route to enforcement than any
new AI statute.

```python
# WHAT this cell draws: the iTutorGroup case on a time axis - when the rejections
# happened, when the remedy arrived, and how far apart those two are.
# WHY: the paragraph above makes two claims that are hard to feel in prose - that
# an old statute did the work, and that enforcement arrives years after the harm.
# A date axis makes the second one impossible to miss.
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(12.4, 4.4))
ax.hlines(0, 2019.6, 2024.8, color="0.35", lw=1.6)          # the timeline itself
ax.axvspan(2020.0, 2023.70, color="#C44E52", alpha=0.09)     # the waiting period

# Three dated events, all taken from the paragraph above.
for x, c in [(2020.0, "#C44E52"), (2023.51, "#4C72B0"), (2023.70, "#55A868")]:
    ax.scatter([x], [0], s=80, color=c, zorder=6)

ax.annotate("2020\nApplication-review software automatically rejects\n"
            "women aged 55+ and men aged 60+",
            xy=(2020.0, 0.06), xytext=(2019.68, 0.62), fontsize=8.8,
            ha="left", va="bottom",
            arrowprops=dict(arrowstyle="->", color="#C44E52", lw=1.2))
ax.annotate("11 Sep 2023\nEEOC consent decree announced:\n"
            "$365,000 to 200+ rejected applicants\n(under $1,825 each)",
            xy=(2023.72, 0.06), xytext=(2024.75, 0.62), fontsize=8.8,
            ha="right", va="bottom",
            arrowprops=dict(arrowstyle="->", color="#55A868", lw=1.2))
ax.annotate("5 Jul 2023  NYC Local Law 144 enforcement begins:\n"
            "independent annual bias audit, published summary,\n"
            "10 business days' notice to candidates",
            xy=(2023.49, -0.06), xytext=(2022.15, -0.34), fontsize=8.8,
            ha="center", va="top",
            arrowprops=dict(arrowstyle="->", color="#4C72B0", lw=1.2))

# The gap between the conduct and the remedy - the point of the whole figure.
ax.annotate("", xy=(2023.70, -0.92), xytext=(2020.0, -0.92),
            arrowprops=dict(arrowstyle="<->", color="#C44E52", lw=1.8))
ax.text(2021.85, -1.02, "about 3 years between the conduct and the remedy",
        ha="center", va="top", color="#8B1A1E", fontsize=10.5, fontweight="bold")
ax.text(2019.68, -1.42,
        "Statute used: Age Discrimination in\nEmployment Act (1967).\n"
        "No AI-specific law was involved.",
        ha="left", va="top", fontsize=8.6,
        bbox=dict(boxstyle="round,pad=0.4", fc="#F3F3F3", ec="0.7"))

ax.set_xlim(2019.6, 2024.8)
ax.set_ylim(-2.25, 1.35)
ax.set_xticks([2020, 2021, 2022, 2023, 2024])
ax.set_yticks([])                       # the y direction only separates labels
ax.set_xlabel("Year")
for s in ("top", "left", "right"):
    ax.spines[s].set_visible(False)
ax.set_title("EEOC v. iTutorGroup: the sector regulator won - "
             "three years after the rejections", fontsize=11.5)
fig.tight_layout()
plt.show()
```

**What to see in this figure.** Three dots on one date axis. The red dot is the conduct
(2020, automatic rejection of women aged 55+ and men aged 60+); the green dot is the remedy
(11 September 2023, a $365,000 consent decree shared among more than 200 people — **under
$1,825 each**); the blue dot is a different kind of rule arriving in between (NYC Local Law
144, enforceable from 5 July 2023, which requires the audit *before* anyone is harmed rather
than a settlement afterwards).

The double-headed red arrow is the figure's whole argument: **about three years separate the
conduct from the remedy.** And the grey box names the statute that did the work — a 1967
age-discrimination law, not an AI law. Hold both facts at once: sector regulators already have
the power, and they arrive late.

## Part 1: Sector-Specific Rules on Top of the Global Baseline

Industries with existing regulators (medicine, finance, transport) apply their own
frameworks to AI. Snapshot below is a **teaching summary as of mid-2025** - always check
the regulator's current guidance.

```python
# Why sector rules matter: horizontal laws (GDPR, AI Act) set the floor -
# but healthcare, finance, and vehicles each add their own regulator with teeth.

# Step 1: Industry regulation map + a requirements lookup
import pandas as pd

# One row per regulator/rule: its scope and the requirement AI teams actually feel.
industry_regs = pd.DataFrame([
    ['Healthcare', 'FDA (USA)', 'AI/ML-enabled medical device pathway (SaMD)',
     'Clinical validation; change-control plans for learning systems'],
    ['Healthcare', 'HIPAA (USA)', 'Health data privacy',
     'Safeguards for patient data used in training'],
    ['Finance', 'Federal Reserve SR 11-7 (USA)', 'Model risk management',
     'Model validation, documentation, ongoing monitoring'],
    ['Finance', 'ECOA / fair lending (USA)', 'Anti-discrimination in credit',
     'Adverse-action notices - i.e., EXPLANATIONS for denials (Unit 4!)'],
    ['Autonomous vehicles', 'NHTSA (USA)', 'Vehicle safety standards',
     'Crash reporting for automated driving systems'],
    ['Autonomous vehicles', 'UNECE WP.29 (international)', 'Type approval',
     'Requirements for automated lane keeping and software updates'],
], columns=['Industry', 'Regulator / rule', 'Scope', 'Key AI-relevant requirement'])

print("INDUSTRY-SPECIFIC AI REGULATION MAP (teaching summary, mid-2025)")
print("=" * 110)
print(industry_regs.to_string(index=False))

# A lookup helper: given an industry, list what an AI product must satisfy there.
def requirements_for(industry):
    rows = industry_regs[industry_regs['Industry'].str.lower() == industry.lower()]
    return list(rows['Key AI-relevant requirement'])

print("\nLookup demo - what must an AI product satisfy per industry?")
for ind in ['Healthcare', 'Finance', 'Autonomous vehicles']:
    print(f"\n  {ind}:")
    for req in requirements_for(ind):
        print(f"    - {req}")
print("\nPattern: every sector demands (1) validation evidence, (2) documentation,")
print("(3) monitoring, and often (4) explanations - the exact skills of Units 2-4.")
```

```python
# WHAT this cell draws: the six rules you just printed, as a grid - which of five
# obligation families each one actually demands.
# WHY: the printed line above claims "every sector demands validation,
# documentation, monitoring, and often explanations". A grid either shows three
# fully-marked columns or it does not - the claim becomes checkable instead of
# rhetorical.
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

# The five families are read off the 'Key AI-relevant requirement' column above.
COLS = ["Pre-deployment\nvalidation evidence", "Technical\ndocumentation",
        "Ongoing\nmonitoring", "Explanation to the\naffected person",
        "Personal-data\nsafeguards"]

# (row label, industry, which column indices that rule demands)
ROWS = [
    ("Horizontal floor: GDPR + EU AI Act", "baseline",   {0, 1, 2, 3, 4}),
    ("Healthcare - FDA (SaMD pathway)",    "Healthcare", {0, 1, 2}),
    ("Healthcare - HIPAA",                 "Healthcare", {4}),
    ("Finance - Fed. Reserve SR 11-7",     "Finance",    {0, 1, 2}),
    ("Finance - ECOA / fair lending",      "Finance",    {3}),
    ("Autonomous vehicles - NHTSA",        "AV",         {2}),
    ("Autonomous vehicles - UNECE WP.29",  "AV",         {0, 1, 2}),
]

# Union per industry, then the intersection across industries: this is COMPUTED,
# not asserted, so the bottom row of the figure cannot flatter the claim.
union = {}
for _, ind, cells in ROWS:
    if ind == "baseline":
        continue
    union.setdefault(ind, set()).update(cells)
common = set.intersection(*union.values())

print("What each industry demands once you merge its rules:")
for ind, cells in union.items():
    print(f"  {ind:<12}: " + ", ".join(COLS[i].replace(chr(10), ' ') for i in sorted(cells)))
print("\nDemanded by ALL THREE industries: "
      + ", ".join(COLS[i].replace(chr(10), ' ') for i in sorted(common)))

order = ROWS + [("Demanded by ALL THREE industries", "summary", common)]
fig, ax = plt.subplots(figsize=(11.6, 4.9))
colmap = {"baseline": "#8C8C8C", "Healthcare": "#4C72B0", "Finance": "#55A868",
          "AV": "#DD8452", "summary": "#C44E52"}
for r, (label, ind, cells) in enumerate(order):
    for c in range(len(COLS)):
        if c in cells:
            ax.add_patch(Rectangle((c - 0.36, r - 0.32), 0.72, 0.64,
                                   color=colmap[ind],
                                   alpha=0.55 if ind == "baseline" else 0.9))
        else:
            ax.add_patch(Rectangle((c - 0.36, r - 0.32), 0.72, 0.64,
                                   facecolor="none", edgecolor="0.85", lw=0.9))
ax.set_xlim(-0.6, len(COLS) - 0.4)
ax.set_ylim(len(order) - 0.4, -0.7)      # first row at the top
ax.set_xticks(range(len(COLS)))
ax.set_xticklabels(COLS, fontsize=8.5)
ax.xaxis.set_ticks_position("top")
ax.set_yticks(range(len(order)))
ax.set_yticklabels([r[0] for r in order], fontsize=9)
ax.axhline(0.5, color="0.3", lw=1.2)                 # baseline is separate
ax.axhline(len(ROWS) - 0.5, color="0.3", lw=1.2)     # summary row is separate
for s in ax.spines.values():
    s.set_visible(False)
ax.tick_params(length=0)
ax.set_title("Every sector regulator asks for the same three things\n"
             "(filled = that rule demands that artefact)", fontsize=11.5, pad=34)
fig.tight_layout()
plt.show()
```

**What to see in this figure.** Read it as a checklist grid: one row per rule, one column
per obligation family, a filled block where that rule demands that artefact.

- The **grey top row** is the horizontal floor — GDPR and the AI Act demand all five. Sector
  rules do not replace it; they sit **below the line, on top of it**.
- The **red bottom row is computed, not asserted**: it is the set intersection of what
  healthcare, finance and autonomous vehicles each demand once you merge their rules. Three
  columns survive — **pre-deployment validation evidence, technical documentation, ongoing
  monitoring**. Those three are filled; explanation and personal-data safeguards are empty.
- The two empty columns are informative too. **Explanation to the affected person** is filled
  only by ECOA (adverse-action notices), and **personal-data safeguards** only by HIPAA. If
  your sector happens to be one of those, that column is not optional for you.

So the practical reading: build validation evidence, documentation and monitoring once, and you
have satisfied the common core of three unrelated regulators — and, per the grey row, most of
the horizontal floor as well.

## 💬 Discuss

Your industry map lists **FDA** (clinical validation and change control for learning
systems), **HIPAA**, **SR 11-7** (model risk management), and **ECOA / fair lending** — four
regimes that all predate modern machine learning and all now apply to it.

1. **What does "change control" mean for a model that learns?** The FDA's pathway assumes a
   device that is fixed once approved. A model retrained weekly is a different device every
   week. Where do you draw the line between an approved update and a new product requiring
   re-approval — and does your answer make continuous improvement impossible, or does it make
   safety impossible? Pick one and live with the consequence.
2. **Would you want Local Law 144 in your country?** It mandates an independent annual bias
   audit and public publication of the summary. That is a real cost, it applies to employers
   who may not have built the tool, and it does not require the audit to *pass* — only to
   exist and be published. Is publication alone enough to change behaviour? What would you add
   or remove, and who would you make responsible: the employer, or the vendor?
3. **Old law versus new law.** iTutorGroup was resolved under a 1967 age-discrimination
   statute; no AI-specific rule was involved. So is a dedicated AI Act necessary, or does it
   mostly duplicate powers regulators already have and add compliance overhead? Argue the
   position you do **not** hold first — that is the harder half, and it is the one you will
   need in a real policy meeting.


## ⚠️ Where this breaks

- **This map is a teaching summary and it expires.** Sector guidance changes more often than
  primary legislation — regulator circulars, updated device pathways, new state statutes. Read
  the current guidance from the regulator itself before relying on anything here.
- **Sector rules stack; they do not replace each other.** A clinical AI product in the EU can
  be simultaneously a medical device, a high-risk AI system under the AI Act, and a personal
  data processing operation under GDPR — three conformity regimes, three sets of documentation,
  possibly three auditors. Budget for the intersection, not for the maximum of the three.
- **A mandated audit is only as good as the auditor and the scope.** Local Law 144 requires an
  independent bias audit and publication; the employer still selects the auditor and, in
  practice, influences the scope and the data supplied. Unit 4 Notebook 07's lesson applies
  directly here — an evaluation the subject controls is weak evidence.
- **"Human review" is the standard escape hatch, and Unit 4 measured why it fails.** Several
  of these regimes soften their requirements where a human makes the final decision. Automation
  bias (AI Act Art. 14(4)(b)), the Horizon prosecutions and the SCHUFA reasoning all say that
  a human who follows the score has not made the decision.
- **Sector regulators are slow, and AI deployment is not.** iTutorGroup's rejections happened
  in 2020 and were resolved in 2023. Enforcement after the fact does not help the 200 people
  who were rejected, and it is not a substitute for testing before deployment.

**The cheaper alternative when you cannot map every regime:** identify the **single most
stringent obligation** across the sectors you touch — usually documented pre-deployment
validation plus disaggregated performance evidence — and build to that. It satisfies most of
the others as a by-product and it is exactly the evidence Units 2 and 4 taught you to produce.
