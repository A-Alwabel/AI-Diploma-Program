## 📌 The case: one company, five regulators, and a practice that is now simply illegal

**Clearview AI** built a facial-recognition search engine by scraping billions of face images
from the public web without anyone's consent, converting each into a biometric template, and
selling lookups. Nothing about that was hidden; the company argued the images were public.

European regulators disagreed, one at a time:

| Authority | Fine |
|---|---|
| France — CNIL (October 2022) | **€20 million** |
| Italy — Garante | **€20 million** |
| Greece — HDPA | **€20 million** |
| United Kingdom — ICO (2022) | **£7.5 million** — *still contested: the First-tier Tribunal held in October 2023 that the ICO had no jurisdiction; the Upper Tribunal reversed that in October 2025, and Clearview has permission to appeal again* |
| Netherlands — Dutch DPA (September 2024) | **€30.5 million** |

Then the law stopped arguing case by case. The **EU AI Act (Regulation (EU) 2024/1689)**,
in force since **1 August 2024**, lists **untargeted scraping of facial images from the
internet or CCTV to build facial-recognition databases** among its **Article 5 prohibited
practices** — and those prohibitions have applied since **2 February 2025**. Breaching them
carries administrative fines of up to **€35 million or 7 % of total worldwide annual
turnover, whichever is higher** (Article 99).

That is the whole shape of AI regulation in one company's history: a practice moves from
"legal because nobody wrote a rule" to "fined under general data-protection law" to "named
and prohibited in a dedicated statute", in about five years.

**The WHY — what goes wrong without knowing the landscape.** An engineer who does not know
what a **risk tier** is will build a high-risk system with no conformity assessment, no
logging and no human oversight, and discover it during a tender. And the tiers are not an EU
curiosity: the risk-based structure has become the template most new AI rules copy, so
learning it once buys you most of the map. Saudi Arabia's own **PDPL** (Royal Decree M/19,
2021, amended 2023; in force September 2023, full enforcement from **14 September 2024**,
supervised by **SDAIA**) reaches any organisation processing the personal data of people in
the Kingdom, wherever that organisation sits.

```python
# WHAT this cell draws: the Clearview AI case as two pictures - the money on the
# left, and the legal status of the practice over time on the right.
# WHY: the paragraph above says a practice moved from "legal because nobody wrote
# a rule" to "fined case by case" to "prohibited by name". That is a claim about
# a sequence, and a sequence is much easier to argue about when you can see it.
import matplotlib.pyplot as plt

# Real, dated enforcement decisions taken from the table above. We plot only the
# four euro-denominated fines: the UK ICO penalty is in pounds (GBP 7.5 m) and is
# still being litigated, so mixing it into a EUR axis would be dishonest.
fines = [("France - CNIL\n(Oct 2022)", 20.0),
         ("Italy - Garante\n(2022)", 20.0),
         ("Greece - HDPA\n(2022)", 20.0),
         ("Netherlands - DPA\n(Sep 2024)", 30.5)]
labels = [f[0] for f in fines]
values = [f[1] for f in fines]

fig, (axL, axR) = plt.subplots(1, 2, figsize=(13.0, 4.6),
                               gridspec_kw={"width_ratios": [1.0, 1.3]})

# --- Left panel: what each regulator actually charged --------------------------
ypos = list(range(len(values)))
bars = axL.barh(ypos, values, color="#4C72B0", height=0.55)
axL.set_yticks(ypos)
axL.set_yticklabels(labels, fontsize=8.5)
axL.set_ylim(4.3, -0.7)                      # top row first, plus room for the note
axL.set_xlabel("Fine imposed (EUR millions)")
axL.set_xlim(0, 40)
axL.set_title("Fined four times under data-protection law\n(EUR 90.5 m in total)",
              fontsize=10)
for b, v in zip(bars, values):
    axL.text(v + 0.7, b.get_y() + b.get_height() / 2, f"{v:g}", va="center", fontsize=9)
# The AI Act ceiling for this exact practice, drawn on the same scale as the fines.
axL.axvline(35, color="#C44E52", ls="--", lw=1.6)
axL.text(34.2, 3.95,
         "AI Act Art. 99 ceiling for a prohibited\npractice: EUR 35 m or 7 % of\n"
         "worldwide turnover, whichever is higher",
         color="#C44E52", fontsize=7.5, ha="right", va="center")
axL.grid(axis="x", alpha=0.3)
axL.set_axisbelow(True)

# --- Right panel: the LEGAL STATUS of the practice, as a step function ----------
# The y axis is not a quantity: it has three named levels, and the line only ever
# steps up. That shape IS the lesson.
xs = [2021.0, 2022.5, 2022.5, 2025.09, 2025.09, 2026.0]
ys = [0, 0, 1, 1, 2, 2]
axR.plot(xs, ys, lw=2.8, color="#55A868")
axR.scatter([2022.5, 2024.75], [1, 1], s=55, color="#4C72B0", zorder=5)  # the fines
axR.axvline(2024.58, color="#6A5ACD", ls=":", lw=1.4)                    # 1 Aug 2024
axR.set_yticks([0, 1, 2])
axR.set_yticklabels(["No rule names\nthe practice",
                     "Fined case by case\nunder GDPR",
                     "Prohibited by name\n(AI Act Art. 5)"], fontsize=8.5)
axR.set_ylim(-0.65, 2.85)
axR.set_xlim(2021.0, 2026.0)
axR.set_xticks([2021, 2022, 2023, 2024, 2025, 2026])
axR.set_xlabel("Year")
axR.set_title("The same practice, as the law escalates around it", fontsize=10)
axR.annotate("2022: France, Italy,\nGreece, UK fine it", xy=(2022.5, 1),
             xytext=(2021.1, 1.55), fontsize=8,
             arrowprops=dict(arrowstyle="->", lw=1.0))
axR.annotate("Sep 2024:\nNetherlands EUR 30.5 m", xy=(2024.72, 1),
             xytext=(2022.9, 0.25), fontsize=8,
             arrowprops=dict(arrowstyle="->", lw=1.0))
axR.annotate("2 Feb 2025: Art. 5\nprohibitions apply", xy=(2025.09, 1.95),
             xytext=(2022.75, 2.45), fontsize=8,
             arrowprops=dict(arrowstyle="->", lw=1.0))
axR.text(2024.5, -0.45, "1 Aug 2024\nAI Act in force", fontsize=7.5,
         color="#6A5ACD", ha="right", va="bottom")
axR.grid(axis="x", alpha=0.3)
axR.set_axisbelow(True)

fig.suptitle("Clearview AI: from unregulated, to fined, to prohibited", fontsize=12.5)
fig.tight_layout(rect=[0, 0, 1, 0.94])
plt.show()
```

**What to see in this figure.** *Left:* four European authorities fined the same company
under general data-protection law — EUR 20 m, 20 m, 20 m and 30.5 m, **EUR 90.5 m in total** —
and every one of those bars sits **below** the dashed red line, which is the AI Act's own
ceiling for a *prohibited* practice (EUR 35 m, or 7 % of worldwide turnover, whichever is
higher). The UK's GBP 7.5 m penalty is deliberately not plotted: different currency, and still
being litigated.

*Right:* the green line is **not a quantity — it is a legal status**, and it only ever steps
up. Flat at "no rule names the practice"; up to "fined case by case under GDPR" in 2022 (the
two blue dots are the 2022 cluster and the Dutch decision of September 2024); up again on
**2 February 2025** to "prohibited by name". That second step is the entire argument of this
notebook: regulators stopped fining one company at a time and wrote the practice into a
statute. Three years, two steps.

## Part 1: The Global Regulatory Landscape (as of mid-2025)

Regulation moves fast - treat the snapshot below as a **starting map, not legal advice**,
and verify current status before relying on it.

```python
# Why a structured table: the regulatory landscape is a moving target -
# comparing instruments by region, bindingness, and core idea reveals the pattern
# (risk-based, human-centred) that individual headlines hide.

# Step 1: Compare the major regulatory instruments in a structured table
import pandas as pd

# One row per instrument. The 'Status' column is dated on purpose:
# regulation changes faster than course material - always verify.
regulations = pd.DataFrame([
    ['EU', 'EU AI Act', 'Binding regulation',
     'Risk-based tiers; obligations scale with risk',
     'In force Aug 2024; obligations phase in 2025-2027'],
    ['EU', 'GDPR', 'Binding regulation',
     'Personal data protection; Art. 22 limits automated decisions',
     'In force since 2018 (Unit 3 covered it)'],
    ['USA', 'Executive Order 14110 (2023)', 'Executive order',
     'Safety testing, reporting for powerful models',
     'Rescinded Jan 2025; successor policy emphasizes AI leadership'],
    ['USA', 'State laws (e.g. Colorado AI Act)', 'State legislation',
     'Algorithmic discrimination in consequential decisions',
     'Varies by state; landscape still forming'],
    ['OECD/International', 'OECD AI Principles', 'Non-binding principles',
     'Human-centred values, transparency, accountability',
     'Adopted 2019; basis for many national policies'],
    ['International', 'Council of Europe AI Convention', 'Treaty',
     'Human rights, democracy, rule of law in AI',
     'Opened for signature Sep 2024'],
], columns=['Region', 'Instrument', 'Type', 'Core idea', 'Status (verify - may have changed!)'])

print("GLOBAL AI REGULATION SNAPSHOT (teaching summary, as of mid-2025)")
print("=" * 110)
print(regulations.to_string(index=False))
```

```python
# Why the EU tiers matter beyond the EU: risk-based classification has become
# THE template - most new AI rules worldwide copy this structure.

# Step 2: The EU AI Act's risk tiers - the model most regulation now follows

# The four tiers: obligations scale with the potential harm of the use case.
risk_tiers = pd.DataFrame([
    ['Unacceptable risk', 'PROHIBITED',
     'Social scoring by governments; manipulative techniques causing harm'],
    ['High risk', 'Strict obligations (conformity assessment, logging, human oversight)',
     'Hiring, credit scoring, medical devices, emergency triage, critical infrastructure'],
    ['Limited risk', 'Transparency duties (disclose AI use)',
     'Chatbots, deepfakes/synthetic media labeling'],
    ['Minimal risk', 'No new obligations',
     'Spam filters, AI in video games'],
], columns=['Tier', 'Regulatory consequence', 'Examples'])

print("EU AI ACT RISK TIERS")
print("=" * 100)
print(risk_tiers.to_string(index=False))

def classify_eu_risk(use_case):
    """Toy classifier mapping a use case to an EU AI Act tier (teaching aid)."""
    prohibited = ['social scoring', 'subliminal manipulation']
    high = ['hiring', 'credit', 'medical', 'law enforcement', 'education scoring',
            'critical infrastructure', 'triage', 'emergency']
    limited = ['chatbot', 'deepfake', 'synthetic media']
    uc = use_case.lower()
    if any(k in uc for k in prohibited): return 'Unacceptable risk - PROHIBITED'
    if any(k in uc for k in high):       return 'High risk - strict obligations'
    if any(k in uc for k in limited):    return 'Limited risk - transparency duties'
    return 'Minimal risk - no new obligations'

# Apply the toy classifier to the system this course actually built, plus the
# use cases we discussed - notice where each one lands.
# The Units 2-4 model decides who is predicted to survive, i.e. a triage-style
# screening of people: Annex III of the AI Act names emergency triage explicitly.
print("\nClassifying this course's running example and its neighbours:")
for uc in ['emergency triage screening model (Units 2-4)', 'credit scoring model',
           'customer service chatbot', 'spam filter',
           'government social scoring system']:
    print(f"  {uc:<45} -> {classify_eu_risk(uc)}")
print("\nNotice where the system we spent Units 2-4 on lands: HIGH RISK - which is")
print("exactly why those units' skills (fairness, privacy, explainability, HITL,")
print("audit trails) map one-to-one onto this tier's obligations.")
```

```python
# WHAT this cell draws: the four tiers you just printed, as a picture - how many
# obligations each tier carries, where the classifier put each of our use cases,
# and what the maximum fine is per tier.
# WHY: the table says "obligations scale with risk". A table cannot show you that
# the scaling is 0 -> 2 -> 9, or that the FINE ceiling does not scale the same way.
import matplotlib.pyplot as plt

TIERS = ['Minimal risk', 'Limited risk', 'High risk', 'Unacceptable risk']

# The obligations counted below are the concrete Articles, so the bar heights are
# checkable against the Regulation rather than being an impression.
OBLIGATIONS = {
    'Minimal risk': [],
    'Limited risk': ['Art. 50(1) disclose the user is interacting with an AI',
                     'Art. 50(2,4) mark and label synthetic media / deepfakes'],
    'High risk': ['Art. 9 risk-management system',
                  'Art. 10 data governance',
                  'Art. 11 technical documentation',
                  'Art. 12 record-keeping (logs)',
                  'Art. 13 transparency and instructions for the deployer',
                  'Art. 14 human oversight',
                  'Art. 15 accuracy, robustness, cybersecurity',
                  'Art. 43 conformity assessment',
                  'Art. 49 registration in the EU database'],
    'Unacceptable risk': [],   # nothing to comply with: the practice is banned
}
# Article 99 fine ceilings, in EUR millions - "or the % of worldwide annual
# turnover, whichever is HIGHER".
MAX_FINE = {'Minimal risk': 0.0, 'Limited risk': 15.0,
            'High risk': 15.0, 'Unacceptable risk': 35.0}
FINE_PCT = {'Minimal risk': '-', 'Limited risk': '3 % turnover',
            'High risk': '3 % turnover', 'Unacceptable risk': '7 % turnover'}

# Re-use the classifier defined above so the picture cannot drift from the code.
SHORT = {'emergency triage screening model (Units 2-4)':
         "emergency triage screening\n(this course's own model)",
         'credit scoring model': 'credit scoring',
         'customer service chatbot': 'customer-service chatbot',
         'spam filter': 'spam filter',
         'government social scoring system': 'government social scoring'}
placed = {t: [] for t in TIERS}
for uc in SHORT:
    tier = classify_eu_risk(uc).split(' - ')[0]
    placed[tier].append(SHORT[uc])

fig, (axL, axR) = plt.subplots(1, 2, figsize=(13.4, 4.6),
                               gridspec_kw={"width_ratios": [2.3, 1.0]})
colors = {'Minimal risk': '#9BBF85', 'Limited risk': '#E8C15C',
          'High risk': '#E08A4B', 'Unacceptable risk': '#C44E52'}
counts = [len(OBLIGATIONS[t]) for t in TIERS]

# --- Left panel: obligation count per tier, plus where our examples landed ------
for i, t in enumerate(TIERS):
    if t == 'Unacceptable risk':
        # Drawn as a band across the WHOLE panel, not a bar of length 9: there is
        # no number of obligations that makes a prohibited system lawful.
        axL.barh(i, 20, color=colors[t], height=0.6, hatch='//',
                 edgecolor=colors[t], alpha=0.22)
        axL.text(0.25, i, 'PROHIBITED - no compliance route exists',
                 ha='left', va='center', color='#8B1A1E', fontsize=9.5,
                 fontweight='bold')
    else:
        axL.barh(i, counts[i], color=colors[t], height=0.6)
        if counts[i] > 0:
            axL.text(counts[i] - 0.15, i, str(counts[i]), ha='right', va='center',
                     color='white', fontsize=10, fontweight='bold')
        else:
            axL.text(0.12, i, '0', ha='left', va='center', fontsize=10,
                     fontweight='bold')
    txt = '; '.join(placed[t]) if placed[t] else '(none of our examples)'
    axL.text(10.4, i, txt, va='center', fontsize=8.2)

axL.set_yticks(range(len(TIERS)))
axL.set_yticklabels(TIERS, fontsize=9.5)
axL.set_xlim(0, 20)
axL.set_ylim(-0.6, 3.6)
axL.set_xticks([0, 2, 4, 6, 8])
axL.set_xlabel("Number of AI Act obligations attached to the tier")
axL.set_title("Obligations scale with risk - and here is where our examples land",
              fontsize=10.5)
axL.axvline(10.15, color='0.55', lw=0.9)
axL.text(10.4, 3.45, "use cases classified by classify_eu_risk() above",
         fontsize=8, style='italic', color='0.35')
axL.grid(axis='x', alpha=0.3)
axL.set_axisbelow(True)

# --- Right panel: the fine ceilings, on their own scale ------------------------
vals = [MAX_FINE[t] for t in TIERS]
axR.barh(range(len(TIERS)), vals, color=[colors[t] for t in TIERS], height=0.6)
for i, t in enumerate(TIERS):
    lab = "no new obligations" if vals[i] == 0 else f"EUR {vals[i]:g} m or {FINE_PCT[t]}"
    axR.text(vals[i] + 0.8, i, lab, va='center', fontsize=8.2)
axR.set_yticks(range(len(TIERS)))
axR.set_yticklabels([])
axR.set_ylim(-0.6, 3.6)
axR.set_xlim(0, 60)
axR.set_xticks([0, 15, 35])
axR.set_xlabel("Maximum administrative fine (EUR m)")
axR.set_title("Art. 99 ceilings: 'whichever is higher'", fontsize=10.5)
axR.grid(axis='x', alpha=0.3)
axR.set_axisbelow(True)

fig.suptitle("EU AI Act risk tiers: what each tier costs you", fontsize=12.5)
fig.tight_layout(rect=[0, 0, 1, 0.94])
plt.show()
```

**What to see in this figure.** *Left panel:* the bars count obligations, and the count is
**0 for minimal risk, 2 for limited risk, 9 for high risk**. The top row is not a longer bar —
it is a hatched band across the whole panel, because for an unacceptable-risk system there is
no number of obligations that makes it lawful. To the right of the grey divider each use case
sits in the tier `classify_eu_risk()` actually returned, and **the model this course built
across Units 2-4 is in the high-risk row, sharing it with credit scoring**.

*Right panel:* the fine ceilings do **not** scale the same way the obligations do. Limited risk
and high risk sit on the same EUR 15 m / 3 % ceiling; only the prohibited tier steps up to
EUR 35 m / 7 %. Read the two panels together: getting the tier wrong mostly costs you the
**nine obligations you never built** — the conformity assessment, the logs, the human
oversight — not a larger fine. That is why the tier is argued over in tenders.

## ⚠️ Where this breaks

- **This snapshot is out of date the day it is read.** Regulation moves faster than any
  teaching material: the AI Act's own obligations phase in across **2 February 2025**
  (prohibitions), **2 August 2025** (general-purpose AI), **2 August 2026** and **2 August
  2027** (high-risk), and guidance, standards and delegated acts keep arriving. Treat the
  table you printed as a **starting map, not legal advice**, and verify before you rely on it.
- **Risk tiers are a classification, and classification is contested.** Whether a system is
  "high risk" is a legal judgement about intended purpose, often argued between a vendor who
  wants a lower tier and a regulator who wants a higher one. The tier is not a property you
  can read off the architecture.
- **Compliance is not safety, and the gap is where the harm lives.** A fully conformity-assessed
  high-risk system can still be biased (Unit 2), unexplainable in practice (Unit 4) and a poor
  idea (Unit 1). The AI Act mandates process and documentation; it does not mandate that your
  fairness numbers be good.
- **Jurisdiction is genuinely hard for anything with users.** A Riyadh team serving European
  customers can be inside GDPR, the AI Act *and* the PDPL simultaneously, with conflicting
  duties on cross-border transfer and consent. "Build to the strictest standard" is a decent
  default and it is not always coherent — one regime may require data localisation that
  another forbids.
- **Enforcement, not text, is what binds.** A rule with no regulator, no budget and no
  cross-border reach is a suggestion. When you assess regulatory risk, read the enforcement
  record — how many decisions, against whom, collected or not — rather than the statute alone.

**The cheaper alternative when full regulatory analysis is disproportionate:** classify the
system by **who is harmed if it is wrong**. If the answer is a person's liberty, money, health
or employment, build to high-risk expectations regardless of what a tier table says — you will
be close to compliant in every regime, and you will not have to redo it when the classification
is challenged.


## 🔭 State of the field (2026)

The map you printed above is still the map, but two things on it have moved. **The dates stopped
being in the future.** The Article 5 prohibitions have applied since 2 February 2025 and the
general-purpose AI obligations since 2 August 2025 — and for that second tranche the European
Commission published a voluntary **General-Purpose AI Code of Practice** on 10 July 2025, the route
by which GPAI providers now demonstrate they meet the Article 53 transparency and copyright duties
(and, for systemic-risk models, Article 55). **And conformity became something you can run rather
than only read.** COMPL-AI translated the Act's legal language into measurable technical
requirements with an open benchmarking suite, ran it on 12 prominent LLMs, and found none fully
compliant — weakest on robustness, safety, diversity and fairness — while also showing that for
several requirements the benchmarks do not yet exist (Guldimann et al. 2024); a 2026 follow-up adds
an open, reproducible AI Act evaluation dataset for NLP and RAG systems, reaching 0.87 and 0.85 F1
on prohibited and high-risk scenario classification (Davvetas et al. 2026).

**Why the tiers are still the thing to learn here.** Every one of those suites is built *on top of*
the classification you just did by hand: a benchmark cannot tell you which obligations apply until
somebody has assigned the tier, and assigning the tier is a legal judgement about intended purpose,
not a measurement. The risk-based structure is also what the official plan examines, and it is the
template most new AI rules worldwide copy. Learn to place a system in a tier here; reach for the
test suites afterwards, to check what you built against the obligations that tier carries.
