## Part 2: Understanding the Framework

### 📚 Prerequisites (What You Need First)
- **Library imports** (from Part 1) - Understanding visualization tools
- **Python classes** - Understanding object-oriented programming

### 🔗 Relationship: What This Builds On
This is the **CORE** of ethical decision-making - we define the framework we'll use!
- Builds on: Python basics, ethical frameworks knowledge
- Comes BEFORE: Scenario analysis (we need the framework first)

### 📖 The Story
**Before framework**: We make ethical decisions based on intuition or gut feeling.
**After framework**: We follow a systematic 6-step process that ensures we consider all important factors!

```python
# Step 2: Create the Ethical Decision-Making Framework class
# This class implements a 6-step process for making ethical decisions in AI

# BEFORE: No structured decision-making process
# AFTER: A complete framework with 6 steps, each with questions to guide us

class EthicalDecisionFramework:
    """
    A framework for making ethical decisions in AI development.

    HOW IT WORKS:
    1. Defines 6 steps for ethical decision-making
    2. Each step has questions to guide thinking
    3. Provides structure for systematic ethical analysis
    4. Can be applied to any AI ethical dilemma
    """
    def __init__(self):
        # Store the 6 steps as data (a list of dicts) rather than code:
        # the same loop can then print, plot, or apply any step uniformly.
        self.steps = [
            {'step': 1, 'name_en': 'Identify the Problem',
             'description_en': 'Clearly define the ethical issue or dilemma',
             'questions': ['What is the AI system supposed to do?',
                           'What ethical concerns might arise?',
                           'Who are the stakeholders?']},
            {'step': 2, 'name_en': 'Gather Information',
             'description_en': 'Collect relevant facts and context',
             'questions': ['What data will be used?',
                           'How will the system be deployed?',
                           'What are the potential impacts?']},
            {'step': 3, 'name_en': 'Identify Stakeholders',
             'description_en': 'List all affected parties',
             'questions': ['Who will use the system?',
                           'Who might be affected?',
                           'Who has decision-making power?']},
            {'step': 4, 'name_en': 'Apply Ethical Frameworks',
             'description_en': 'Analyze using multiple ethical perspectives',
             'questions': ['What would utilitarianism suggest?',
                           'What would deontology require?',
                           'What rights are at stake?']},
            {'step': 5, 'name_en': 'Evaluate Options',
             'description_en': 'Consider different approaches and their consequences',
             'questions': ['What are the alternative approaches?',
                           'What are the trade-offs?',
                           'What are the risks and benefits?']},
            {'step': 6, 'name_en': 'Make Decision',
             'description_en': 'Choose the most ethical course of action',
             'questions': ['Which option best balances ethical considerations?',
                           'Can the decision be justified?',
                           'Is it transparent and accountable?']},
        ]
        print("Framework created with 6 steps!")
        print("   Each step has guiding questions to help you think through ethical decisions")

    def analyze_scenario(self, scenario_name, scenario_data):
        """Walk through all 6 steps for one scenario, printing the analysis."""
        print(f"\n{'='*80}")
        print(f"ETHICAL ANALYSIS: {scenario_name}")
        print(f"{'='*80}")
        for step in self.steps:
            print(f"\nStep {step['step']}: {step['name_en']}")
            print("-" * 60)
            print(f"Description: {step['description_en']}")
            print("Key Questions:")
            for i, question in enumerate(step['questions'], 1):
                print(f"  {i}. {question}")
            if scenario_data and step['step'] in scenario_data:
                print("Analysis:")
                print(f"  {scenario_data[step['step']]}")

# Create framework instance to demonstrate
print("Creating framework instance...")
framework = EthicalDecisionFramework()
print("Framework ready to use!")
```

## Part 3: Visualizing the Framework

### 📚 Prerequisites (What You Need First)
- ✅ **Framework defined** (from Part 2) - Understanding the 6-step framework
- ✅ **Visualization libraries** (from Part 1) - Understanding matplotlib

### 🔗 Relationship: What This Builds On
This builds on Part 2 - we now visualize the framework we created!
- Builds on: Framework definition, matplotlib knowledge
- Shows: How the 6 steps flow together

### 📖 The Story
**Before visualization**: We have a framework but can't easily see the flow.
**After visualization**: We can see the complete decision-making process as a flowchart!

```python
# Step 3: Create decision-making flowchart visualization
# This shows the 6-step process as a visual flowchart

print("\n" + "="*80)
print("📊 CREATING DECISION-MAKING FLOWCHART")
print("="*80)

fig, ax = plt.subplots(figsize=(8, 10))
ax.set_xlim(0, 10)
ax.set_ylim(1.5, 14)
ax.axis('off')

# Color code by role: green = start, blue = analysis steps, orange = judgment steps, purple = end.
# Color tells the reader which steps need human ethical judgment (orange) vs fact-gathering (blue).
start_color, process_color = '#2ecc71', '#3498db'
decision_color, end_color = '#f39c12', '#9b59b6'

# Lay out the boxes on an invisible grid: same x (centered column), decreasing y (top to bottom flow).
steps = [
    {'x': 5, 'y': 13,   'w': 2,   'color': start_color,    'text': 'Start'},
    {'x': 5, 'y': 11.5, 'w': 2.5, 'color': process_color,  'text': '1. Identify\nProblem'},
    {'x': 5, 'y': 10,   'w': 2.5, 'color': process_color,  'text': '2. Gather\nInformation'},
    {'x': 5, 'y': 8.5,  'w': 2.5, 'color': process_color,  'text': '3. Identify\nStakeholders'},
    {'x': 5, 'y': 7,    'w': 2.5, 'color': decision_color, 'text': '4. Apply Ethical\nFrameworks'},
    {'x': 5, 'y': 5.5,  'w': 2.5, 'color': decision_color, 'text': '5. Evaluate\nOptions'},
    {'x': 5, 'y': 4,    'w': 2.5, 'color': process_color,  'text': '6. Make\nDecision'},
    {'x': 5, 'y': 2.5,  'w': 2,   'color': end_color,      'text': 'End'},
]
H = 0.6  # box height
# Draw each step as a rounded box with its label - one patch + one text per step.
for s in steps:
    box = FancyBboxPatch((s['x'] - s['w']/2, s['y'] - H/2), s['w'], H,
                         boxstyle="round,pad=0.1", edgecolor='black',
                         facecolor=s['color'], linewidth=2, alpha=0.8)
    ax.add_patch(box)
    ax.text(s['x'], s['y'], s['text'], ha='center', va='center', fontsize=9,
            fontweight='bold',
            color='white' if s['color'] != '#f39c12' else 'black')
# Connect consecutive boxes with arrows so the eye follows the mandatory order of the process.
for i in range(len(steps) - 1):
    arrow = FancyArrowPatch((steps[i]['x'], steps[i]['y'] - H/2),
                            (steps[i+1]['x'], steps[i+1]['y'] + H/2),
                            arrowstyle='->', mutation_scale=20, linewidth=2,
                            color='black')
    ax.add_patch(arrow)
ax.set_title('Ethical Decision-Making Framework for AI',
             fontsize=14, fontweight='bold', pad=20)
plt.tight_layout()
plt.show()
print("Flowchart displayed - the 6 steps flow top to bottom from Start to End.")
```

## 🎯 Solving the Autonomous Vehicle Problem

**Remember**: In the previous notebook, we saw that different frameworks gave different answers to the autonomous vehicle problem. Now let's use our **decision-making framework** to resolve this conflict!

**The Problem**: Autonomous vehicle must choose between:
- Option A: Swerve and hit wall (kill passenger)
- Option B: Continue forward (kill pedestrians)  
- Option C: Try to minimize harm (risk both)

**Framework Conflicts**:
- Utilitarianism → Option A
- Virtue Ethics → Option C
- Deontology → No clear answer
- Rights-Based → No clear answer
- Care Ethics → Option C

**How the Decision-Making Framework Solves This**: We'll go through all 6 steps systematically to resolve the conflict!

```python
# Step 4: Solve the autonomous vehicle problem using the decision-making framework
# This demonstrates how the framework resolves framework conflicts

# Map step number -> our written analysis for that step of the framework.
# This is the 'filled-in worksheet' for the autonomous-vehicle dilemma from Notebook 1.
av_scenario = {
    1: "An autonomous vehicle must choose between options that all cause harm "
       "(hit wall / continue forward / swerve to minimize harm).",
    2: "Sensor data is reliable; outcomes are probabilistic. Passengers chose the "
       "vehicle; pedestrians did not choose the risk.",
    3: "Passenger, pedestrians, manufacturer, regulators, and society "
       "(future riders whose trust depends on this policy).",
    4: "Utilitarianism -> Option A (fewest total deaths). Virtue/Care Ethics -> "
       "Option C (responsible effort to protect everyone). Deontology and "
       "Rights-Based -> no option fully respects every duty/right.",
    5: "Option A minimizes expected deaths but sacrifices the passenger by design. "
       "Option B protects the passenger but ignores pedestrians. Option C spreads "
       "risk and reflects a defensible, careful policy.",
    6: "Choose Option C: it is justifiable to every stakeholder group, survives "
       "the transparency test (it can be published as policy), and balances the "
       "frameworks instead of letting one override the others."
}

# Walk the scenario through all 6 steps - the framework prints each step next to our analysis.
framework.analyze_scenario("Autonomous Vehicle Dilemma", av_scenario)

# Summarize the key teaching point: the process (not any single framework) broke the deadlock.
print("\n" + "="*80)
print("HOW THE FRAMEWORK RESOLVED THE CONFLICT")
print("="*80)
print("""
Individual frameworks disagreed (Unit 1, Notebook 1). The 6-step process
resolved the deadlock because it:
  1. Forced us to state the problem and gather facts BEFORE judging
  2. Surfaced ALL stakeholders (not just passenger vs pedestrians)
  3. Used the frameworks as inputs (Step 4), not as final verdicts
  4. Made the final decision explicit and justifiable (Step 6)
Decision reached: Option C, with a written justification we can defend.
""")
```
