## 📌 The case: Germany wrote an answer to the trolley problem into national policy

The scenario above stopped being a thought experiment in **June 2017**, when the German
Federal Ministry of Transport's **Ethics Commission on Automated and Connected Driving**
(chaired by former constitutional-court judge Udo Di Fabio) published **20 ethical rules**
for automated vehicles — the first national ethics code for self-driving cars.

Its **Rule 9** does not answer the dilemma. It rules one whole family of answers *out*:

> In unavoidable accident situations, **any distinction based on personal features
> (age, gender, physical or mental constitution) is strictly prohibited**, and victims
> may not be offset against one another.

Read that against Options A, B and C. "One death is better than five" — the clean
utilitarian arithmetic — is exactly the reasoning the Commission forbade. A regulator did
not choose the framework with the most comfortable answer. It chose a framework, in public,
and then accepted the consequences of that choice.

And the public does not agree with it. The MIT **Moral Machine** experiment
(Awad et al., *Nature* 563:59–64, 2018) put these same scenarios online and collected
**about 40 million decisions from people in 233 countries and territories**. The
preferences did not converge on one human morality; they separated into **three cultural
clusters** whose differences track institutions and cultural traits. People in different
places genuinely want different cars.

**The WHY — what goes wrong without frameworks.** Germany had to write Rule 9 because
engineers were already shipping the decision. Somebody's collision-avoidance code already
implies an answer to "who does the car hit", whether or not anyone chose it deliberately.
Without a named framework, that answer is made once, on a Tuesday, by whoever was editing
that file — and afterwards nobody can say whether it was defensible, only that it was the
one in the binary. Frameworks do not make the choice easy. They make it **arguable**, which
is the precondition for anybody being held to it.

## Part 3: Comparing Ethical Frameworks

### 📚 Prerequisites (What You Need First)
- **Frameworks defined** (from Part 2) - Understanding the five ethical frameworks
- **Visualization libraries** (from Part 1) - Understanding matplotlib

### 🔗 Relationship: What This Builds On
This builds on Part 2 - we now visualize and compare the frameworks we defined!
- Builds on: Framework definitions, matplotlib
- Shows: How frameworks compare in strengths, weaknesses, and practicality

### 📖 The Story
**Before visualization**: We have framework definitions but can't easily see how they compare.
**After visualization**: We can see at a glance which frameworks are strongest, which have weaknesses, and which are most practical!

```python
# Step 3: Create framework comparison visualization
# This shows us how frameworks compare in strengths, weaknesses, and practicality

# BEFORE: We have framework data but no visual comparison
# AFTER: We'll have a diverging bar chart in which "up" always means "counts in
#        the framework's favour" and "down" always means "counts against it"

print("\n" + "="*80)
print("CREATING FRAMEWORK COMPARISON VISUALIZATION")
print("="*80)
print("\nThis visualization shows, on the course authors' own 1-10 teaching scores:")
print("  - Strengths    (plotted UP)   : how well the framework works")
print("  - Practicality (plotted UP)   : how easy it is to apply")
print("  - Weaknesses   (plotted DOWN) : how problematic it is\n")


def create_framework_comparison():
    """
    Draw a diverging bar chart comparing the five ethical frameworks.

    HOW IT WORKS:
    1. Pull the three scores for each framework out of the `frameworks` dictionary
    2. Plot strengths and practicality ABOVE the zero line (good), weaknesses BELOW it (bad)
    3. Label every bar with its exact score and mark the framework that dominates
    4. Show the chart inline

    WHY plot weaknesses downward: in the usual grouped bar chart all three bars point up,
    so a tall red "weakness" bar looks like an achievement. Flipping it below the axis makes
    one direction mean one thing, which is the whole job of a chart axis.

    WHEN to use: after defining the frameworks - to compare them at a glance.
    """
    # Extract data: pull names and the three author-assigned scores out of the dictionary
    names = [f['name_en'] for f in frameworks.values()]
    strengths = [f['strength'] for f in frameworks.values()]
    weaknesses = [f['weakness'] for f in frameworks.values()]
    practicality = [f['practicality'] for f in frameworks.values()]

    x = np.arange(len(names))   # one x slot per framework
    width = 0.34                # two bars sit above the line, one below it

    fig, ax = plt.subplots(figsize=(11, 6))

    # Bars ABOVE the zero line: things that count in the framework's favour
    b_str = ax.bar(x - width/2, strengths, width, label='Strengths (higher = better)',
                   color='#2ecc71', edgecolor='black', linewidth=0.6)
    b_pra = ax.bar(x + width/2, practicality, width, label='Practicality (higher = easier to apply)',
                   color='#3498db', edgecolor='black', linewidth=0.6)
    # Bars BELOW the zero line: the one score where a bigger number is worse.
    # We negate the value so its direction on the page matches its meaning.
    b_wea = ax.bar(x, [-w for w in weaknesses], width, label='Weaknesses (longer downward = worse)',
                   color='#e74c3c', edgecolor='black', linewidth=0.6)

    # Write the true (positive) score on every bar so the reader never has to guess
    for bars, vals in ((b_str, strengths), (b_pra, practicality), (b_wea, weaknesses)):
        for bar, v in zip(bars, vals):
            h = bar.get_height()
            ax.text(bar.get_x() + bar.get_width()/2, h + (0.25 if h >= 0 else -0.25), str(v),
                    ha='center', va='bottom' if h >= 0 else 'top', fontsize=9, fontweight='bold')

    ax.axhline(0, color='black', linewidth=1.2)               # the zero line separates for/against
    ax.set_xticks(x)
    ax.set_xticklabels(names, fontsize=10)
    ax.set_xlabel('Ethical framework', fontsize=11, fontweight='bold')
    ax.set_ylabel('Author-assigned teaching score (1-10)\ndown = against   |   up = in favour',
                  fontsize=10, fontweight='bold')
    ax.set_ylim(-11.5, 14.8)   # headroom so the legend and the callout never sit on a bar
    ax.set_yticks([-10, -5, 0, 5, 10])
    ax.set_yticklabels(['10', '5', '0', '5', '10'])           # show magnitudes, not negative numbers
    ax.grid(axis='y', alpha=0.3, linestyle='--')
    ax.set_title('Ethical frameworks compared - author-assigned teaching scores, not measurements',
                 fontsize=12, fontweight='bold', pad=14)
    ax.legend(fontsize=9, loc='upper left', framealpha=0.95, ncol=1)

    # Mark the framework that comes out ahead on ALL THREE scores, because the caption
    # below the figure asks you to be suspicious of exactly that result.
    best = max(range(len(names)),
               key=lambda i: (strengths[i] - weaknesses[i] + practicality[i]))
    ax.annotate('best on all three\n(be suspicious - see caption)',
                xy=(best, strengths[best] + 0.75), xytext=(best, 12.4),
                ha='center', va='center', fontsize=9, color='#7f1d1d',
                arrowprops=dict(arrowstyle='->', color='#7f1d1d'))

    plt.tight_layout()
    plt.show()
    return names[best]


print("Creating comparison chart...")
winner = create_framework_comparison()
print(f"Framework comparison visualization complete. Dominates on all three scores: {winner}")
```

### 👁️ Read the chart before you trust it

**What you should see.** Five groups of bars. In each group two bars rise above the black
zero line — green *Strengths*, blue *Practicality*, both "higher is better" — and one red
*Weaknesses* bar hangs below it, because on that score a bigger number is worse. Putting it
below the line is the point: in an ordinary grouped bar chart all three point upward and a
tall red bar looks like an achievement.

**What it proves — and the trap.** Read the fourth group, **Rights-Based Ethics**: tallest
green bar (9), tallest blue bar (8), shortest red bar (4). On these numbers it beats every
other framework on every axis. If a 2,400-year-old argument between five moral traditions
could be settled by a bar chart, it would have been settled. What the picture actually
measures is *the course authors' opinion*, which is why the axis says so. Question 3 of the
💬 Discuss section below asks you to attack this chart; the chart is drawn honestly enough
that you can.

## 🎯 Applying Frameworks to the Autonomous Vehicle Problem

Now let's apply our frameworks to the **exact problem** we started with - the autonomous vehicle dilemma!

**The Scenario**: An autonomous vehicle must choose between:
- **Option A**: Swerve and hit a wall, potentially killing the passenger
- **Option B**: Continue forward, potentially killing the pedestrians
- **Option C**: Try to minimize harm, but risk both

Let's see how each framework evaluates this problem:

```python
# Step 5: Apply frameworks to the autonomous vehicle problem
# This demonstrates how different frameworks give different answers to the same problem

# BEFORE: We have frameworks but haven't seen them applied to our original problem
# AFTER: We'll see how each framework evaluates the autonomous vehicle dilemma

print("\n" + "="*80)
print("🚗 APPLYING FRAMEWORKS TO AUTONOMOUS VEHICLE PROBLEM")
print("="*80)
print("\nScenario: Autonomous vehicle must choose between:")
print("  Option A: Swerve and hit wall (kill passenger)")
print("  Option B: Continue forward (kill pedestrians)")
print("  Option C: Try to minimize harm (risk both)\n")

# Each framework's reading of the three options, kept at module scope (not hidden inside the
# function) so that the next cell can PLOT the same verdicts it prints here - one source of truth.
framework_evaluations = {
    'Utilitarianism': {  # Framework 1: Utilitarian perspective
        'option_a': 'Kill 1 passenger - minimize total harm',
        'option_b': 'Kill multiple pedestrians - maximize harm',
        'option_c': 'Uncertain outcome - hard to calculate',
        'recommendation': 'Option A (minimize total deaths)',
        'lands_on': 'Option A',
        'reasoning': 'Maximize overall good by minimizing total deaths'
    },
    'Deontology': {  # Framework 2: Deontological perspective
        'option_a': 'Violates duty to protect passenger',
        'option_b': 'Violates duty to protect pedestrians',
        'option_c': 'Uncertain - violates clear duties',
        'recommendation': 'No clear answer - all violate duties',
        'lands_on': 'No clear answer',
        'reasoning': 'Moral rules conflict - no perfect solution'
    },
    'Rights-Based': {  # Framework 3: Rights-based perspective
        'option_a': 'Violates passenger right to life',
        'option_b': 'Violates pedestrian rights to life',
        'option_c': 'Uncertain - may violate all rights',
        'recommendation': 'No clear answer - all violate rights',
        'lands_on': 'No clear answer',
        'reasoning': "All options violate someone's right to life"
    },
    'Virtue Ethics': {  # Framework 4: Virtue ethics perspective
        'option_a': 'Shows courage but not justice',
        'option_b': 'Shows no virtue - reckless',
        'option_c': 'Shows prudence - trying to minimize harm',
        'recommendation': 'Option C (most virtuous)',
        'lands_on': 'Option C',
        'reasoning': 'Shows prudence and care for all lives'
    },
    'Care Ethics': {  # Framework 5: Relationship-focused perspective
        'option_a': 'Abandons passenger relationship',
        'option_b': 'Abandons pedestrian relationships',
        'option_c': 'Tries to care for all relationships',
        'recommendation': 'Option C (cares for all)',
        'lands_on': 'Option C',
        'reasoning': 'Considers relationships with all affected parties'
    }
}


def analyze_autonomous_vehicle_problem():
    """
    Analyze the autonomous vehicle problem using different ethical frameworks.
    
    HOW IT WORKS:
    1. Define how each framework evaluates the three options
    2. Show that different frameworks lead to different recommendations
    3. Demonstrate the limitation: frameworks conflict!
    
    ⏰ WHEN to use: After learning frameworks - see them applied to real problem
    💡 WHY use: Shows that frameworks give different answers - creates need for decision-making process
    """
    # The evaluations live at module scope above - the function only reports them.
    # Print evaluations: Show how each framework evaluates the problem
    print("Framework Evaluations:\n")
    for framework, evaluation in framework_evaluations.items():  # Loop through frameworks: Process each framework
        print(f"{framework}:")  # Framework header: Show framework name
        print("-" * 60)  # Separator: Visual divider
        print(f"  Option A: {evaluation['option_a']}")  # Option A evaluation: Framework's view
        print(f"  Option B: {evaluation['option_b']}")  # Option B evaluation: Framework's view
        print(f"  Option C: {evaluation['option_c']}")  # Option C evaluation: Framework's view
        print(f"  Recommendation: {evaluation['recommendation']}")  # Recommendation: What framework suggests
        print(f"  Reasoning: {evaluation['reasoning']}")  # Reasoning: Why framework recommends this
        print()  # Blank line: Spacing
    
    # Show the conflict: Different frameworks give different answers
    print("\n" + "="*80)
    print("⚠️ THE LIMITATION: Frameworks Give Different Answers!")
    print("="*80)
    print("\nNotice that:")
    print("  - Utilitarianism recommends Option A")
    print("  - Virtue Ethics recommends Option C")
    print("  - Deontology and Rights-Based find no clear answer")
    print("  - Care Ethics recommends Option C")
    print("\n🚫 This is a DEAD END: We have frameworks, but they conflict!")
    print("   We need a way to resolve these conflicts...")
    print("   → This leads us to the next notebook: Ethical Decision-Making!")

# Run the analysis
print("Analyzing autonomous vehicle problem with frameworks...")
analyze_autonomous_vehicle_problem()
print("\n✅ Framework analysis complete!")
```

## 📊 The experiment of this notebook, in one picture

The cell above printed five verdicts on **one unchanged dilemma**. Printed one after
another they read like five paragraphs. Put on a single axis they read like a result:
the frameworks do not land in the same place.

```python
# Step 5b: Plot where each framework LANDS on the same unchanged dilemma
# WHY this figure: the notebook's central claim is "hold the situation fixed, change the
# ethical lens, and watch the verdict move". A printed list cannot show movement; a single
# axis with four possible landing places can. The data plotted here is exactly the
# `framework_evaluations` dictionary printed above - nothing new is invented.

# The four places a verdict can land. Option B is included even though nothing lands there,
# because an empty column is itself a finding worth seeing.
columns = ['Option A\nswerve into wall\n(passenger dies)',
           'Option B\ncontinue forward\n(pedestrians die)',
           'Option C\nminimise harm\n(risk both)',
           'No clear answer\n(all options violate\na duty or a right)']
col_key = ['Option A', 'Option B', 'Option C', 'No clear answer']

# Group the frameworks by the column they land in, keeping the printed order.
landing = {k: [] for k in col_key}
for name, ev in framework_evaluations.items():
    landing[ev['lands_on']].append(name)

fig, ax = plt.subplots(figsize=(11, 4.6))
colors = {'Option A': '#2ecc71', 'Option B': '#bdc3c7',
          'Option C': '#3498db', 'No clear answer': '#e67e22'}

# Draw one labelled box per framework, stacked inside its column.
for j, key in enumerate(col_key):
    members = landing[key]
    for k, name in enumerate(members):
        ax.add_patch(plt.Rectangle((j - 0.40, 0.55 + k * 0.85), 0.80, 0.68,
                                   facecolor=colors[key], edgecolor='black',
                                   linewidth=1.1, alpha=0.9))
        ax.text(j, 0.89 + k * 0.85, name, ha='center', va='center',
                fontsize=10, fontweight='bold',
                color='white' if key != 'Option B' else 'black')
    # Tally under each column so the split is countable, not just visible.
    ax.text(j, 0.18, f"{len(members)} of 5 frameworks", ha='center', va='center',
            fontsize=10, fontweight='bold',
            color='#7f8c8d' if not members else 'black')

ax.axhline(0.45, color='black', linewidth=1.0)
ax.set_xlim(-0.65, 3.65)
ax.set_ylim(0, 2.45)   # just enough for two stacked boxes - no dead space
ax.set_xticks(range(4))
ax.set_xticklabels(columns, fontsize=9.5)
ax.set_yticks([])
ax.set_xlabel('Where the framework lands on the SAME unchanged dilemma',
              fontsize=11, fontweight='bold', labelpad=8)
ax.set_title('One dilemma, five ethical lenses, three different answers',
             fontsize=13, fontweight='bold', pad=12)
for side in ('top', 'right', 'left'):
    ax.spines[side].set_visible(False)
plt.tight_layout()
plt.show()

# Print the same split as text, so the figure and the numbers cannot drift apart.
print("Landing places (counted from the dictionary that was printed above):")
for key in col_key:
    who = ", ".join(landing[key]) if landing[key] else "- nobody -"
    print(f"  {key:<16} : {len(landing[key])} -> {who}")
print()
print("Three of the four columns are occupied. That is the conflict, and it is why the")
print("next notebook is about a decision PROCESS rather than a better framework.")
```

### 👁️ Read the figure

**What you should see.** Four columns, one per possible verdict, and five labelled boxes
distributed across them: **Utilitarianism alone in Option A**, **Virtue Ethics and Care
Ethics together in Option C**, **Deontology and Rights-Based together in "No clear
answer"** — and **Option B empty**. The count printed under each column matches the boxes
above it.

**What it proves.** The situation never changed between boxes; only the lens did. Three
occupied columns out of four is the failure that ends this notebook: you cannot settle the
dilemma by picking the "right" framework, because five defensible frameworks give three
different answers. That is the exact gap Notebook 02 fills with a six-step *process*.

**The empty column is worth a sentence of its own.** No framework here endorses Option B —
plough on — yet a vehicle whose collision code was never given a policy does Option B by
default. That is the point of the Germany case at the top of this notebook: refusing to
choose is not neutral, it is a choice that nobody argued for.

## 🚫 When Ethical Frameworks Hit a Dead End

### The Limitation We Discovered

We've learned five ethical frameworks, and we've seen how they can evaluate AI scenarios. **But there's a critical problem:**

**Different frameworks give different answers to the same ethical question!**

As we saw with the autonomous vehicle problem:
- **Utilitarianism** recommends Option A (minimize total deaths)
- **Virtue Ethics** recommends Option C (most virtuous approach)
- **Deontology** and **Rights-Based** find no clear answer (all options violate principles)
- **Care Ethics** recommends Option C (cares for all relationships)

### Why This Is a Problem

When frameworks conflict, we're back to the original problem:
- **How do we decide** which framework to follow?
- **How do we resolve** conflicts between frameworks?
- **How do we make** a final ethical decision when frameworks disagree?

### The Solution: Ethical Decision-Making Framework

We need a **systematic process** to:
1. Apply multiple frameworks
2. Resolve conflicts between them
3. Make a final, justified decision

## ⚠️ Where this breaks

**The assumption that has to hold:** that disagreement between frameworks is *productive* —
that seeing one problem through five lenses sharpens the decision. Three situations where it
is not.

- **When the frameworks are used to shop for a verdict.** Five frameworks hand a motivated
  team five defensible-sounding conclusions, and "we conducted an ethical analysis" becomes
  a way to launder a decision already taken. Guard against it by committing to the framework
  and its weighting **before** you see which product it kills.
- **When author-assigned numbers get mistaken for measurements.** The strength, weakness and
  practicality scores in this notebook are integers the course authors typed, and the bar
  chart makes them look like data. Utilitarianism does not *have* a strength of 8. Nothing
  in this notebook was measured. Measurement starts in Unit 2, on a real passenger manifest.
- **When the real failure is not a dilemma at all.** The trolley framing assumes a perceiving
  system that correctly sees two bad options and must choose. In the 2018 Uber ATG fatality
  (next notebook) the system never got that far — it saw *something* for 5.6 seconds and
  could not decide what. Most real AI harm is banal: a mis-set threshold, a missing test, a
  safety feature switched off for convenience. No amount of framework comparison finds those.

**The cheaper alternative when this does not fit:** for routine product decisions, a
one-page harms checklist with a named owner (Unit 4's RACI matrix) resolves more cases,
faster, than a five-framework analysis. Reserve the frameworks for decisions where
reasonable people genuinely disagree about the **goal**, not the implementation.

**The same reasoning appears elsewhere.** Medical ethics committees run the same
multi-framework argument over triage protocols; central banks run it over whether to
optimise for employment or price stability; Islamic finance boards run it over whether a
contract's *form* or its *effect* decides permissibility. In each case the institutional
fix is identical to Germany's: pick a rule in public, write it down, and live with it.
