## The $62 million proposal

The **University of Texas MD Anderson Cancer Center** spent about four years building the **Oncology Expert Advisor** on IBM Watson: a system that would read a patient's record plus the oncology literature and recommend a treatment. A **48-page University of Texas System audit report** — dated November 2016, made public in early 2017 — ended it. The audit records what the money bought: at least **$62 million** spent; an original contract to deliver the leukaemia product in **six months for $2.4 million**, extended **twelve times** to total fees of **$39.2 million**; a procurement that had gone around MD Anderson's normal IT process. And the finding that matters most here — as of September 2016 the system **was not in clinical use** and had never been piloted outside MD Anderson. IBM's own agreement stated that the system "is not ready for human investigational or clinical use, and its use in the treatment of patients is prohibited", except as needed to test and evaluate it.

Nobody on that project was incompetent, and the technology was not the problem. The failure is visible entirely in the documents this notebook teaches you to write: the objective was never narrowed to something a fixed budget could reach, *done* was never defined as a number anyone could check, and there was no point at which somebody was obliged to say **stop**.

### What goes wrong without this

A capstone with no written scope does not fail loudly in week 1. It fails quietly in week 9. You are still "improving" the model; the dataset you assumed was available turned out to need a licence; one extra feature looked cheap and cost two weeks; and there is no document anyone can point at to settle whether that feature was ever part of the project. That is **scope creep**, and its cost is always paid in the same place — the report and the demo, the last two deliverables, get compressed into whatever days are left. In AIAT 126 those two are gates 5 and 6: **30 of the 100 points**.

The proposal is the cheapest insurance in the course. It is a few hours of writing that turns *"no, that is out of scope"* into a sentence you are allowed to say — to your supervisor, to a helpful classmate with a great idea, and to yourself at 2am in week 9.

### The same reasoning, three other domains

- **Zillow Offers, November 2021** — an automated home-pricing model bought houses at prices above what Zillow could later sell them for. Q3 2021 carried a **$304.4 million** inventory write-down, the programme was wound down and roughly **25%** of the workforce went with it. Zillow's own explanation, in the shareholder letter filed the same day, was that it had "been unable to accurately forecast future home prices at different times in both directions by much more than we modeled as possible" — the success criterion was never stress-tested against the market it would actually meet.
- **Google Flu Trends, 2013** — a system built to predict the CDC's influenza reports ended up predicting **more than double** the CDC's actual proportion of flu-related doctor visits (Lazer et al., *Science*, 2014). The objective was perfectly well defined; the assumption underneath it expired without anyone noticing.
- **The Netflix Prize, 2009** — a $1 million competition, won with a 10% RMSE improvement, whose winning ensemble Netflix never put into production because "the additional accuracy gains that we measured did not seem to justify the engineering effort" (Netflix, 2012). The metric was hit and the goal was still missed.

Different industries, one pattern: **the goal that was written down and the goal that mattered were not the same goal.** That gap is closed on paper, in this unit, or it is not closed at all.

### The $62 million proposal, drawn

The section above gives you the audit's numbers in a sentence: **$2.4 million**, **$39.2 million**, **$62 million**, **six months**, **four years**. Read as prose they are five similar-looking figures. Put on one axis they are not similar at all, and the shape of the failure becomes visible before you write a single line of your own proposal.

Everything plotted below is a **reported figure from the November 2016 University of Texas System audit** (reference 4 at the end of this notebook). Nothing here is simulated.

```python
# WHAT: the opening case, drawn - the audit's dollar figures on one axis, and its two
# schedule figures on another. WHY draw it: a reader skims "$2.4 million" and "$62 million"
# as two numbers of the same kind. On a shared axis the entire original contract turns out
# to be a sliver, which is the whole point of the case.
# These are REPORTED FIGURES from the audit (reference 4), not a simulation.

# (label, US$ millions, colour) - written in the order the money was committed
MDA_SPEND = [
    ("Contracted: leukaemia product\nin 6 months", 2.4, "#4C72B0"),
    ("Paid to IBM after 12\ncontract extensions", 39.2, "#DD8452"),
    ("All external firms,\nthrough 31 Aug 2016", 62.1, "#B03030"),
]
# (label, months, colour) - the same project measured in calendar instead of dollars
MDA_MONTHS = [
    ("Contracted\nto deliver", 6, "#4C72B0"),
    ("Elapsed\n(about 4 years)", 48, "#B03030"),
]

fig, (ax_money, ax_time) = plt.subplots(
    1, 2, figsize=(11.8, 4.0), gridspec_kw={"width_ratios": [1.85, 1.0]})

# --- left panel: the money ------------------------------------------------------
labels = [r[0] for r in MDA_SPEND][::-1]      # reversed so the contract sits at the TOP
values = [r[1] for r in MDA_SPEND][::-1]
colours = [r[2] for r in MDA_SPEND][::-1]
plan_usd = MDA_SPEND[0][1]                    # the original contract is the baseline

ax_money.barh(range(len(values)), values, height=0.5, color=colours)
for i, v in enumerate(values):
    # The multiple is COMPUTED from the two reported numbers, never typed in by hand
    tag = "the plan" if v == plan_usd else f"{v / plan_usd:.1f}× the plan"
    ax_money.text(v + 1.5, i, f"${v:.1f}M  ({tag})", va="center", fontsize=9.5)
# The dashed line is the entire original contract: everything to its right was never planned
ax_money.axvline(plan_usd, color="#4C72B0", ls="--", lw=1.4)
ax_money.annotate("the whole\noriginal contract", xy=(plan_usd, 2.45), xytext=(21, 2.45),
                  fontsize=8, color="#4C72B0", va="center",
                  arrowprops=dict(arrowstyle="->", color="#4C72B0", lw=1.0))
ax_money.set_yticks(range(len(labels)))
ax_money.set_yticklabels(labels, fontsize=8.5)
ax_money.set_xlim(0, 96)
ax_money.set_ylim(-0.6, 2.9)
ax_money.set_xlabel("US$ millions paid")
ax_money.set_title("Money: one contract, twelve extensions", fontsize=10)

# --- right panel: the calendar --------------------------------------------------
t_labels = [r[0] for r in MDA_MONTHS][::-1]
t_values = [r[1] for r in MDA_MONTHS][::-1]
t_colours = [r[2] for r in MDA_MONTHS][::-1]
plan_months = MDA_MONTHS[0][1]

ax_time.barh(range(len(t_values)), t_values, height=0.42, color=t_colours)
for i, v in enumerate(t_values):
    tag = "the plan" if v == plan_months else f"{v / plan_months:.1f}× the plan"
    ax_time.text(v + 1.5, i, f"{v} months  ({tag})", va="center", fontsize=9.5)
ax_time.set_yticks(range(len(t_labels)))
ax_time.set_yticklabels(t_labels, fontsize=8.5)
ax_time.set_xlim(0, 90)
ax_time.set_ylim(-0.6, 1.9)
ax_time.set_xlabel("months")
ax_time.set_title("Calendar: six months, about four years", fontsize=10)

for ax in (ax_money, ax_time):
    ax.grid(axis="x", alpha=0.25)
    ax.spines[["top", "right"]].set_visible(False)

fig.suptitle("MD Anderson / IBM Watson \"Oncology Expert Advisor\": what was planned, "
             "and what was spent", fontsize=11.5, y=0.99)
# The outcome is not a bar, because there is no quantity to draw - it is stated instead
fig.text(0.5, 0.135, "Delivered into clinical use: NONE. The audit found that as of "
                     "September 2016 the system was not in clinical use\n"
                     "and had never been piloted outside MD Anderson.",
         ha="center", fontsize=9, color="#B03030", weight="bold")
fig.text(0.5, 0.035, "Nested totals: the 39.2M of IBM fees is PART OF the 62.1M - the three "
                     "bars are levels of the same spend in US dollars, not three separate costs.\n"
                     "Source: University of Texas System audit, November 2016 (reference 4).",
         ha="center", fontsize=7.5, color="#444444", style="italic")
fig.tight_layout(rect=[0, 0.20, 1, 0.94])
plt.show()

# The same numbers as a table - a picture you cannot re-read as numbers is half a result
case_table = pd.DataFrame(
    [{"item": lab.replace("\n", " "), "US$M": v, "x_of_contract": round(v / plan_usd, 1)}
     for lab, v, _ in MDA_SPEND]
    + [{"item": lab.replace("\n", " ") + " (months)", "US$M": None,
        "x_of_contract": round(v / plan_months, 1)} for lab, v, _ in MDA_MONTHS])
print("=" * 72)
print("MD ANDERSON / IBM WATSON - reported figures, UT System audit (Nov 2016)")
print("=" * 72)
print(case_table.to_string(index=False))
print(f"\nCost overrun vs the original contract : {62.1 / 2.4:.1f}\u00d7")
print(f"Schedule overrun vs the original plan : {48 / 6:.1f}\u00d7")
print("Clinical use delivered                : none (audit finding, September 2016)")
```

**Read the figure.** Two panels, the same project measured two ways, both against the same baseline — **the original contract, drawn in blue**.

- **Left, the money.** The blue bar at the top is the *entire* original agreement: a leukaemia product, six months, $2.4M. It is so short that the dashed blue line marking it sits almost against the axis. Every dollar to the right of that line was spent on work nobody had committed to at the start — $39.2M in IBM fees alone (**16.3× the plan**) and $62.1M across all external firms (**25.9×**).
- **Right, the calendar.** The same two-bar comparison in months: **6 contracted, about 48 elapsed — 8× the plan.** Cost and schedule did not drift apart from each other; both ran an order of magnitude past what was written down.
- **The red sentence under the panels is the part that has no bar.** There is no quantity to draw for what was delivered, because the audit found the system was not in clinical use and had never been piloted outside MD Anderson. A figure with 25× the budget on it and nothing on the delivery side is the entire lesson.

Read the three money bars as **nested levels of one spend, not three separate costs** — IBM's $39.2M sits inside the $62.1M. The figure says so in its own footer, because summing the bars would invent a total this project never spent.

**What this has to do with your capstone.** Scale down by four orders of magnitude and this is a 14-week project that spends week 9 still "improving the model". Nothing on this figure required anyone to behave badly — it required only that no document said, in advance and in numbers, *what done looks like and when we stop*. ## Part 3: Project Timeline and Resource Planning

Create a realistic timeline for your graduation project.


```python
# Project timeline template: the 14-week capstone broken into 5 milestones,
# each with tasks and CONCRETE deliverables — a deliverable is what makes a
# milestone checkable instead of a vague intention.
# Anchor the plan at today; adapt the week ranges to your own calendar
start_date = datetime.now()

timeline = {
    "Week 1-2": {
        "milestone": "Project Proposal", "tasks": [
            "Define project scope and objectives",
            "Conduct initial literature review",
            "Write project proposal document"
        ],
        "deliverables": ["Project Proposal Document"]
    },
    "Week 3-5": {
        "milestone": "Data Collection & Preparation",
        "tasks": [
            "Identify and acquire datasets",
            "Data cleaning and preprocessing",
            "Exploratory data analysis",
            "Feature engineering"
        ],
        "deliverables": ["Clean dataset", "EDA notebook"]
    },
    "Week 6-10": {
        "milestone": "Model Development",
        "tasks": [
            "Model architecture design",
            "Training and validation",
            "Hyperparameter tuning",
            "Model evaluation"
        ],
        "deliverables": ["Trained models", "Evaluation results"]
    },
    "Week 11-12": {
        "milestone": "Deployment & Testing",
        "tasks": [
            "Model deployment",
            "System testing",
            "Performance validation"
        ],
        "deliverables": ["Deployed system"]
    },
    "Week 13-14": {
        "milestone": "Documentation & Presentation",
        "tasks": [
            "Write final report",
            "Create presentation",
            "Prepare demonstration"
        ],
        "deliverables": ["Final Report", "Presentation", "Video Demo"]
    }
}

# Print each milestone with a sample of its tasks and every deliverable
print("=" * 60)
print("Project Timeline Template (14 weeks)")
print("=" * 60)
for week, details in timeline.items():
    print(f"\n{week}: {details['milestone']}")
    print(f"  Tasks: {', '.join(details['tasks'][:2])}...")
    print(f"  Deliverables: {', '.join(details['deliverables'])}")

print("\n✅ Adjust timeline based on your project complexity!")
```

```python
# Comparative check: does the 14-week timeline above spend its weeks where AIAT 126
# actually awards its points? WHY compare the two: a plan is only credible when the
# calendar matches the rubric, and that comparison is arithmetic, not opinion.
# The gate weights are the real ones from PROJECT_GUIDELINES.md ("the six milestone gates").
# Gate 6 (the defense) is left out on purpose: it happens in its own session AFTER the
# build calendar, so charging it zero weeks here would be misleading rather than informative.
GATE_POINTS = {
    "Gate 1 - Proposal":       10,
    "Gate 2 - Design review":  10,
    "Gate 3 - Implementation": 30,
    "Gate 4 - Evaluation":     20,
    "Gate 5 - Final report":   15,
}

# Which gate does each milestone in `timeline` feed? Keys are read from the plan above,
# so this check can never drift away from the plan it is checking.
MILESTONE_TO_GATE = {
    "Project Proposal":              "Gate 1 - Proposal",
    "Data Collection & Preparation": "Gate 3 - Implementation",
    "Model Development":             "Gate 3 - Implementation",
    "Deployment & Testing":          "Gate 4 - Evaluation",
    "Documentation & Presentation":  "Gate 5 - Final report",
}


def weeks_in(label):
    """'Week 6-10' -> 5. The plan is written as ranges; the comparison needs a count."""
    span = label.replace("Week", "").strip()
    first, _, last = span.partition("-")
    return int(last or first) - int(first) + 1


weeks_per_gate = {gate: 0 for gate in GATE_POINTS}
unmapped = []
for week_label, details in timeline.items():
    gate = MILESTONE_TO_GATE.get(details["milestone"])
    if gate is None:
        unmapped.append(details["milestone"])          # a milestone nothing grades
    else:
        weeks_per_gate[gate] += weeks_in(week_label)

total_weeks = sum(weeks_per_gate.values())
plan_vs_rubric = pd.DataFrame({
    "points": pd.Series(GATE_POINTS),
    "weeks_planned": pd.Series(weeks_per_gate),
})
plan_vs_rubric["%_of_marks"] = (
    plan_vs_rubric["points"] / plan_vs_rubric["points"].sum() * 100).round(1)
plan_vs_rubric["%_of_calendar"] = (
    plan_vs_rubric["weeks_planned"] / total_weeks * 100).round(1)
plan_vs_rubric["gap_pp"] = (
    plan_vs_rubric["%_of_calendar"] - plan_vs_rubric["%_of_marks"]).round(1)

print("=" * 78)
print(f"PLAN vs RUBRIC - {total_weeks} planned weeks against "
      f"{plan_vs_rubric['points'].sum()} points (gates 1-5)")
print("=" * 78)
print(plan_vs_rubric.to_string())

# The two findings a reviewer would draw from this table, computed rather than asserted
starved = plan_vs_rubric[plan_vs_rubric["weeks_planned"] == 0]
print(f"\nGates with NO milestone anywhere in the timeline: "
      f"{list(starved.index) if len(starved) else 'none'} "
      f"({int(starved['points'].sum())} points)")
print(f"Milestones that feed no gate: {unmapped if unmapped else 'none'}")
print(f"Biggest over-spend  : {plan_vs_rubric['gap_pp'].idxmax():<24} "
      f"{plan_vs_rubric['gap_pp'].max():+.1f} percentage points of calendar vs marks")
print(f"Biggest under-spend : {plan_vs_rubric['gap_pp'].idxmin():<24} "
      f"{plan_vs_rubric['gap_pp'].min():+.1f} percentage points of calendar vs marks")
```

```python
# The table above, drawn. WHY plot it: this is a COMPARISON of two shares - marks and
# weeks - and a comparison is read far faster as two bars on one axis than as two columns
# of numbers. Nothing new is computed here: every value is read straight out of
# `plan_vs_rubric`, so the picture can never drift away from the table it illustrates.
gates = list(plan_vs_rubric.index)[::-1]      # reversed so Gate 1 ends up at the TOP
marks = plan_vs_rubric.loc[gates, "%_of_marks"].to_numpy()
cal = plan_vs_rubric.loc[gates, "%_of_calendar"].to_numpy()
gap = plan_vs_rubric.loc[gates, "gap_pp"].to_numpy()
y, h = range(len(gates)), 0.38

fig, ax = plt.subplots(figsize=(9.0, 4.0))
# Two bars per gate: what the rubric PAYS for it, and what the plan SPENDS on it
ax.barh([i + h / 2 for i in y], marks, height=h, color="#4C72B0", label="% of marks (rubric)")
ax.barh([i - h / 2 for i in y], cal, height=h, color="#DD8452", label="% of calendar (plan)")
for i, (m, c, g) in enumerate(zip(marks, cal, gap)):
    ax.text(m + 0.6, i + h / 2, f"{m:.1f}", va="center", fontsize=8, color="#4C72B0")
    ax.text(c + 0.6, i - h / 2, f"{c:.1f}", va="center", fontsize=8, color="#DD8452")
    # Red for the gaps big enough to argue about, grey for the ones inside a few points
    ax.text(63, i, f"{g:+.1f} pp", va="center", fontsize=9, weight="bold",
            color=("#B03030" if abs(g) >= 9 else "#666666"))
ax.text(63, len(gates) - 0.35, "calendar - marks", fontsize=8, color="#444444", style="italic")
ax.set_yticks(list(y))
ax.set_yticklabels(gates, fontsize=9)
ax.set_xlim(0, 78)
ax.set_xlabel("percent")
ax.axvline(0, color="black", lw=0.8)
ax.set_title("Where the plan spends weeks vs where the rubric awards points",
             fontsize=11, pad=18)
ax.spines[["top", "right"]].set_visible(False)
ax.grid(axis="x", alpha=0.25)
fig.legend(loc="lower center", ncol=2, fontsize=9, frameon=False, bbox_to_anchor=(0.5, -0.02))
fig.tight_layout(rect=[0, 0.07, 1, 1])
plt.show()
```

**Read the figure.** Every gate gets two bars: **blue = its share of the marks**, **orange = its share of the 14 weeks**. In a plan that matched its rubric, the two bars on each row would be about the same length. Four things are visible here that the numbers alone do not shout:

- **Gate 2 has no orange bar at all.** Not a short bar - a missing one. The design review is 11.8% of the marks and 0 of the 14 weeks. It is also the gate that `PROJECT_GUIDELINES.md` makes a hard stop before implementation may begin.
- **Gate 3 is the one row where orange overshoots blue by more than a couple of points**, and it overshoots by a lot: the orange bar reaches 57.1 while the blue stops at 35.3 (+21.8 pp).
- **Gates 4 and 5 lean the other way** - blue longer than orange. Evaluation and the final report are worth more than the calendar pays them.
- **The colour of the `pp` numbers on the right is the summary.** Three are printed in **red** — that is the code's own threshold for a gap of 9 percentage points or more, big enough that a supervisor will ask about it. Two are printed in **grey**, Gate 1 (+2.5 pp) and Gate 5 (−3.3 pp): near-matches, which is what a well-planned row looks like.

The picture does not tell you the plan is wrong. It tells you exactly which four rows you will be asked to justify, and in which direction.


### Reading the comparison: the template you just printed has a hole in it

The numbers above are not a criticism of you — they are a criticism of the generic 14-week plan in the previous cell, which is the plan most students copy. Two things the table says, and nothing more:

1. **Gate 2 (the design review, 10 points) gets zero weeks.** The timeline jumps straight from "Project Proposal" to "Data Collection & Preparation". But gate 2 is not just 10 points, it is a **hard stop**: `PROJECT_GUIDELINES.md` says implementation does not start until the design review passes. A calendar that allocates no time to it is a calendar that plans to fail a gate, or to skip it and rebuild later.
2. **Implementation takes 57.1% of the calendar for 35.3% of the marks** (+21.8 percentage points), while evaluation takes 14.3% of the calendar for 23.5% of the marks (−9.2). The build feels like the project; the rubric says the build is barely a third of it.

Neither number tells you the plan is *wrong*. Building genuinely does take longer than evaluating. What the table gives you is the sentence a supervisor will ask for: **"I am spending 57% of my weeks on 35% of my marks, deliberately, because X."** If you cannot finish that sentence, move a week.

**Fix your own copy before you submit it:** add a design-review milestone between the proposal and the data work, and re-run this cell.

### The plan under pressure: a four-frame storyboard

The "⚠️ Where this breaks" section below warns that **the 14-week timeline assumes the dataset works** - there is no milestone anywhere in it for *"the data turned out to be unusable"*. The next cell stops warning and draws it.

Read the frames as a **what-if, not a measurement.** They re-time the exact plan printed above under one assumption, and it is the assumption a real capstone lives under: **the week-14 deadline does not move.** So weeks that the data phase takes have to be taken from something else, and they come out of the *last* milestone first - because the last milestone is the one still unfinished when the clock runs out.


```python
# WHAT: re-plan the SAME timeline four times, letting the data phase run 0, 1, 2 and 3
# weeks late. WHY a grid of four small plans instead of one plot: the lesson is not any
# single frame, it is the direction of travel - watch the bottom bar from left to right.
# WHY only four frames: this must stay a classroom-cheap picture, and four is enough to
# see a bar disappear.
DEADLINE_WEEK = 14        # the assumption stated in the markdown above: the deadline is fixed

# Read the base plan straight out of `timeline`, so the storyboard cannot drift from the plan
base = []
for week_label, details in timeline.items():
    base.append((details["milestone"], weeks_in(week_label)))


def replan(slip_weeks):
    """Push the data phase `slip_weeks` late; a FIXED deadline absorbs the overflow.

    Returns [(milestone, start_week, duration_weeks), ...]. The overflow is taken from
    the END of the project backwards - the order in which work really does get cut.
    """
    duration = dict(base)
    duration["Data Collection & Preparation"] += slip_weeks
    overflow = sum(duration.values()) - DEADLINE_WEEK
    for milestone in ["Documentation & Presentation", "Deployment & Testing", "Model Development"]:
        if overflow <= 0:
            break
        taken = min(overflow, duration[milestone])
        duration[milestone] -= taken
        overflow -= taken
    rows, start = [], 1
    for milestone, _ in base:
        rows.append((milestone, start, duration[milestone]))
        start += duration[milestone]
    return rows


SLIPS = [0, 1, 2, 3]
COLORS = {"Project Proposal": "#8C8C8C", "Data Collection & Preparation": "#DD8452",
          "Model Development": "#4C72B0", "Deployment & Testing": "#55A868",
          "Documentation & Presentation": "#C44E52"}
names = [m for m, _ in base]

fig, axes = plt.subplots(1, len(SLIPS), figsize=(13.0, 3.4), sharey=True)
for ax, slip in zip(axes, SLIPS):
    for i, (milestone, start, dur) in enumerate(replan(slip)):
        row = len(names) - 1 - i                     # first milestone drawn at the top
        if dur > 0:
            ax.barh(row, dur, left=start - 1, height=0.62, color=COLORS[milestone])
        else:
            # A milestone squeezed to zero weeks has no bar to draw - mark it instead
            ax.plot([start - 1], [row], marker="x", ms=9, mew=2.2, color="#C44E52")
    ax.axvline(DEADLINE_WEEK, color="black", ls="--", lw=1.4)   # the deadline that never moves
    weeks_left = replan(slip)[-1][2]
    ax.set_title(f"data phase slips {slip} wk\nreport gets {weeks_left} wk"
                 + (" - NONE" if weeks_left == 0 else ""),
                 fontsize=9.5, color=("#B03030" if weeks_left == 0 else "black"))
    ax.set_xlim(0, 15.4)
    ax.set_xticks([0, 7, 14])
    ax.set_xlabel("week")
    ax.grid(axis="x", alpha=0.2)
    ax.spines[["top", "right"]].set_visible(False)
axes[0].set_yticks(range(len(names)))
axes[0].set_yticklabels(names[::-1], fontsize=8)
fig.suptitle("Same 14-week plan, a FIXED week-14 deadline, and a data phase that runs late\n"
             "(x marker = milestone squeezed to zero weeks)", fontsize=10.5)
fig.tight_layout(rect=[0, 0, 1, 0.86])
plt.show()

# The same four frames as a step table - a picture you cannot re-read as numbers is half a result
SHORT = {"Project Proposal": "proposal", "Data Collection & Preparation": "data",
         "Model Development": "model", "Deployment & Testing": "deploy",
         "Documentation & Presentation": "report+demo"}
step_table = pd.DataFrame(
    {f"slip {s} wk": {SHORT[m]: d for m, _, d in replan(s)} for s in SLIPS}
).T
step_table["TOTAL"] = step_table.sum(axis=1)

print("=" * 72)
print("STEP TABLE - milestone durations in weeks, under each slip")
print("=" * 72)
print(step_table.to_string())

# The conclusion is computed, not asserted: find the first slip that erases the last milestone
first_zero = next((s for s in SLIPS if replan(s)[-1][2] == 0), None)
print(f"\nA data phase just {first_zero} weeks late leaves the final report and demo ZERO weeks.")
print("That milestone delivers the final report, the presentation and the video demo:")
print("gate 5 (15 points) plus the gate 6 defense (15 points) = 30 of the 100 marks.")
```

**Read the storyboard.** Left to right, the only thing that changes between panels is how late the data phase runs. Follow the bottom row - the red one, `Documentation & Presentation`.

- **Panel 1 (on plan):** the report and demo have their 2 weeks, and the last bar stops right at the dashed deadline.
- **Panel 2 (1 week late):** the red bar is half as long. One week. Tight, survivable.
- **Panel 3 (2 weeks late):** **the red bar is gone**, replaced by a red `x` sitting on the deadline. Two weeks of data trouble - a licence to chase, labels that turn out to be missing - and the entire final-report milestone has no weeks left in the calendar.
- **Panel 4 (3 weeks late):** the squeeze has nowhere left to go at the end, so it moves backwards into `Deployment & Testing`, whose green bar drops from 2 weeks to 1.

Nothing dramatic happens in any frame. That is the finding. The deliverables in that vanishing bottom bar are the final report, the presentation and the video demo - **gate 5 (15 points) and the gate 6 defense (15 points), 30 of the 100 marks** - and they are lost to arithmetic, not to a disaster.

This is the sentence from the opening section drawn to scale: *"the report and the demo, the last two deliverables, get compressed into whatever days are left."*

**What to do with it, in your own plan:** the storyboard only bites because every milestone in this template is the minimum it could be, so there is nothing to give. Mark one milestone in your own timeline as **deliberately compressible** - review six papers instead of twenty, train one model instead of three - and write down in the proposal which one it is. A plan with a named shock absorber survives a slip. A plan without one deletes whatever is last.

## ⚠️ Where this breaks

A proposal is a commitment made at the moment you know least about the project. The templates above are worth writing anyway — but here is where they stop helping.

- **A proposal cannot be tested, so it cannot be wrong yet.** Every number in `success_criteria` was typed before anyone looked at the data. This is exactly why Unit 2 makes you compute a **baseline on the real file**: in that lesson a one-line domain rule ("predict survived if female") beats *both* trained models on F1, which would have exposed an over-confident proposal target in an afternoon. The assumption that must hold for your proposal numbers to mean anything: **you have seen enough of the data to know its ceiling.** Until then, treat every target as provisional and say so in writing. Do not defend a proposal number after the data has contradicted it — amend it, and record why.
- **The 14-week timeline assumes the dataset works.** There is no milestone for *"the data turned out to be unusable"*, which is the most common reason a capstone gets re-planned. The plan quietly assumes you can obtain, licence and load your data during weeks 3–5. Test that assumption in **week 1**: downloading one file and printing `df.shape` costs an hour and can save the project. If the file will not load, the licence forbids your use, or the labels are not in it, you have eleven weeks to change course instead of two.
- **15–25 papers is a lot of reading for one semester, and a bibliography is not a review.** The cheaper alternative when time is short: a **narrow** review — three papers that define the task, two that give you a baseline number you must beat, one dataset paper — read properly. Depth on six beats a skim of twenty-five, and only the depth survives a defense question.
- **When NOT to write a long proposal at all.** If your project is a reuse-and-extend of something you already built in AIAT 111–125 (your Prior Work Inventory names it), a one-page proposal plus a working `load → baseline → metric` script is stronger evidence than ten pages of plan. Planning is what you do instead of knowing; running code is knowing.
- **A plan that is never amended is not a good plan, it is an unread one.** Every departure from this proposal is legitimate if it is written down with a reason. A silent departure is scored as scope drift at gate 3.

## Summary

### Key Components of Project Proposal:
1. **Project Title**: Clear, descriptive title
2. **Problem Statement**: What problem are you solving?
3. **Objectives**: Specific, measurable goals
4. **Research Questions**: Key questions your project answers
5. **Scope**: What's included and excluded
6. **Literature Review**: Relevant research and state-of-the-art
7. **Methodology**: Technical approach and methods
8. **Timeline**: Realistic schedule with milestones
9. **Success Criteria**: How you'll measure success
10. **Risk Assessment**: Potential challenges and mitigation

### Evaluation Metrics:
- **Classification**: Accuracy, Precision, Recall, F1-score, ROC-AUC
- **Regression**: MSE, RMSE, MAE, R²
- **Deployment**: Response time, uptime, user satisfaction

## 🔭 State of the field (2026)

*Context, not examinable content — the official unit examines the proposal, the literature review and the timeline exactly as taught above. This box says where that planning work sits on the 2026 map.*

The worked scope in Part 1 defines *done* as `accuracy > 90%`, and for one trained classifier scored on a fixed test set that is still the right shape of answer. What has changed is what happens when the deliverable is not a single classifier but a **system that calls tools and retries** — the shape a growing share of capstone proposals now take. For those, a threshold met once is not yet a claim about the system: Rabanser *et al.* (2026) argue that reliability, not peak capability, is the binding constraint, decompose it into twelve metrics across **consistency, robustness, predictability and safety**, and report across 15 models that recent capability gains did not bring reliability gains with them. Reporting hardened alongside it — Evaluation Cards (Ghosh *et al.*, 2026) give evaluation results a unified schema and interpretive signals, the way model cards standardised model reporting, so a number arrives together with the conditions that produced it.

The consequence for your proposal is **one extra clause per criterion, not a different document**: beside each target in `success_criteria`, write *how many runs, on which fixed data, and what spread you will accept* — and if your project really is one model on one split, write that, because it is a legitimate answer and it is now written down. Nothing else in this lesson moves. Scope boundaries you can point at, objectives that are measurable, a calendar with named deliverables, and a literature review that ends in a gap are the mechanism the rest of the field is built on, and they are exactly what the six AIAT 126 gates score.
