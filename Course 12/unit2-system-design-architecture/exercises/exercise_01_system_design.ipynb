## System Design Primer

The Unit 2 lesson notebook is still being authored, so here is the minimum
you need to complete this exercise from this notebook alone.

A capstone system design document answers **five questions**:

1. **Components** — what are the parts of your system (data ingestion,
   preprocessing, model, serving/UI, ...), and what does each one do?
2. **Data flow** — in what order does data move between the components?
3. **Model selection** — which model(s) will you use, *why* do they fit your
   data and constraints, and what alternatives did you consider?
4. **Data formats and storage** — what format does data arrive in, what
   format do results leave in, and where is anything stored?
5. **Risks and dependencies** — what could block you (licenses, APIs,
   compute, imbalanced data), and what do you depend on?

The next cell shows a compact worked example for a sample project. Study its
shape, then produce the same five answers for **your** project in Task 1.

```python
# Worked mini-example: a complete (small) system design for a sample project —
# a web app that classifies customer reviews as positive or negative.
# Your own design in Task 1 follows the same five-part shape.

example_design = {
    "project_name": "Customer Review Sentiment Web App",
    # Part 1: components — every box you would draw in the architecture diagram
    "components": [
        "data ingestion — load the labeled reviews CSV (public Kaggle dataset)",
        "preprocessing — clean text, TF-IDF vectorization",
        "model — logistic regression classifier (scikit-learn)",
        "API — Flask endpoint /predict returning JSON",
        "UI — simple web page that calls the API and shows the label",
    ],
    # Part 2: data flow — the arrows between the boxes, in order
    "data_flow": "CSV -> preprocessing -> trained model -> Flask API -> web UI",
    # Part 3: model selection WITH justification (never just name a model)
    "model_selection": {
        "chosen_model": "Logistic regression on TF-IDF features",
        "justification": "50k short texts; a linear model is strong, fast and interpretable at this scale",
        "alternatives_considered": [
            "Naive Bayes (simpler but weaker probability estimates)",
            "fine-tuned transformer (overkill for the scope and compute budget)",
        ],
    },
    # Part 4: data formats in/out and storage choices
    "data_formats_storage": "input CSV (text,label); model saved as .joblib; API JSON in/out",
    # Part 5: risks and dependencies you plan around from day one
    "risks_dependencies": [
        "dataset license must allow coursework use",
        "class imbalance (75% positive) -> stratified split + F1 reporting",
        "Flask + scikit-learn versions pinned in requirements.txt",
    ],
}

# Print the example design document, section by section
print("=" * 60)
print("Example System Design Document")
print("=" * 60)
for key, value in example_design.items():
    print(f"\n{key}:")
    if isinstance(value, list):
        for item in value:
            print(f"  - {item}")
    elif isinstance(value, dict):
        for k, v in value.items():
            print(f"  {k}: {v}")
    else:
        print(f"  {value}")
```
