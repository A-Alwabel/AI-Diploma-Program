## The demo that cost $100 billion in a day

On 6 February 2023 Google announced **Bard**, its answer to ChatGPT, and published a short promotional demonstration. In it, Bard is asked what new discoveries from the **James Webb Space Telescope** a nine-year-old might enjoy. One of its answers was that JWST *"took the very first pictures of a planet outside of our own solar system."*

It did not. The first images of an exoplanet were captured by the European Southern Observatory's Very Large Telescope in 2004. Astronomers spotted the error within hours of the demo appearing — before Google's launch event in Paris on 8 February. Alphabet's shares slid as much as **9%** during trading that day, wiping about **$100 billion** off the company's market value (NPR, 9 February 2023).

The model was not the failure. The **demo** was. Three sentences went out under a company's name with nobody in the loop who checked the third one.

### What goes wrong without this

The report, the slides and the demo are the only parts of your project most people will ever see. Your data pipeline, your grid search, your careful three-way split — an examiner meets all of them through fifteen minutes of talking and twenty pages of writing. In AIAT 126 that presentation layer is graded at **30 of the 100 points** (gate 5, the report; gate 6, the defence), the same weight as the entire implementation.

The specific failure this unit prevents happens in the room. You run over time, the demo will not load, and you arrive at *Limitations* with ninety seconds left — so the examiner never hears the honest analysis that was the best work you did all semester. The plans below exist so that the **last three minutes of your defence are the ones you chose**.

## Part 2: Presentation Structure

Effective presentation slide deck template.


```python
# Presentation slide structure: a 15-20 minute defense deck, slide by slide,
# mirroring TEMPLATES/presentation_template.md so your slides and this plan match.
presentation_structure = {
    "1. Title Slide": {
        "content": ["Project title", "Student name(s)", "Course and date"],
        "minutes": 1
    },
    "2. Problem Statement": {
        "content": ["What problem you solve", "Why it matters", "Current solutions and their limits"],
        "minutes": 2
    },
    "3. Objectives": {
        "content": ["2-4 measurable objectives", "Success criteria"],
        "minutes": 1
    },
    "4. Methodology": {
        "content": ["Approach overview", "Technologies used", "Algorithms/models chosen and why"],
        "minutes": 3
    },
    "5. System Architecture": {
        "content": ["Architecture diagram", "Data flow between modules"],
        "minutes": 2
    },
    "6. Implementation": {
        "content": ["Key components built", "Challenges faced and how you solved them"],
        "minutes": 2
    },
    "7. Results": {
        "content": ["Performance metrics vs baseline", "Confusion matrix / ROC visuals", "Key achievements"],
        "minutes": 3
    },
    "8. Demo": {
        "content": ["Live demo or recorded video", "Have a backup recording ready"],
        "minutes": 3
    },
    "9. Future Work": {
        "content": ["Limitations", "Potential improvements and extensions"],
        "minutes": 1
    },
    "10. Q&A": {
        "content": ["Thank the audience", "Invite questions"],
        "minutes": 2
    }
}

# Print the deck plan slide by slide, with each slide's time budget
print("=" * 60)
print("Presentation Slide Deck Structure")
print("=" * 60)
for slide, details in presentation_structure.items():
    print(f"\n{slide} ({details['minutes']} min):")
    for item in details['content']:
        print(f"  - {item}")

# Compute the total talk time from the per-slide budget (do not hardcode it)
total_minutes = sum(d['minutes'] for d in presentation_structure.values())
print(f"\n✅ Total planned time: {total_minutes} minutes (target: 15-20 minutes)")
print("📝 Slide-by-slide template: see TEMPLATES/presentation_template.md")
print("📝 Practice with a timer — running long is the most common defense mistake")
```

```python
# Comparative check: do the two plans above actually FIT their requirements?
# WHY compute this rather than trust it: "20-40 pages" and "15-20 minutes" are the two hard
# constraints of this unit, and a plan that only fits at one end of its own range is a plan
# with no room for the section that turns out to need more. The arithmetic is the argument.
import re

REPORT_TARGET = (20, 40)     # pages,   PROJECT_GUIDELINES.md deliverable 5 (gate 5)
TALK_TARGET = (15, 20)       # minutes, PROJECT_GUIDELINES.md deliverable 6 (gate 6)


def page_range(text):
    """'5-8 pages' -> (5, 8); '1-2 pages' -> (1, 2); 'Variable' -> (0, 0)."""
    numbers = [int(n) for n in re.findall(r"\d+", text)]
    return (numbers[0], numbers[-1]) if numbers else (0, 0)


# Read the budget straight out of report_structure, so this check cannot drift from the plan
budget = [(name, *page_range(details["length"])) for name, details in report_structure.items()]
min_pages = sum(lo for _, lo, _ in budget)
max_pages = sum(hi for _, _, hi in budget)

print("=" * 72)
print("PLAN vs REQUIREMENT - the report")
print("=" * 72)
for name, lo, hi in budget:
    share = f"{hi / max_pages * 100:5.1f}%" if max_pages else "     -"
    print(f"  {name:<24} {lo:>2}-{hi:<2} pages  {share} of the plan's maximum")
print(f"\n  Nine sections            {min_pages}-{max_pages} pages "
      f"(appendices count as 0 - their length is 'Variable')")
print(f"  Requirement              {REPORT_TARGET[0]}-{REPORT_TARGET[1]} pages")
print(f"  Every section at its MINIMUM -> {min_pages} pages "
      f"({'exactly the floor, no slack' if min_pages <= REPORT_TARGET[0] else 'above the floor'})")
print(f"  Every section at its MAXIMUM -> {max_pages} pages "
      f"({REPORT_TARGET[1] - max_pages} pages below the ceiling)")

# Same question for the talk: planned minutes against the minutes you are actually given
talk_total = sum(d["minutes"] for d in presentation_structure.values())
demo_minutes = presentation_structure["8. Demo"]["minutes"]

print("\n" + "=" * 72)
print("PLAN vs REQUIREMENT - the talk")
print("=" * 72)
print(f"  Planned total    {talk_total} minutes")
print(f"  Requirement      {TALK_TARGET[0]}-{TALK_TARGET[1]} minutes")
print(f"  Slack            {TALK_TARGET[1] - talk_total} minutes before you are over time")
print(f"  Demo             {demo_minutes} of {talk_total} minutes "
      f"({demo_minutes / talk_total * 100:.0f}% of the talk) - the only part where the")
print(f"                   examiner sees the SYSTEM instead of your description of it")
```

```python
# FIGURE: the two plans above, drawn to scale against the two limits they must fit inside.
# WHAT this draws: the report's nine sections stacked as pages (once at every section's
# MINIMUM, once at every section's MAXIMUM) over the 20-40 page band, and the ten defence
# slides stacked as minutes over the 15-20 minute window.
# WHY draw what the table already printed: "20 pages" and "slack 0 minutes" are numbers you
# read past. Drawn to scale you SEE the bar's edge touching the limit line, and the edge
# touching the line is the entire argument of this section.
import numpy as np
import matplotlib.pyplot as plt

fig, (ax_r, ax_t) = plt.subplots(2, 1, figsize=(11.0, 6.6))

# --- panel 1: the report. Reuse `budget` from the cell above so the drawing cannot
# drift from the arithmetic. Appendices are "Variable" -> (0, 0), so they draw nothing.
sections = [(n, lo, hi) for n, lo, hi in budget if hi > 0]
colors = plt.cm.Blues(np.linspace(0.30, 0.88, len(sections)))
ax_r.axvspan(*REPORT_TARGET, color="#2e7d32", alpha=0.09, zorder=0)  # the allowed band
for row, key in enumerate(["hi", "lo"]):          # row 0 = all maxima, row 1 = all minima
    left = 0.0
    for (name, lo, hi), c in zip(sections, colors):
        w = hi if key == "hi" else lo
        ax_r.barh(row, w, left=left, height=0.5, color=c, edgecolor="white", lw=1.0, zorder=3)
        if w >= 2:                                 # only label segments wide enough to read
            ax_r.text(left + w / 2, row, name.split(".")[0], ha="center", va="center",
                      fontsize=8, color="white", fontweight="bold", zorder=4)
        left += w
    ax_r.text(left + 0.6, row, f"{left:.0f} pp", va="center", fontsize=9.5, fontweight="bold",
              color="#c62828" if left <= REPORT_TARGET[0] else "#555")
for x, lab in [(REPORT_TARGET[0], "floor 20"), (REPORT_TARGET[1], "ceiling 40")]:
    ax_r.axvline(x, color="#2e7d32", ls="--", lw=1.4, zorder=2)
    ax_r.text(x, -0.62, lab, ha="center", fontsize=8, color="#2e7d32")
ax_r.set_yticks([0, 1])
ax_r.set_yticklabels(["every section\nat its MAX", "every section\nat its MIN"], fontsize=9)
ax_r.set_xlim(0, 45); ax_r.set_ylim(-0.85, 1.75); ax_r.set_xticks(range(0, 46, 5))
ax_r.set_xlabel("pages", fontsize=9)
ax_r.set_title("The report plan drawn to scale — section page budgets vs the 20–40 requirement",
               fontsize=11, fontweight="bold", pad=10)
ax_r.annotate("the MIN bar stops exactly ON the floor — no slack",
              xy=(min_pages + 0.4, 1.28), xytext=(24.0, 1.48), fontsize=9, color="#c62828",
              arrowprops=dict(arrowstyle="->", color="#c62828", lw=1.2))
ax_r.annotate(f"{REPORT_TARGET[1] - max_pages} pp still spare", xy=(max_pages + 3.0, 0.42),
              xytext=(34.5, 0.95), fontsize=9, color="#555",
              arrowprops=dict(arrowstyle="->", color="#888", lw=1.0))
ax_r.legend([plt.Rectangle((0, 0), 1, 1, color=c) for c in colors],
            [n for n, _, _ in sections], fontsize=7.5, ncol=4, loc="upper center",
            bbox_to_anchor=(0.5, -0.30), frameon=False, handlelength=1.2, columnspacing=1.0)

# --- panel 2: the talk. Same idea, minutes instead of pages, and the 20-minute limit
# drawn as a solid red wall because it is enforced by a person with a clock.
slides = list(presentation_structure.items())
tcolors = plt.cm.Oranges(np.linspace(0.28, 0.86, len(slides)))
ax_t.axvspan(*TALK_TARGET, color="#2e7d32", alpha=0.09, zorder=0)
left = 0.0
for (name, d), c in zip(slides, tcolors):
    m = d["minutes"]
    ax_t.barh(0, m, left=left, height=0.5, color=c, edgecolor="white", lw=1.0,
              hatch="//" if name.startswith("8.") else None, zorder=3)  # hatch = the demo
    ax_t.text(left + m / 2, 0, name.split(".")[0], ha="center", va="center",
              fontsize=8, color="white", fontweight="bold", zorder=4)
    left += m
ax_t.axvline(TALK_TARGET[0], color="#2e7d32", ls="--", lw=1.4, zorder=2)
ax_t.axvline(TALK_TARGET[1], color="#c62828", ls="-", lw=2.2, zorder=5)
ax_t.text(TALK_TARGET[0], 0.36, "15 min", ha="center", fontsize=8, color="#2e7d32")
ax_t.text(TALK_TARGET[1], 0.36, "20 min — hard stop", ha="center", fontsize=8.5,
          color="#c62828", fontweight="bold")
ax_t.annotate(f"the plan ends ON the wall: slack = {TALK_TARGET[1] - talk_total} min",
              xy=(talk_total - 0.2, -0.30), xytext=(9.0, -0.62), fontsize=9, color="#c62828",
              arrowprops=dict(arrowstyle="->", color="#c62828", lw=1.2))
ax_t.set_yticks([]); ax_t.set_xlim(0, 22.5); ax_t.set_ylim(-0.85, 0.62)
ax_t.set_xticks(range(0, 23, 2)); ax_t.set_xlabel("minutes", fontsize=9)
ax_t.set_title("The defence deck drawn to scale — ten slides vs the 15–20 minute window",
               fontsize=11, fontweight="bold", pad=10)
ax_t.legend([plt.Rectangle((0, 0), 1, 1, color=c) for c in tcolors],
            [n for n, _ in slides], fontsize=7.5, ncol=5, loc="upper center",
            bbox_to_anchor=(0.5, -0.34), frameon=False, handlelength=1.2, columnspacing=1.0)
fig.subplots_adjust(hspace=1.15, top=0.93, bottom=0.11)
plt.show()
```

**What to look at in the figure above** (both panels are the same numbers the cell above
printed, drawn against the limits they must fit inside):

  the requirement's floor is at 20 pages: they are the same line. Written at every section's

### Reading the comparison: two plans, and neither has any slack

Both plans "fit" — and both fit in the way that means trouble.

- **The talk lands on exactly 20 minutes, the top of the 15–20 minute window. Slack: zero.** Every defence overruns somewhere: a demo that takes forty seconds to load, an examiner who interrupts at slide 5, a projector that will not accept your laptop. A plan with no slack does not absorb that — it drops whatever is last on the list. Last here are *Future Work* (where your limitations live) and *Q&A*. **Cut two minutes now, deliberately, from a slide you choose, or the room will choose for you.**
- **The report's nine sections total 20–32 pages against a 20–40 requirement.** Writing every section at its stated minimum puts you on exactly 20 pages: the floor, with nothing spare. That is fine if it is a decision and dangerous if it is an accident — because the section most likely to overrun is *Results and Analysis*, already the largest at up to 8 pages, a quarter of the whole plan, and the one carrying every number you are graded on.
- **The demo is 3 of 20 minutes.** Fifteen percent of the talk, inside a gate worth 15 of the 100 course points, and the only stretch where the examiner sees the system rather than your account of it. Rehearse it more than proportionally.

```python
# FRAME-GRID STORYBOARD: four snapshots of the same deck, at four sizes of overrun.
# THE MODEL (stated, because the figure is only as honest as its assumption): slides run in
# order, and the room stops you at 20 minutes. So an unplanned overrun of D minutes anywhere
# early in the talk shifts everything right by D and deletes the LAST D minutes of the plan.
# WHY this is worth drawing rather than asserting: "a plan with no slack drops whatever is
# last" names no victim. The storyboard names it, and the victims are the two slides you
# least want to lose.
HARD_STOP = TALK_TARGET[1]          # 20 minutes - the wall
DELAYS = [0, 2, 4, 6]               # four snapshots, cheap: no animation needed to see this

fig, ax = plt.subplots(figsize=(11.0, 5.0))
step_rows = []                       # collected for the printed step table below the figure

# Draw one timeline per delay, newest state at the bottom, all sharing the one red wall.
for row, delay in zip(range(len(DELAYS) - 1, -1, -1), DELAYS):
    if delay:                        # the overrun itself: grey, hatched, at the front
        ax.barh(row, delay, left=0, height=0.38, color="#bdbdbd", edgecolor="#616161",
                hatch="xx", lw=0.8, zorder=3)
        ax.text(delay / 2, row, f"+{delay}", ha="center", va="center", fontsize=7.5,
                color="#212121", fontweight="bold", zorder=4)
    t = float(delay)
    lost_names, lost_total, full_count = [], 0.0, 0
    for (name, d), c in zip(slides, tcolors):
        m = float(d["minutes"])
        # delivered = the part of this slide that falls before the wall
        delivered = max(0.0, min(t + m, HARD_STOP) - min(t, HARD_STOP))
        lost = m - delivered
        if delivered > 0:
            ax.barh(row, delivered, left=t, height=0.38, color=c, edgecolor="white",
                    lw=0.8, zorder=3)
            if delivered >= 1.4:
                ax.text(t + delivered / 2, row, name.split(".")[0], ha="center", va="center",
                        fontsize=7.5, color="white", fontweight="bold", zorder=4)
        if lost > 0:                 # the part that never happens: white, red hatch
            ax.barh(row, lost, left=t + delivered, height=0.38, color="#ffffff",
                    edgecolor="#c62828", hatch="//", lw=0.8, alpha=0.95, zorder=3)
            lost_names.append(name.split(". ", 1)[1]
                              + ("" if lost == m else f" (−{lost:g} of {m:g} min)"))
        lost_total += lost
        full_count += int(delivered == m)
        t += m
    ax.text(-0.4, row, f"delay\n+{delay} min", ha="right", va="center", fontsize=8.5)
    msg = "nothing lost — every slide delivered" if not lost_names else "lost: " + ", ".join(lost_names)
    ax.text(0.2, row - 0.33, msg, fontsize=8.2, va="top",
            color="#555" if not lost_names else "#c62828")
    step_rows.append((delay, lost_total, full_count,
                      lost_names[0].split(" (")[0] if lost_names else "—"))

ax.axvline(HARD_STOP, color="#c62828", lw=2.4, zorder=6)
ax.axvspan(HARD_STOP, 27, color="#c62828", alpha=0.05, zorder=0)
ax.text(HARD_STOP + 0.25, len(DELAYS) - 0.45, "the room stops you here (20 min)",
        fontsize=9, color="#c62828", fontweight="bold", va="bottom")
ax.set_yticks([]); ax.set_xlim(-3.2, 27); ax.set_ylim(-0.75, len(DELAYS) - 0.15)
ax.set_xticks(range(0, 27, 2)); ax.set_xlabel("minutes into the defence", fontsize=9)
ax.set_title("Storyboard: a zero-slack plan meets a delay — what actually gets deleted",
             fontsize=11.5, fontweight="bold", pad=12)
ax.spines[["top", "right", "left"]].set_visible(False)
ax.legend([plt.Rectangle((0, 0), 1, 1, fc="#bdbdbd", ec="#616161", hatch="xx"),
           plt.Rectangle((0, 0), 1, 1, fc=tcolors[5], ec="white"),
           plt.Rectangle((0, 0), 1, 1, fc="white", ec="#c62828", hatch="//")],
          ["unplanned overrun", "slide delivered", "slide never reached"],
          fontsize=8, ncol=3, loc="upper center", bbox_to_anchor=(0.5, -0.13), frameon=False)
fig.text(0.5, -0.05, "model: slides run in order and the clock stops at 20 min, so a D-minute "
                     "overrun deletes the last D minutes of the plan",
         ha="center", fontsize=7.5, color="#777")
plt.tight_layout()
plt.show()

# The same four frames as numbers, so nothing in the picture has to be eyeballed
print("=" * 72)
print("STEP TABLE - the storyboard above, frame by frame")
print("=" * 72)
print(f"{'delay':>7} | {'minutes deleted':>15} | {'slides fully delivered':>22} | first casualty")
for delay, lost_total, full_count, first in step_rows:
    print(f"{'+' + str(delay) + ' min':>7} | {lost_total:>15g} | {full_count:>22} | {first}")
print(f"\n  Every minute you lose comes off the END of the deck, never the middle.")
print(f"  A {DELAYS[-1]}-minute overrun costs you the demo, the limitations and the questions.")
```

**Read the storyboard top to bottom.** Each row is the same deck; the only thing that
changes is how many minutes an unplanned overrun eats before you start.

- **Row 1 (+0 min): the plan works exactly once — when nothing goes wrong.** Every slide is
  solid orange and the bar stops on the wall.
- **Row 2 (+2 min): a two-minute overrun costs the whole of Q&A.** Not a shortened Q&A — the
  red-hatched block past the wall is the part that never happens. Two minutes is one slow demo
  load or one examiner interrupting at slide 5.
- **Row 3 (+4 min): the cut reaches back into the Demo.** You lose Q&A, *Future Work* — which
  is where your limitations live — and a third of the only segment where the examiner sees the
  system instead of your account of it.
- **Row 4 (+6 min): the demo, the limitations and the questions are all gone.** Seven slides
  delivered out of ten; the three you lose are the last three, and the step table confirms it.

The point is not that six minutes is likely. It is that the deletions are never chosen — they
are whatever sits at the end of the list. That is the argument for cutting two minutes now,
deliberately, from a slide you pick, which is exactly what discussion question 1 below asks you
to do.

## ⚠️ Where this breaks

Documentation is the deliverable students plan for least and lose the most marks on. Here is where the templates above stop being enough.

- **A checklist proves a file exists, not that it works.** "☐ requirements.txt" is ticked by an empty file. The only version of that check worth anything is the one you run: clone your own submission into a fresh folder, follow your own README on a machine that is not yours, and time it. If it does not run in ten minutes it does not run. Every cohort produces at least one submission whose notebook only executes because a data file happens to sit in the author's Downloads folder.
- **A page count is not a report and a slide count is not a talk.** The nine sections and ten slides above are a floor, not a definition of quality. An examiner reading your Results chapter wants one thing: a number, the baseline it beat, and the honest gap between them. A 40-page report that never says what the system was compared against scores below a 22-page one that does.
- **The plan treats the demo as the last risk; it is usually the first.** The failure modes are boringly consistent — no network in the room, a model file too large to carry, a package version that differs on the borrowed laptop, a dataset you may not legally display on a projector because it contains personal data. Record the demo **before** the week of the defence, while there is still time to fix whatever the recording exposes.
- **"Address limitations honestly" is cheap to write and expensive to do.** The honest version names a number: *"validation F1 0.8125, test F1 0.7302 — our validation estimate was optimistic by 8.2 points on 179 test rows."* The version that is not honest says "future work includes further evaluation". Examiners can tell instantly, and CLO5 is graded on which one you wrote.
- **Nothing in this notebook produces the ethical considerations document.** That is a separate required submission on the Unit 5 checklist, graded at gates 5 and 6 and examined in Part 4 of the final evaluation. It is not a paragraph inside the Discussion chapter, and none of the templates above will generate it for you.
- **When NOT to use this structure at all.** These are the conventions of an academic project report, and they are graded because your reader is an examiner. If your reader were a company or a government body, the executive summary would become the whole document, the literature review would disappear, and the limitations would move to the **front** — a decision-maker needs to know what the system cannot do before reading what it can. Ask who reads it, then choose the structure.

## Summary

### Key Components:
1. **Final Report**: Comprehensive 20-40 page document covering all project aspects
2. **Presentation**: 15-20 minute slide deck with clear structure and visualizations
3. **Demonstration**: Live demo or video showing your system in action
4. **Submission Package**: Well-organized collection of all project materials

### Best Practices:
- **Writing**: Clear, concise, well-structured, properly cited
- **Visualizations**: Use graphs, tables, confusion matrices, ROC curves
- **Presentation**: Practice timing, prepare for questions, engage audience
- **Code Organization**: Clean, commented, documented, reproducible
- **Documentation**: README files, user guides, setup instructions

### Presentation Tips:
- Start with problem and motivation
- Show methodology clearly
- Highlight key results with visualizations
- Include live demo when possible
- Address limitations honestly
- End with clear contributions
