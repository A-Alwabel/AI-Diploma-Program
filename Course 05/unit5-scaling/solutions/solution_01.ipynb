```python
import pandas as pd
import numpy as np
import time

np.random.seed(42)
large_data = pd.DataFrame({
    'id': range(1, 100001), 'value1': np.random.randn(100000), 'value2': np.random.randn(100000),
    'category': np.random.choice(['A','B','C','D'], 100000), 'score': np.random.uniform(0, 100, 100000)
})
```

### Task 1: Efficient processing (chunking)

```python
print("=" * 60)
print("Task 1: Efficient Processing")
print("=" * 60)
t0 = time.perf_counter()
chunk_sums = []
for chunk in np.array_split(large_data, 10):
    chunk_sums.append(chunk['score'].sum())
total = sum(chunk_sums)
print(f"Chunked sum (10 chunks): {total:.2f}, time: {time.perf_counter()-t0:.4f}s")
```

### Task 2: Aggregation

```python
print("\n" + "=" * 60)
print("Task 2: Aggregation")
print("=" * 60)
t0 = time.perf_counter()
agg = large_data.groupby('category').agg({'score': ['mean','std','count'], 'value1': 'sum'})
print(agg)
print(f"Time: {time.perf_counter()-t0:.4f}s")
```

### Task 3: Memory optimization

```python
print("\n" + "=" * 60)
print("Task 3: Memory Optimization")
print("=" * 60)
before = large_data.memory_usage(deep=True).sum() / 1024**2
large_data['category'] = large_data['category'].astype('category')
large_data['id'] = large_data['id'].astype('int32')
after = large_data.memory_usage(deep=True).sum() / 1024**2
print(f"Before: {before:.2f} MB, After: {after:.2f} MB, Saved: {before-after:.2f} MB")
```

### Task 4: Performance profiling

```python
print("\n" + "=" * 60)
print("Task 4: Performance Profiling")
print("=" * 60)
def time_op(name, f):
    t = time.perf_counter(); f(); elapsed = time.perf_counter() - t; print(f"  {name}: {elapsed:.4f}s")
time_op('groupby mean', lambda: large_data.groupby('category')['score'].mean())
time_op('sort', lambda: large_data.sort_values('score'))
time_op('query', lambda: large_data[large_data['score'] > 50])
print("\n" + "=" * 60)
print("Exercise 1 Complete!")
print("=" * 60)
```
