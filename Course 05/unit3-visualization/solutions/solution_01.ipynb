```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams['axes.unicode_minus'] = False
sns.set_style("whitegrid")

np.random.seed(42)
data = {'month': ['Jan','Feb','Mar','Apr','May','Jun'], 'sales': [1200,1500,1800,1600,2000,2200],
        'region': np.random.choice(['North','South','East','West'],6), 'product': np.random.choice(['A','B','C'],6),
        'rating': np.random.uniform(3,5,6)}
df = pd.DataFrame(data)
advanced_data = pd.DataFrame({'x': np.random.randn(100), 'y': np.random.randn(100),
    'category': np.random.choice(['Type1','Type2','Type3'],100), 'value': np.random.uniform(10,100,100)})
```

### Task 1: Basic plots

```python
print("=" * 60)
print("Task 1: Basic Plots")
print("=" * 60)
fig, ax = plt.subplots(1, 3, figsize=(12, 4))
ax[0].plot(df['month'], df['sales'], marker='o'); ax[0].set_title('Sales over months'); ax[0].tick_params(axis='x', rotation=45)
df.groupby('region')['sales'].sum().plot(kind='bar', ax=ax[1]); ax[1].set_title('Sales by region'); ax[1].tick_params(axis='x', rotation=45)
ax[2].scatter(advanced_data['x'], advanced_data['y'], alpha=0.6); ax[2].set_title('x vs y')
plt.tight_layout(); plt.show()
```

### Task 2: Statistical plots

```python
print("\n" + "=" * 60)
print("Task 2: Statistical Plots")
print("=" * 60)
fig, ax = plt.subplots(1, 3, figsize=(12, 4))
sns.histplot(advanced_data['value'], kde=True, ax=ax[0]); ax[0].set_title('Distribution')
sns.boxplot(data=advanced_data, x='category', y='value', ax=ax[1]); ax[1].set_title('Box by category')
sns.heatmap(advanced_data[['x','y','value']].corr(), annot=True, ax=ax[2], cmap='coolwarm'); ax[2].set_title('Correlation')
plt.tight_layout(); plt.show()
```

### Task 3: Customization

```python
print("\n" + "=" * 60)
print("Task 3: Plot Customization")
print("=" * 60)
ax = df.groupby('region')['sales'].sum().plot(kind='bar', color=['#2ecc71','#3498db','#e74c3c','#9b59b6'], edgecolor='black')
ax.set_xlabel('Region'); ax.set_ylabel('Sales'); ax.set_title('Sales by Region (customized)')
ax.legend(['Sales']); ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('unit3_ex1_task3.png', dpi=100, bbox_inches='tight')
plt.show()
print("Saved unit3_ex1_task3.png (in cwd)")
```

### Task 4: Multiple subplots

```python
print("\n" + "=" * 60)
print("Task 4: Multiple Subplots")
print("=" * 60)
fig, axes = plt.subplots(2, 2, figsize=(10, 8))
axes[0,0].plot(df['month'], df['sales']); axes[0,0].set_title('Line')
df['sales'].hist(ax=axes[0,1], bins=6); axes[0,1].set_title('Histogram')
axes[1,0].scatter(advanced_data['x'], advanced_data['y'], c=advanced_data['value'], cmap='viridis')
axes[1,0].set_title('Scatter')
df.groupby('product')['sales'].sum().plot(kind='bar', ax=axes[1,1]); axes[1,1].set_title('Bar')
fig.suptitle('2x2 Subplots'); plt.tight_layout(); plt.show()
print("\n" + "=" * 60)
print("Exercise 1 Complete!")
print("=" * 60)
```
