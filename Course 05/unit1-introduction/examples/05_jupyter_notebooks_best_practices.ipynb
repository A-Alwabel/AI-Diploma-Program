## Why Jupyter Notebook Best Practices Matter

Best practices are essential because:
- **Reproducibility**: Others (and future you) can run your code
- **Organization**: Clear structure makes notebooks easy to follow
- **Professionalism**: Well-organized notebooks impress employers/clients
- **Efficiency**: Good practices save time and prevent errors

**Common Student Questions:**
- **Q: Why use markdown cells?**
  - Answer: Markdown explains WHAT and WHY, code shows HOW
  - Example: Markdown explains "We're calculating average", code shows the calculation
  - Benefit: Makes notebooks readable and educational
  
- **Q: Why set random seeds?**
  - Answer: Ensures results are reproducible (same random numbers each run)
  - Example: `np.random.seed(42)` before generating random data
  - Benefit: Others get the same results when running your notebook

## 🎯 The case: a spreadsheet row range that shaped austerity policy

In 2010 Carmen Reinhart and Kenneth Rogoff published *"Growth in a Time of Debt"*,
reporting that countries whose public debt passes 90% of GDP see average real growth of
about **−0.1%**. The finding was quoted in budget debates across Europe and the United
States. In 2013 Thomas Herndon, a graduate student at UMass Amherst, asked for the
original spreadsheet and — with Michael Ash and Robert Pollin — found that an averaging
formula covered the wrong range of rows, silently omitting several countries, alongside
disputed exclusions and an unusual weighting scheme. Recomputed, the average growth above
90% debt was **+2.2%**, not −0.1%. The sign of the headline result flipped.

Nothing was faked. The analysis simply could not be re-run by anyone else until somebody
asked for the file — three years later.

**What goes wrong without this.** Nothing dramatic on the day. Months later somebody asks
you to re-derive a number, you open the notebook, and it fails on cell three — or worse, it
runs and gives a different answer. The finding you published cannot be defended, and you
cannot say why. **And this is not rare.** Pimentel, Murta, Braganholo and Freire (MSR 2019) downloaded
**1.4 million Jupyter notebooks** from GitHub and tried to execute them. About **24% ran
without an error at all**, and only about **4% reproduced the same results** their authors
had committed. Your notebook is, statistically, more likely to be in the 96% than the 4%.

**The other side of the same coin.** The notebooks that let anyone re-derive the first
observation of gravitational waves are published by the LIGO Open Science Center, and
Project Jupyter's steering committee received the **2017 ACM Software System Award** for
exactly this: making computational work something other people can re-run. The difference
between the Reinhart–Rogoff spreadsheet and the LIGO notebooks is not talent. It is the
handful of habits in this lesson.

```python
# ============================================================================
# PART 2: CODE ORGANIZATION
# ============================================================================
print("\n" + "=" * 70)
print("PART 2: Code Organization")
print("=" * 70)

# Example 1: Clear variable names
print("\n✅ Example 1: Clear Variable Names")
print("-" * 70)

# ❌ Bad: Unclear names
x = [1, 2, 3, 4, 5]
y = sum(x) / len(x)

# ✅ Good: Clear, descriptive names
student_scores = [85, 90, 78, 92, 88]
average_score = sum(student_scores) / len(student_scores)

print(f"   Student scores: {student_scores}")
print(f"   Average score: {average_score}")
print("   ✅ Clear names make code self-documenting!")

# Example 2: Comments and explanations
print("\n✅ Example 2: Comments and Explanations")
print("-" * 70)

# Calculate statistics with clear comments
data = [10, 20, 30, 40, 50]
```

```python
# Calculate mean (average)
mean_value = sum(data) / len(data)

# Calculate variance (spread of data)
variance = sum((x - mean_value) ** 2 for x in data) / len(data)

print(f"   Data: {data}")
print(f"   Mean: {mean_value}")
print(f"   Variance: {variance}")
print("   ✅ Comments explain WHY, not just WHAT")

# ============================================================================
# ============================================================================
print("\n" + "=" * 70)
print("PART 3: Reproducibility")
print("=" * 70)

import numpy as np
import random

# Example 1: Set random seeds
print("\n✅ Example 1: Setting Random Seeds")
print("-" * 70)

# Set seed for reproducibility
np.random.seed(42)  # Same seed = same random numbers
random.seed(42)

# Generate random data
random_data_1 = np.random.rand(5)
random_data_2 = np.random.rand(5)
```

```python
# WHAT: Show that re-seeding NumPy's random generator reproduces identical numbers, then list dependency-documentation tips.
# WHY: Reproducibility is the heart of a trustworthy notebook - the same inputs must give the same outputs on any machine.

print(f"   Random data (seed=42): {random_data_1}")
print("   ✅ Setting seed ensures same results every time!")
# Reset seed to show reproducibility
np.random.seed(42)
random_data_3 = np.random.rand(5)
print(f"   Same seed again: {random_data_3}")
print(f"   ✅ Results match: {np.array_equal(random_data_1, random_data_3)}")

# The two arrays above are NOT a dataset - they exist only to prove that seeding
# works. Randomness is the lesson here, so random numbers are the right material.
# Now the same idea where it actually bites in daily work: sampling REAL data.
print("\n✅ Example 1b: The Same Rule on a Real Dataset")
print("-" * 70)
import pandas as pd
# --- Data setup. Works from any folder, and on Google Colab. -------------------------
# WHAT: find the repository root and put it on sys.path, then import the shared loader.
# WHY:  a hard-coded '../../../Course 04/datasets/raw/titanic.csv' only resolves when the
#       kernel's working directory happens to be this notebook's folder. This does not care.
import sys, pathlib

_here = pathlib.Path.cwd().resolve()
_root = next((p for p in [_here, *_here.parents] if (p / "tools" / "data.py").exists()), None)
if _root is None:                     # Google Colab, or a stray copy of the notebook
    import urllib.request
    pathlib.Path("tools").mkdir(exist_ok=True)
    try:
        urllib.request.urlretrieve("https://raw.githubusercontent.com/A-Alwabel/AI-Diploma-Program/main/tools/data.py", "tools/data.py")
    except Exception as _e:
        raise RuntimeError(
            "Could not find the AI Diploma repository from this folder, and could not "
            "download the data loader either. Open this notebook inside a clone of "
            "https://github.com/A-Alwabel/AI-Diploma-Program, or connect to the internet "
            f"and re-run this cell. (underlying error: {_e})") from None
    _root = pathlib.Path.cwd()
sys.path.insert(0, str(_root))

from tools.data import load
# -------------------------------------------------------------------------------------
titanic = load("titanic")

# random_state is pandas' equivalent of a seed. Without it, every re-run of your
# notebook analyses a DIFFERENT sample - and your reported numbers drift silently.
sample_a = titanic.sample(5, random_state=42)
sample_b = titanic.sample(5, random_state=42)
sample_c = titanic.sample(5)          # no random_state: a different draw each run

print(f"   Passengers drawn with random_state=42: {list(sample_a['PassengerId'])}")
print(f"   Drawn again with random_state=42:      {list(sample_b['PassengerId'])}")
print(f"   ✅ Identical sample: {sample_a.equals(sample_b)}")
print(f"   Drawn with NO random_state:            {list(sample_c['PassengerId'])}")
print("   ⚠ That last line changes every time you run this cell. If a figure or a")
print("     reported mean depends on it, your notebook is not reproducible.")

# Example 2: Document dependencies
print("\n✅ Example 2: Document Dependencies")
print("-" * 70)
print("""
In your notebook, always include:
- import statements at the top
- requirements.txt or environment.yml

✅ This helps others install correct versions!
""")
# ============================================================================
# PART 4: NOTEBOOK BEST PRACTICES CHECKLIST
# ============================================================================
print("\n" + "=" * 70)
```

### 👁️ Read the figure

- **Three bars, each a subset of the one above it.** The blue bar is every notebook the study
  tried to execute. The orange bar is the fraction that ran to the end without raising an
  error — about **a quarter**. The red bar is the fraction that produced the *same results
  their own authors had committed* — about **one in twenty-five**.
- **Do the second division yourself:** 4 out of 24. Of the notebooks that ran at all, roughly
  **five in six still gave a different answer** than the version saved in the repository.
- **The source line sits inside the figure.** These are Pimentel et al.'s percentages, not
  this notebook's; a figure that borrows someone else's measurement has to say so where the
  figure is, not three paragraphs away.

**What it proves:** "it works on my machine" is the default state of a public notebook, not
the exception. The distance from the blue bar to the orange bar is *hidden state* — cells run
out of order, variables from cells since deleted — and **Restart Kernel & Run All** is the only
test that closes it. The distance from orange to red is *environment and data* — unpinned
versions, a CSV that changed upstream — which is what the `pandas.__version__` line above and
a recorded data source are for. Reinhart and Rogoff's spreadsheet failed at the red bar, and
it took three years and one graduate student to notice.

## ⚠️ Where this breaks

A tidy notebook is not a reproducible one. The specific ways this lesson's advice falls
short:

- **Hidden state beats good intentions.** Cells run in whatever order you clicked them,
  and a variable defined in a cell you later deleted still lives in the kernel. A notebook
  can produce perfect output on your screen and fail on the first cell for everyone else.
  The only honest test is *Restart Kernel and Run All* before you share it — the test the
  96% in the MSR 2019 study did not pass.
- **A seed does not pin the environment.** `np.random.seed(42)` fixes NumPy's stream. It
  does not fix your pandas version, your BLAS threading, GPU non-determinism, or a
  `sample()` call whose behaviour changed between releases. This notebook prints
  `pandas 2.3.3` and `numpy 2.4.4` — those two lines do more for reproducibility than the
  seed does.
- **Notebooks are hostile to version control.** The `.ipynb` file is JSON with embedded
  outputs and base64 images, so `git diff` is unreadable and two people editing the same
  notebook get merge conflicts they cannot resolve by hand.
- **The assumption that must hold: the data is the same data.** None of this saves you if
  the CSV was overwritten upstream. Record where the file came from and when, and prefer a
  checksum to a promise.
- **When a notebook is the wrong artifact:** anything scheduled, tested, or imported by
  other code belongs in a `.py` module. Use notebooks for exploration and explanation;
  move logic that has to *run reliably* into scripts with tests, and pin the environment
  (`requirements.txt` with exact versions, or a container). Course 05 Unit 5 lessons 05
  and 10 do exactly that migration.
