Best practices are essential because:

- **Organization**: Clear structure makes notebooks easy to follow
- **Professionalism**: Well-organized notebooks impress employers/clients
- **Efficiency**: Good practices save time and prevent errors

**Common Student Questions:**
- **Q: Why use markdown cells?**
  - Answer: Markdown explains WHAT and WHY, code shows HOW
  - Example: Markdown explains "We're calculating average", code shows the calculation
  - Benefit: Makes notebooks readable and educational
  
- **Q: Why set random seeds?**
  - Answer: Ensures results are reproducible (same random numbers each run)
  - Example: `np.random.seed(42)` before generating random data
  - Benefit: Others get the same results when running your notebook

```python
# ============================================================================
# PART 1: NOTEBOOK STRUCTURE
# ============================================================================
print("=" * 70)
print("Jupyter Notebooks Best Practices")
print("=" * 70)
print("\n📝 PART 1: Notebook Structure")
print("-" * 70)
# Example: Good notebook structure
print("""
✅ Good Notebook Structure:
1. Header Section (Markdown):
   - Title and learning objectives
   - Prerequisites checklist
   - Overview of what you'll learn
2. Introduction (Markdown):
   - Context and background
   - Why this topic matters
   - What you'll accomplish
3. Code Sections (Code + Markdown):
   - Markdown: Explains what you're doing
   - Code: Shows how to do it
   - Markdown: Explains the results
4. Summary (Markdown):
   - Key takeaways
   - Next steps
   - References
""")
```

```python
# ============================================================================
# PART 2: CODE ORGANIZATION
# ============================================================================
print("\n" + "=" * 70)
print("PART 2: Code Organization")
print("=" * 70)

# Example 1: Clear variable names
print("\n✅ Example 1: Clear Variable Names")
print("-" * 70)

# ❌ Bad: Unclear names
x = [1, 2, 3, 4, 5]
y = sum(x) / len(x)

# ✅ Good: Clear, descriptive names
student_scores = [85, 90, 78, 92, 88]
average_score = sum(student_scores) / len(student_scores)

print(f"   Student scores: {student_scores}")
print(f"   Average score: {average_score}")
print("   ✅ Clear names make code self-documenting!")

# Example 2: Comments and explanations
print("\n✅ Example 2: Comments and Explanations")
print("-" * 70)

# Calculate statistics with clear comments
data = [10, 20, 30, 40, 50]
```

```python
# Calculate mean (average)
mean_value = sum(data) / len(data)

# Calculate variance (spread of data)
variance = sum((x - mean_value) ** 2 for x in data) / len(data)

print(f"   Data: {data}")
print(f"   Mean: {mean_value}")
print(f"   Variance: {variance}")
print("   ✅ Comments explain WHY, not just WHAT")

# ============================================================================
# ============================================================================
print("\n" + "=" * 70)
print("PART 3: Reproducibility")
print("=" * 70)

import numpy as np
import random

# Example 1: Set random seeds
print("\n✅ Example 1: Setting Random Seeds")
print("-" * 70)

# Set seed for reproducibility
np.random.seed(42)  # Same seed = same random numbers
random.seed(42)

# Generate random data
random_data_1 = np.random.rand(5)
random_data_2 = np.random.rand(5)
```

```python
# WHAT: Show that re-seeding NumPy's random generator reproduces identical numbers, then list dependency-documentation tips.
# WHY: Reproducibility is the heart of a trustworthy notebook - the same inputs must give the same outputs on any machine.

print(f"   Random data (seed=42): {random_data_1}")
print("   ✅ Setting seed ensures same results every time!")
# Reset seed to show reproducibility
np.random.seed(42)
random_data_3 = np.random.rand(5)
print(f"   Same seed again: {random_data_3}")
print(f"   ✅ Results match: {np.array_equal(random_data_1, random_data_3)}")

# The two arrays above are NOT a dataset - they exist only to prove that seeding
# works. Randomness is the lesson here, so random numbers are the right material.
# Now the same idea where it actually bites in daily work: sampling REAL data.
print("\n✅ Example 1b: The Same Rule on a Real Dataset")
print("-" * 70)
import pandas as pd
titanic = pd.read_csv('../../../Course 04/datasets/raw/titanic.csv')

# random_state is pandas' equivalent of a seed. Without it, every re-run of your
# notebook analyses a DIFFERENT sample - and your reported numbers drift silently.
sample_a = titanic.sample(5, random_state=42)
sample_b = titanic.sample(5, random_state=42)
sample_c = titanic.sample(5)          # no random_state: a different draw each run

print(f"   Passengers drawn with random_state=42: {list(sample_a['PassengerId'])}")
print(f"   Drawn again with random_state=42:      {list(sample_b['PassengerId'])}")
print(f"   ✅ Identical sample: {sample_a.equals(sample_b)}")
print(f"   Drawn with NO random_state:            {list(sample_c['PassengerId'])}")
print("   ⚠ That last line changes every time you run this cell. If a figure or a")
print("     reported mean depends on it, your notebook is not reproducible.")

# Example 2: Document dependencies
print("\n✅ Example 2: Document Dependencies")
print("-" * 70)
print("""
In your notebook, always include:
- import statements at the top
- requirements.txt or environment.yml

✅ This helps others install correct versions!
""")
# ============================================================================
# PART 4: NOTEBOOK BEST PRACTICES CHECKLIST
# ============================================================================
print("\n" + "=" * 70)
```

```python
# WHAT: Define a checklist of notebook best practices grouped into four categories.
# WHY: Storing the checklist as a dictionary lets the next cell print it programmatically - content and presentation stay separate.

print("PART 4: Best Practices Checklist")
print("=" * 70)

best_practices = {
    "Structure": [
        "✅ Clear title and objectives",
        "✅ Logical flow (top to bottom)",
        "✅ Markdown cells explain code",
        "✅ Section headers organize content"
    ],
    "Code Quality": [
        "✅ Clear variable names",
        "✅ Comments explain complex logic",
        "✅ One concept per cell",
        "✅ Remove unnecessary code"
    ],
    "Reproducibility": [
        "✅ Set random seeds",
        "✅ Document all dependencies",
        "✅ Include data sources",
        "✅ Clear execution order"
    ],
    "Professionalism": [
        "✅ Clean, readable output",
        "✅ Consistent formatting",
        "✅ Error handling where needed",
        "✅ Summary at the end"
    ]
}
```

```python
# WHAT: Print the checklist by looping over the dictionary, then recap the whole notebook.
# WHY: Iterating over dict items is the standard pattern for rendering structured content.

for category, practices in best_practices.items():
    print(f"\n📋 {category}:")
    for practice in practices:
        print(f"   {practice}")
# ============================================================================
# ============================================================================
print("\n" + "=" * 70)
print("Summary")
print("=" * 70)
print("""
✅ What you learned:
1. Notebook Structure: Organize with markdown and code cells
2. Code Organization: Clear names, comments, logical flow
3. Reproducibility: Seeds, dependencies, clear execution
4. Best Practices: Professional, shareable notebooks

🎯 Key Takeaways:
- Structure: Markdown explains, code demonstrates
- Organization: Clear names and comments
- Reproducibility: Seeds and documentation
- Professionalism: Clean, organized, shareable

📚 Next Steps:
- Apply these practices to all future notebooks
- Example 6: Data Structures (use these practices!)
- Unit 2: Data Cleaning (professional notebooks essential)
""")
print("✅ Jupyter Notebook best practices understood!")
```
