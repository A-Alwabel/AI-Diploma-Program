## 🎯 The case: an exception that destroyed a rocket in 37 seconds

**Kourou, French Guiana, 4 June 1996, 12:34 UTC.** Ariane 5 flight V88 lifted off. Inside
the inertial reference system, a piece of Ada code converted a 64-bit floating-point value
— the horizontal bias — into a **16-bit signed integer**. On Ariane 4 that number had
never been large enough to overflow. Ariane 5 flew a different trajectory, the number was
larger, and the conversion overflowed.

The software **did** detect the problem and raise an exception. What it did next is the
lesson: the design treated an unhandled exception as proof of hardware failure, so the
processor shut itself down — and so did the identical backup unit, running the identical
code, a fraction of a second earlier. The active unit then emitted a diagnostic bit
pattern which the on-board computer read as *flight data*, commanded full nozzle
deflection, and the launcher broke up and self-destructed **37 seconds after liftoff**.
More than **US$370 million** was lost. The ESA/CNES inquiry board chaired by
Jacques-Louis Lions wrote that the exception *"was detected, but inappropriately handled
because the view had been taken that software should be considered correct until it is
shown to be at fault."*

Three things you are about to practise are all in that paragraph: the **data type** a
number is stored in, the **condition** that should have guarded the conversion, and the
**`try`/`except`** that decides what happens when reality disagrees with your assumption.

**What goes wrong without this, at your scale.** In Part 4 below you loop over 891 real
Titanic passengers. 177 of them have no recorded age. A loop that does arithmetic on
`Age` without a condition or a `try` does not give you a slightly wrong answer — it stops
at the first blank row and gives you nothing at all, after having done the work.


```python
# WHAT: Practise arithmetic operators, then apply them to REAL ticket fares from the Titanic manifest.
# WHY: The operators are worth learning because they compute real quantities - so the
#      "real-world example" below reads real numbers instead of numbers we typed in.

# No external libraries needed for basic Python operations - csv is standard library
import csv

print("✅ Ready to learn Python basics!")
print("\n" + "=" * 70)
print("=" * 70)

# ============================================================================
# ============================================================================
print("\n" + "=" * 70)
print("PART 1: Arithmetic Operations")
print("=" * 70)

# Basic arithmetic operations
a = 10
b = 3

print(f"\nGiven: a = {a}, b = {b}")
print(f"Addition (a + b): {a + b}")
print(f"Subtraction (a - b): {a - b}")
print(f"Multiplication (a * b): {a * b}")
print(f"Division (a / b): {a / b}")
print(f"Floor division (a // b): {a // b}  # Rounds down to nearest integer")
print(f"Modulo (a % b): {a % b}  # Returns remainder")
print(f"Exponentiation (a ** b): {a ** b}  # a to the power of b")

# Real-world example: read real data and average it
DATA_DIR = '../../../Course 04/datasets/raw/'
with open(DATA_DIR + 'titanic.csv', newline='', encoding='utf-8') as f:
    passengers = list(csv.DictReader(f))

fares = [float(p['Fare']) for p in passengers]
total = sum(fares)
average = total / len(fares)
print(f"\n📊 Real-world example: average ticket fare on the RMS Titanic")
print(f"   Rows read from titanic.csv: {len(fares)} real passengers")
print(f"   First five fares: {[round(x, 2) for x in fares[:5]]}")
print(f"   Sum: £{total:,.2f}")
print(f"   Count: {len(fares)}")
print(f"   Average: £{average:.2f}")

# ============================================================================
```

```python
# WHAT: Practise if / elif / else, then use conditions to categorise REAL ticket fares.
# WHY: Conditions are how you turn a continuous measurement into a decision - here, into fare bands.

# ============================================================================
print("\n" + "=" * 70)
print("PART 2: Conditional Statements (if/elif/else)")
print("=" * 70)

# Simple if statement
age = 20
print(f"\nExample 1: Simple if statement")
print(f"   Age: {age}")
if age >= 18:
    print("   ✅ Result: Adult (age >= 18)")

# if-else statement
score = 75
print(f"\nExample 2: if-else statement")
print(f"   Score: {score}")
if score >= 60:
    print("   ✅ Result: Pass (score >= 60)")
else:
    print("   ❌ Result: Fail (score < 60)")

# if-elif-else statement
temperature = 25
print(f"\nExample 3: if-elif-else (multiple conditions)")
print(f"   Temperature: {temperature}°C")
if temperature > 30:
    print("   🔥 Result: Hot")
elif temperature > 20:
    print("   ☀️  Result: Warm")
elif temperature > 10:
    print("   🌤️  Result: Mild")
else:
    print("   ❄️  Result: Cold")

# Real-world example: categorise REAL fares
print(f"\n📊 Real-world example: categorise the first 8 real ticket fares")
for fare in fares[:8]:
    if fare >= 100:
        category = "Expensive"
    elif fare >= 30:
        category = "Mid-range"
    elif fare > 0:
        category = "Cheap"
    else:
        category = "Free (£0.00 - a real value in this file, not missing data)"
    print(f"   £{fare:7.2f} → {category}")

# ============================================================================
```

```python
# WHAT: Practise for and while loops, then loop over REAL passenger records.
# WHY: Loops let you process every row of a dataset - which is what the rest of this course does.

# ============================================================================
print("\n" + "=" * 70)
print("PART 3: Loops")
print("=" * 70)

# for loop: Iterate over a list
print("\nExample 1: for loop - iterate over list")
fruits = ["apple", "banana", "orange"]
print(f"   Fruits: {fruits}")
for fruit in fruits:
    print(f"   - {fruit}")

# for loop with range
print("\nExample 2: for loop - using range()")
print("   Counting from 1 to 5:")
for i in range(1, 6):  # range(1, 6) gives [1, 2, 3, 4, 5]
    print(f"   - Number {i}")

# while loop
print("\nExample 3: while loop - repeat until condition")
count = 0
print("   Counting from 0 to 4:")
while count < 5:
    print(f"   - Count: {count}")
    count += 1  # Increment count by 1

# Real-world example: accumulate over real records
print("\n📊 Real-world example: running total of the first 5 real fares")
running_total = 0
for fare in fares[:5]:
    running_total += fare
    print(f"   - Fare: £{fare:7.2f}, Running total: £{running_total:8.2f}")
print(f"   ✅ Total for those 5 passengers: £{running_total:.2f}")
print(f"   ✅ Total for all {len(fares)} passengers: £{sum(fares):,.2f}")

# List comprehension (advanced, but useful)
print("\nExample 4: List comprehension (advanced)")
numbers = [1, 2, 3, 4, 5]
squared = [x ** 2 for x in numbers]  # Square each number
print(f"   Original: {numbers}")
print(f"   Squared: {squared}")
print(f"   Same idea on real data - every passenger's age, as a list:")
ages = [float(p['Age']) for p in passengers if p['Age']]   # skip the blanks
print(f"   {len(ages)} ages recorded, {len(passengers) - len(ages)} blank in the file")

# ============================================================================
```

```python
# WHAT: Combine loops and conditions to filter REAL passenger records, then summarise the notebook.
# WHY: Filter-with-a-loop is the pattern every pandas query hides - writing it once explains the rest.

# PART 4: COMBINING LOOPS AND CONDITIONS
# ============================================================================
print("\n" + "=" * 70)
print("PART 4: Combining Loops and Conditions")
print("=" * 70)

# Filter data using loop + condition, on the real manifest
print("\n📊 Real-world example: find the first-class passengers who survived")
survivors_first_class = []
for p in passengers:
    if int(p['Pclass']) == 1 and int(p['Survived']) == 1:
        survivors_first_class.append(p['Name'])

print(f"   Total passengers on the manifest: {len(passengers)}")
print(f"   First-class survivors found: {len(survivors_first_class)}")
print("   First five:")
for name in survivors_first_class[:5]:
    print(f"   ✅ {name}")

# Two counters in one pass - the pattern behind groupby()
survived = sum(1 for p in passengers if int(p['Survived']) == 1)
print(f"\n   Survived: {survived} of {len(passengers)} "
      f"({survived / len(passengers):.1%})")

# ============================================================================
# ============================================================================
print("\n" + "=" * 70)
print("Summary")
print("=" * 70)
print(f"""
✅ What you learned (and applied to {len(passengers)} real Titanic passenger records):
   1. Arithmetic Operations: +, -, *, /, //, %, **
      -> average real ticket fare: £{average:.2f}
   2. Conditional Statements: if, elif, else (make decisions)
      -> sorted real fares into Expensive / Mid-range / Cheap / Free
   3. Loops: for, while (repeat code)
      -> running totals over real fares
   4. Combining loops and conditions (filter/process data)
      -> {len(survivors_first_class)} first-class survivors found by hand

🎯 Key Takeaways:
   - Arithmetic: Calculate values, transform data
   - Conditions: Make decisions, filter data
   - Loops: Process multiple items, repeat operations
   - Together: Foundation for all data science programming
   - Real files have gaps: {len(passengers) - len(ages)} of {len(passengers)} passengers have no recorded age,
     which is why `if p['Age']` appears in the comprehension above

📚 Next Steps:
   - Example 5: Jupyter Notebooks Best Practices
   - Example 2: Pandas & NumPy Basics (uses loops and conditions!)
   - Unit 2: Data Cleaning (heavily uses loops and conditions)
""")
print("✅ Python basics concepts understood!")
```


**Lambda functions** are small anonymous functions: `lambda x: x * 2`. Useful with `map`, `filter`, `sort`, or pandas.

**Exception handling** (`try` / `except`) catches errors so your program can recover instead of crashing. Essential when reading files, calling APIs, or processing messy data.

```python
# WHAT: Demonstrate lambda functions (with map/filter) and exception handling with try/except.
# WHY: Lambdas give quick one-line transforms, and exceptions let a program handle bad input gracefully instead of crashing.

# Lambda functions
print("Lambda functions")
print("-" * 40)
double = lambda x: x * 2
print(f"double(5) = {double(5)}")
vals = [1, 2, 3, 4, 5]
tripled = list(map(lambda x: x * 3, vals))
print(f"map(lambda x: x*3, {vals}) = {tripled}")
evens = list(filter(lambda x: x % 2 == 0, vals))
print(f"filter(even, {vals}) = {evens}")

# Exception handling (try / except)
print("-" * 40)
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("  Cannot divide by zero; returning None")
        return None
print(f"safe_divide(10, 2) = {safe_divide(10, 2)}")
print(f"safe_divide(10, 0) = {safe_divide(10, 0)}")

try:
    num = int("not a number")
except ValueError as e:
    print(f"  ValueError caught: {e}")
print("  Program continues after handling the error.")
```

## ⚠️ Where this breaks

- **Python loops do not scale, and this is measurable.** Unit 1 lesson 02 timed the same
  computation on 1,000,000 real rows: **0.0053 s** vectorized against **3.76 s** row by
  row — 713× on identical data. Every loop you write here is correct and, past a few
  hundred thousand rows, the wrong tool. From Unit 2 onwards, a loop over a DataFrame is a
  code smell.
- **`try`/`except` can hide the bug instead of handling it.** A bare `except:` that
  swallows everything turns a crash into silently wrong numbers, which is strictly worse:
  a crash gets fixed on Monday, a wrong average gets published. Catch the specific
  exception you expect (`ValueError`, `KeyError`, `ZeroDivisionError`), and let anything
  else through.
- **The assumption that must hold is that your data matches your types.** Ariane 5's
  conversion was correct code under Ariane 4's assumptions. `int(row['Age'])` is correct
  code under the assumption that every age is recorded. 177 rows in this file say
  otherwise — and the file never warns you.
- **Cheaper alternative:** when the rule is "apply this test to every row", stop writing
  the loop. `df[df['Pclass'] == 1]` is faster, shorter, and cannot silently skip a row the
  way a hand-rolled `try` inside a loop can. Keep loops for genuinely sequential work —
  reading a stream, retrying a request, walking a directory.
