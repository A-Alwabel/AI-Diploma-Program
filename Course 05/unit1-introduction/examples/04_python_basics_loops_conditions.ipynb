## Why Python Basics Matter | لماذا يهم أساسيات بايثون

- **Arithmetic Operations**: Calculate statistics, transform data, perform computations
- **Loops**: Process every row in a dataset, repeat operations, iterate over data
- **Conditions**: Filter data, make decisions, handle different cases
- **Together**: They form the foundation for all data science programming

**Common Student Questions:**
- **Q: Why do I need loops?**
  - Answer: Loops let you repeat code for every item in a list/dataset
  - Example: Calculate average of 1000 numbers (without loops, you'd write 1000 lines!)
  - In data science: Process every row, apply transformations to all data
  
- **Q: When do I use if/else?**
  - Answer: Use conditions to make decisions based on data values
  - Example: "If age > 18, then adult; else, child"
  - In data science: Filter data, handle missing values, categorize data

## Introduction

**Python programming fundamentals** are essential for data science, covering arithmetic operations, control flow with loops and conditions, and basic problem-solving. This notebook provides hands-on practice with working examples.

```python
# No external libraries needed for basic Python operations
print("✅ Ready to learn Python basics!")
print("\n" + "=" * 70)
print("Python Programming: Basic Tasks | البرمجة الأساسية في بايثون")
print("=" * 70)

# ============================================================================
# PART 1: ARITHMETIC OPERATIONS | العمليات الحسابية
# ============================================================================
print("\n" + "=" * 70)
print("PART 1: Arithmetic Operations | العمليات الحسابية")
print("=" * 70)

# Basic arithmetic operations
a = 10
b = 3

print(f"\nGiven: a = {a}, b = {b}")
print(f"Addition (a + b): {a + b}")
print(f"Subtraction (a - b): {a - b}")
print(f"Multiplication (a * b): {a * b}")
print(f"Division (a / b): {a / b}")
print(f"Floor division (a // b): {a // b}  # Rounds down to nearest integer")
print(f"Modulo (a % b): {a % b}  # Returns remainder")
print(f"Exponentiation (a ** b): {a ** b}  # a to the power of b")

# Real-world example: Calculate average
numbers = [85, 90, 78, 92, 88]
total = sum(numbers)  # Sum all numbers
average = total / len(numbers)  # Divide by count
print(f"\n📊 Real-world example: Calculate average")
print(f"   Numbers: {numbers}")
print(f"   Sum: {total}")
print(f"   Count: {len(numbers)}")
print(f"   Average: {average}")

# ============================================================================
```

```python
# PART 2: CONDITIONAL STATEMENTS | البيانات الشرطية
# ============================================================================
print("\n" + "=" * 70)
print("PART 2: Conditional Statements (if/elif/else) | البيانات الشرطية")
print("=" * 70)

# Simple if statement
age = 20
print(f"\nExample 1: Simple if statement")
print(f"   Age: {age}")
if age >= 18:
    print("   ✅ Result: Adult (age >= 18)")

# if-else statement
score = 75
print(f"\nExample 2: if-else statement")
print(f"   Score: {score}")
if score >= 60:
    print("   ✅ Result: Pass (score >= 60)")
else:
    print("   ❌ Result: Fail (score < 60)")

# if-elif-else statement
temperature = 25
print(f"\nExample 3: if-elif-else (multiple conditions)")
print(f"   Temperature: {temperature}°C")
if temperature > 30:
    print("   🔥 Result: Hot")
elif temperature > 20:
    print("   ☀️  Result: Warm")
elif temperature > 10:
    print("   🌤️  Result: Mild")
else:
    print("   ❄️  Result: Cold")

# Real-world example: Categorize data
data_values = [15, 25, 35, 5, 45]
print(f"\n📊 Real-world example: Categorize data values")
print(f"   Values: {data_values}")
for value in data_values:
    if value >= 30:
        category = "High"
    elif value >= 20:
        category = "Medium"
    else:
        category = "Low"
    print(f"   {value} → {category}")

# ============================================================================
```

```python
# PART 3: LOOPS | الحلقات
# ============================================================================
print("\n" + "=" * 70)
print("PART 3: Loops | الحلقات")
print("=" * 70)

# for loop: Iterate over a list
print("\nExample 1: for loop - iterate over list")
fruits = ["apple", "banana", "orange"]
print(f"   Fruits: {fruits}")
for fruit in fruits:
    print(f"   - {fruit}")

# for loop with range
print("\nExample 2: for loop - using range()")
print("   Counting from 1 to 5:")
for i in range(1, 6):  # range(1, 6) gives [1, 2, 3, 4, 5]
    print(f"   - Number {i}")

# while loop
print("\nExample 3: while loop - repeat until condition")
count = 0
print("   Counting from 0 to 4:")
while count < 5:
    print(f"   - Count: {count}")
    count += 1  # Increment count by 1

# Real-world example: Process data
print("\n📊 Real-world example: Process data with loop")
sales = [100, 150, 200, 180, 220]
total_sales = 0
print(f"   Daily sales: {sales}")
for sale in sales:
    total_sales += sale
    print(f"   - Day sales: ${sale}, Running total: ${total_sales}")
print(f"   ✅ Total sales: ${total_sales}")

# List comprehension (advanced, but useful)
print("\nExample 4: List comprehension (advanced)")
numbers = [1, 2, 3, 4, 5]
squared = [x ** 2 for x in numbers]  # Square each number
print(f"   Original: {numbers}")
print(f"   Squared: {squared}")

# ============================================================================
```

```python
# PART 4: COMBINING LOOPS AND CONDITIONS | دمج الحلقات والشروط
# ============================================================================
print("\n" + "=" * 70)
print("PART 4: Combining Loops and Conditions | دمج الحلقات والشروط")
print("=" * 70)

# Filter data using loop + condition
print("\n📊 Real-world example: Filter data")
scores = [45, 78, 92, 56, 88, 67, 95, 42]
print(f"   All scores: {scores}")
print("   Finding passing scores (>= 60):")
passing_scores = []
for score in scores:
    if score >= 60:
        passing_scores.append(score)
        print(f"   ✅ {score} - Pass")
    else:
        print(f"   ❌ {score} - Fail")
print(f"   Passing scores: {passing_scores}")

# ============================================================================
# SUMMARY | الملخص
# ============================================================================
print("\n" + "=" * 70)
print("Summary | الملخص")
print("=" * 70)
print("""
✅ What you learned:
   1. Arithmetic Operations: +, -, *, /, //, %, **
   2. Conditional Statements: if, elif, else (make decisions)
   3. Loops: for, while (repeat code)
   4. Combining loops and conditions (filter/process data)

🎯 Key Takeaways:
   - Arithmetic: Calculate values, transform data
   - Conditions: Make decisions, filter data
   - Loops: Process multiple items, repeat operations
   - Together: Foundation for all data science programming

📚 Next Steps:
   - Example 5: Jupyter Notebooks Best Practices
   - Example 2: Pandas & NumPy Basics (uses loops and conditions!)
   - Unit 2: Data Cleaning (heavily uses loops and conditions)
""")
print("✅ Python basics concepts understood!")
```

## Lambda functions and exception handling (DETAILED_UNIT_DESCRIPTIONS.md Unit 1)

**Lambda functions** are small anonymous functions: `lambda x: x * 2`. Useful with `map`, `filter`, `sort`, or pandas.

**Exception handling** (`try` / `except`) catches errors so your program can recover instead of crashing. Essential when reading files, calling APIs, or processing messy data.

```python
# Lambda functions
print("Lambda functions | دوال لامدا")
print("-" * 40)
double = lambda x: x * 2
print(f"double(5) = {double(5)}")
vals = [1, 2, 3, 4, 5]
tripled = list(map(lambda x: x * 3, vals))
print(f"map(lambda x: x*3, {vals}) = {tripled}")
evens = list(filter(lambda x: x % 2 == 0, vals))
print(f"filter(even, {vals}) = {evens}")

# Exception handling (try / except)
print("\nException handling | معالجة الاستثناءات")
print("-" * 40)
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("  Cannot divide by zero; returning None")
        return None
print(f"safe_divide(10, 2) = {safe_divide(10, 2)}")
print(f"safe_divide(10, 0) = {safe_divide(10, 0)}")

try:
    num = int("not a number")
except ValueError as e:
    print(f"  ValueError caught: {e}")
print("  Program continues after handling the error.")
```
