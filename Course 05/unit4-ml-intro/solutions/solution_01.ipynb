```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import mean_squared_error, r2_score, accuracy_score, confusion_matrix, classification_report
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams['axes.unicode_minus'] = False
sns.set_style("whitegrid")

np.random.seed(42)
n = 200
X_reg = np.random.randn(n, 3)
y_reg = 2.5*X_reg[:,0] + 1.5*X_reg[:,1] - 0.5*X_reg[:,2] + np.random.randn(n)*0.5
X_clf = np.random.randn(n, 2)
y_clf = (X_clf[:,0] + X_clf[:,1] > 0).astype(int)
```

### Task 1: Linear regression

```python
print("=" * 60)
print("Task 1: Linear Regression")
print("=" * 60)
X_tr, X_te, y_tr, y_te = train_test_split(X_reg, y_reg, test_size=0.25, random_state=42)
model = LinearRegression().fit(X_tr, y_tr)
pred = model.predict(X_te)
print(f"MSE: {mean_squared_error(y_te, pred):.4f}, R²: {r2_score(y_te, pred):.4f}")
plt.figure(figsize=(5,4))
plt.scatter(y_te, pred, alpha=0.6); plt.xlabel('Actual'); plt.ylabel('Predicted'); plt.title('Pred vs Actual')
plt.plot([y_te.min(), y_te.max()], [y_te.min(), y_te.max()], 'r--'); plt.tight_layout(); plt.show()
```

### Task 2: Classification

```python
print("\n" + "=" * 60)
print("Task 2: Classification")
print("=" * 60)
Xc_tr, Xc_te, yc_tr, yc_te = train_test_split(X_clf, y_clf, test_size=0.25, random_state=42)
lr = LogisticRegression(random_state=42).fit(Xc_tr, yc_tr)
dt = DecisionTreeClassifier(random_state=42).fit(Xc_tr, yc_tr)
print(f"LogReg accuracy: {accuracy_score(yc_te, lr.predict(Xc_te)):.4f}")
print(f"DecisionTree accuracy: {accuracy_score(yc_te, dt.predict(Xc_te)):.4f}")
print("LogReg confusion matrix:"); print(confusion_matrix(yc_te, lr.predict(Xc_te)))
print("DT confusion matrix:"); print(confusion_matrix(yc_te, dt.predict(Xc_te)))
```

### Task 3: Model comparison

```python
print("\n" + "=" * 60)
print("Task 3: Model Comparison")
print("=" * 60)
print("LogReg report:"); print(classification_report(yc_te, lr.predict(Xc_te)))
print("DT report:"); print(classification_report(yc_te, dt.predict(Xc_te)))
print("\n" + "=" * 60)
print("Exercise 1 Complete!")
print("=" * 60)
```
