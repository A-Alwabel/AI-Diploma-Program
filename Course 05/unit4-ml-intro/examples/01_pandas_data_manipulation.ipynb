## 🎯 The case: Berkeley, 1973 — a bias that vanished when someone used groupby

The University of California, Berkeley graduate admissions for autumn 1973 looked
indefensible. Of **8,442 male applicants, 44% were admitted**; of **4,321 women, 35%**.
Across 12,763 applicants the gap was far too large to be chance, and the university faced
the obvious accusation.

Then Bickel, Hammel and O'Connell broke the same numbers down **by department** and
published the result in *Science* in 1975 (*Sex Bias in Graduate Admissions: Data From
Berkeley*, 187:398–404). Department by department, there was a small bias **in favour of
women**. Women had applied disproportionately to the departments with the lowest admission
rates for everybody; men had applied to the ones with the highest. The aggregate reversed
the direction of every one of its parts.

This is **Simpson's paradox**, and the only tool that exposes it is the one this notebook
is about: `groupby`. The single number was not wrong. It was an average over groups that
should never have been averaged.

**What goes wrong without this.** You are about to compute that first-class women survived
the Titanic at **96.8%** and third-class men at **13.5%**, from two filters. Report only
the overall 38.4% survival rate and both facts disappear. The aggregate is always
available, always defensible, and frequently the least informative number in the table.

```python
# Imports
import pandas as pd
import numpy as np
```

```python
# WHAT: Load the real Titanic passenger manifest.
# WHY: Every pandas skill below - selecting, filtering, grouping, string work -
#      is worth learning because real tables are messy; this one has 177 missing ages.

print("PART 1: Load the data")
print("-" * 70)

DATA_DIR = '../../../Course 04/datasets/raw/'
df = pd.read_csv(DATA_DIR + 'titanic.csv')

print("Real Titanic manifest (first 6 rows):")
print(df.head(6).to_string())
print(f"\nDataFrame shape: {df.shape}")
print("DataFrame info:")
print(df.info())
print(f"\nMissing values per column (real gaps, nothing removed):")
print(df.isnull().sum()[df.isnull().sum() > 0].to_string())

# ============================================================================
```

```python
# WHAT: Select rows and columns two ways - loc (by label) and iloc (by position).
# WHY: Mixing the two up is the most common pandas bug; doing both side by side fixes it early.

# ============================================================================
print("\n" + "=" * 70)
print("PART 2: Data Selection")
print("=" * 70)

# Using loc (label-based)
print("\n✅ Example 1: Using loc (Label-based Selection)")
print("-" * 70)
print("Select rows by label (index 0-2):")
print(df.loc[0:2, ['Name', 'Sex', 'Age', 'Pclass']].to_string())
print("\nSelect specific columns:")
print(df.loc[:, ['Name', 'Age', 'Fare']].head().to_string())
print("\nSelect rows and columns:")
print(df.loc[0:2, ['Name', 'Fare']].to_string())

# Using iloc (integer position-based)
print("\n✅ Example 2: Using iloc (Position-based Selection)")
print("-" * 70)
print("Select first 3 rows:")
print(df.iloc[0:3, 0:6].to_string())
print("\nSelect specific columns by position (Survived, Pclass, Name):")
print(df.iloc[:, [1, 2, 3]].head().to_string())

# ============================================================================
```

```python
# WHAT: Filter rows with boolean masks - single and combined conditions.
# WHY: Filtering is how you ask a table a question; every analysis below starts with one.

# ============================================================================
print("\n" + "=" * 70)
print("PART 3: Data Filtering")
print("=" * 70)

# Boolean indexing
print("\n✅ Example 3: Boolean Indexing (Filtering)")
print("-" * 70)
first_class = df[df['Pclass'] == 1]
print(f"First-class passengers: {len(first_class)} of {len(df)}")
print(first_class[['Name', 'Sex', 'Age', 'Fare']].head().to_string())

expensive = df[df['Fare'] > 100]
print(f"\nPassengers who paid more than £100: {len(expensive)}")
print(f"  Their survival rate: {expensive['Survived'].mean():.1%}")
print(f"  Everyone else's survival rate: {df[df['Fare'] <= 100]['Survived'].mean():.1%}")

# Multiple conditions
print("\n✅ Example 4: Multiple Conditions")
print("-" * 70)
women_first = df[(df['Sex'] == 'female') & (df['Pclass'] == 1)]
print(f"Women in first class: {len(women_first)}, "
      f"survival rate {women_first['Survived'].mean():.1%}")
men_third = df[(df['Sex'] == 'male') & (df['Pclass'] == 3)]
print(f"Men in third class:  {len(men_third)}, "
      f"survival rate {men_third['Survived'].mean():.1%}")
print("\n💡 Two filters just uncovered the single starkest pattern in the manifest.")

# ============================================================================
```

```python
# WHAT: Add computed columns, then group and aggregate.
# WHY: groupby answers 'how does this differ between groups?' - the workhorse of analysis.

# ============================================================================
print("\n" + "=" * 70)
print("PART 4: Data Manipulation")
print("=" * 70)

# Adding columns
print("\n✅ Example 5: Adding Columns")
print("-" * 70)
df['family_size'] = df['SibSp'] + df['Parch'] + 1   # siblings/spouses + parents/children + self
df['is_alone'] = (df['family_size'] == 1)
print(df[['Name', 'SibSp', 'Parch', 'family_size', 'is_alone']].head().to_string())
print(f"\nTravelling alone: {df['is_alone'].sum()} of {len(df)} passengers "
      f"({df['is_alone'].mean():.1%})")

# Grouping and aggregation
print("\n✅ Example 6: Grouping and Aggregation")
print("-" * 70)
class_stats = df.groupby('Pclass').agg({
    'Fare': ['mean', 'max', 'count'],
    'Age': 'mean',
    'Survived': 'mean'
}).round(2)
print("Statistics by travel class:")
print(class_stats.to_string())
print("\n💡 'Age' counts differ from 'Fare' counts because 177 ages are missing -")
print("   pandas skips NaN in mean() silently. Always check what you averaged over.")

# ============================================================================
```

```python
# WHAT: Sort rows, then pull structure out of the free-text Name column with .str.
# WHY: String columns hide features - here the social title (Mr, Mrs, Miss, Master) is buried in a name.

# PART 5: DATA TRANSFORMATION
# ============================================================================
print("\n" + "=" * 70)
print("PART 5: Data Transformation")
print("=" * 70)

# Sorting
print("\n✅ Example 7: Sorting Data")
print("-" * 70)
sorted_by_fare = df.sort_values('Fare', ascending=False)
print("The five most expensive tickets on the manifest:")
print(sorted_by_fare[['Name', 'Pclass', 'Fare', 'Survived']].head().to_string())

# String operations
print("\n✅ Example 8: String Operations")
print("-" * 70)
# Names look like 'Braund, Mr. Owen Harris' - the title sits between ', ' and '.'
df['title'] = df['Name'].str.split(', ').str[1].str.split('.').str[0].str.strip()
df['name_length'] = df['Name'].str.len()
print("Extracted titles (top 6):")
print(df['title'].value_counts().head(6).to_string())
print("\nSurvival rate by extracted title (groups of 5+ passengers):")
title_counts = df['title'].value_counts()
big_titles = title_counts[title_counts >= 5].index
print(df[df['title'].isin(big_titles)].groupby('title')['Survived']
        .agg(['mean', 'size']).round(3).to_string())

# ============================================================================
# ============================================================================
print("\n" + "=" * 70)
print("Summary")
print("=" * 70)
print(f"""
✅ What you learned, on the real {len(df)}-passenger Titanic manifest:
   1. Data Selection: loc (labels), iloc (positions)
   2. Data Filtering: Boolean indexing, multiple conditions
      -> women in 1st class survived at {women_first['Survived'].mean():.0%},
         men in 3rd class at {men_third['Survived'].mean():.0%}
   3. Data Manipulation: computed columns (family_size), groupby, aggregate
   4. Data Transformation: sort, and .str to mine a title out of every name

🎯 Key Takeaways:
   - loc: Label-based selection (use index/column names)
   - iloc: Position-based selection (use integer positions)
   - Boolean indexing: Filter rows based on conditions
   - Groupby: Aggregate data by categories
   - Real data has gaps: {df['Age'].isna().sum()} ages are missing, and mean() ignores them silently

📚 Next Steps:
   - Example 2: Data Preparation for ML (we clean this same manifest for a model)
   - Example 3: Implementing ML Models
""")
print("✅ Pandas concepts understood!")
```

## ⚠️ Where this breaks

- **Every aggregate can reverse when you split it.** Simpson's paradox is not exotic; it
  is what happens whenever group sizes differ and the groups behave differently. There is
  no statistical test that tells you which level of aggregation is "correct" — that is a
  question about the causal structure, and it has to be answered outside the data.
- **`groupby` will happily average things that must not be averaged.** The mean of a
  ratio is not the ratio of means; the mean of `Survived` is a rate only because the column
  is 0/1; averaging a percentage across groups of unequal size is almost always wrong.
- **Chained assignment fails silently.** `df[df.Age > 30]['Fare'] = 0` may modify a copy
  and leave the original untouched — with a warning most people have learned to ignore.
  Use `.loc[mask, 'Fare'] = 0`.
- **`.str` extraction is regex on human text.** The titles came out clean here because a
  1912 manifest is formatted consistently. Real free-text fields have typos, mixed
  languages, missing separators and encodings. Always print the value counts of what you
  extracted before using it — including the tail.
- **The assumption that must hold: one row is one unit of the thing you are counting.**
  Unit 2 lesson 02 showed 210 passengers sharing ticket numbers. If you group by ticket
  you are counting *bookings*, not *people*, and every rate changes.
- **When pandas is the wrong tool:** if the data already lives in a database, do the
  grouping in SQL on the server. If the table is bigger than memory, Unit 5's Dask and
  Spark lessons exist for that. And if the answer needs a confidence interval rather than
  a point estimate, `groupby().mean()` is the start of the analysis, not the end.
