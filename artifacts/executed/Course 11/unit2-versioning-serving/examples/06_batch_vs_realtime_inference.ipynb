## Introduction

**Batch inference** processes multiple inputs at once, while **real-time inference** processes requests immediately. Each approach has different use cases and trade-offs.

```python
import numpy as np
import time

print("✅ Libraries imported!")
print("\nTesting Batch vs Real-Time Inference")
print("=" * 60)

print("\nBatch Inference:")
print("  - Process multiple inputs together")
print("  - Higher throughput")
print("  - Lower latency per item")
print("  - Use case: Large datasets, offline processing")

print("\nReal-Time Inference:")
print("  - Process requests immediately")
print("  - Lower latency per request")
print("  - Higher latency per item")
print("  - Use case: Interactive applications, APIs")

print("\nComparison:")
print("  Batch: Better for throughput, offline")
print("  Real-time: Better for latency, interactive")

print("\n✅ Batch vs real-time concepts understood!")
```
