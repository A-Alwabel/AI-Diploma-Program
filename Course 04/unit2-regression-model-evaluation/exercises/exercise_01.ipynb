```python
# Import the evaluation toolkit: three linear models (plain, Ridge, Lasso),
# a scaler, and the error metrics you will compare them with.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score
```

Generate sample data with multicollinearity


```python
# Load a REAL dataset whose features are genuinely correlated with one another.
# WHY: correlated features are exactly the situation where Ridge and Lasso behave
# differently from plain Linear Regression — and in the California Housing data the
# correlations are real, not something we engineered. Latitude and Longitude both encode
# the same thing (where in California the block group is), and richer neighbourhoods
# genuinely have more rooms per household. The cell prints every pair with |r| > 0.5.
from sklearn.datasets import fetch_california_housing

california = fetch_california_housing()
df = pd.DataFrame(california.data, columns=california.feature_names)
df["target"] = california.target          # median house value, in $100,000s

# Classroom-size it: 300 census block groups drawn at random (random_state=42).
df = df.sample(n=300, random_state=42).reset_index(drop=True)

print("Dataset loaded!")
print(f"Shape: {df.shape}")
print(f"Source: California Housing (1990 US census), sampled from "
      f"{california.data.shape[0]:,} block groups with random_state=42")
print(f"Features: {list(california.feature_names)}")
print()
print("Real correlations between the features (this is what Ridge and Lasso must cope with):")
corr = df[list(california.feature_names)].corr()
strong = [(a, b, corr.loc[a, b]) for i, a in enumerate(california.feature_names)
          for b in california.feature_names[i + 1:] if abs(corr.loc[a, b]) > 0.5]
for a, b, r in sorted(strong, key=lambda t: -abs(t[2])):
    print(f"  {a:12s} vs {b:12s}: r = {r:+.2f}")
```
