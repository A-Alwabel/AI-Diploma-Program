```python
# Import the evaluation toolkit: three linear models (plain, Ridge, Lasso),
# a scaler, and the error metrics you will compare them with.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score
```

Generate sample data with multicollinearity


```python
# Create the practice dataset: 8 features where two are CORRELATED copies.
# WHY: correlated features are exactly the situation where Ridge and Lasso
# behave differently from plain Linear Regression.
np.random.seed(42)
n_samples = 300

X = np.random.randn(n_samples, 8)
# Make some features correlated
X[:, 2] = X[:, 0] + 0.3 * np.random.randn(n_samples)
X[:, 3] = X[:, 1] + 0.3 * np.random.randn(n_samples)

y = (2 * X[:, 0] + 1.5 * X[:, 1] - X[:, 2] + 3 * X[:, 3] +
     0.5 * X[:, 4] + np.random.normal(0, 0.5, n_samples))

df = pd.DataFrame(X, columns=[f'feature_{i}' for i in range(8)])
df['target'] = y

print("Dataset loaded!")
print(f"Shape: {df.shape}")
```
