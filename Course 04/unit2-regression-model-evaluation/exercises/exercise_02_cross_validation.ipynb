```python
# Import the cross-validation toolkit: KFold and cross_val_score do the
# splitting and scoring loops for you.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, KFold, cross_val_score
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score
```

Generate sample regression data (5 features, known coefficients, small noise)

```python
# Load REAL regression data for the cross-validation exercise.
# WHY: cross-validation exists because a single train/test split of REAL data can be
# lucky or unlucky. You will only see that spread between folds on data with genuine
# variability — a made-up dataset drawn from one clean formula barely moves.
from sklearn.datasets import fetch_california_housing

california = fetch_california_housing()
FEATURES = ["MedInc", "HouseAge", "AveRooms", "AveOccup", "Population"]

full = pd.DataFrame(california.data, columns=california.feature_names)
full["target"] = california.target        # median house value, in $100,000s

# Classroom-size it: 300 census block groups drawn at random (random_state=123).
df = full[FEATURES + ["target"]].sample(n=300, random_state=123).reset_index(drop=True)

print("Dataset loaded!")
print(f"Shape: {df.shape}")
print(f"Source: California Housing (1990 US census), 5 real features")
print(f"Features: {FEATURES}")
print(f"Target: median house value in $100,000s, real range "
      f"{df['target'].min():.2f} to {df['target'].max():.2f}")
print()
print("Watch the spread between your folds — on real data it will be noticeably wider")
print("than you might expect, and that spread is the whole point of the exercise.")
```
