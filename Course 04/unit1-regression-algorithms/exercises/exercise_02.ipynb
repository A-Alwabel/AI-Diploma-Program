```python
# Import the regression toolkit: data handling, the LinearRegression model,
# a train/test splitter, and the three error metrics you will report.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.datasets import fetch_california_housing
```

```python
# Load REAL housing data: 20,640 California census block groups, each with the median
# household income of its residents and the median value of its houses.
# WHY real data: there is no hidden formula to recover here. The model can only find
# how strongly the two really move together — which is the honest version of this task.
california = fetch_california_housing()

housing = pd.DataFrame(california.data, columns=california.feature_names)
housing["MedHouseVal"] = california.target

# Classroom-size it: 300 block groups drawn at random (random_state=42 so everyone
# in the class works with exactly the same 300 rows).
housing = housing.sample(n=300, random_state=42).reset_index(drop=True)

df = pd.DataFrame({
    "median_income": housing["MedInc"] * 10_000,        # convert to dollars
    "median_house_value": housing["MedHouseVal"] * 100_000,   # convert to dollars
})

print("Dataset Info:")
print(f"Shape: {df.shape}")
print(df.head())
print()
print(df.describe().round(0))
print()
print("Note the maximum house value: the 1990 census capped recorded values at $500,001,")
print("so some real block groups sit exactly on that ceiling. Watch for it in your plots.")
```
