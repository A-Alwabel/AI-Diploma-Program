```python
# Import the analysis toolkit: pandas/numpy handle the data,
# matplotlib/seaborn draw the plots you will use to explore it.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

**Dataset:** Real 911 emergency dispatch records (Montgomery County, PA)

```python
# Load a REAL, messy dataset to practise exploration on.
# WHY: this dispatch log arrives with genuine missing values (many calls have no ZIP
# recorded) and genuine duplicate rows (the same call logged twice) — the exact defects
# you are about to learn to find. Nothing below was injected for the exercise.
df = pd.read_csv(
    "../../datasets/raw/montgomery_911_calls.csv",
    usecols=["lat", "lng", "zip", "title", "timeStamp", "twp", "addr"],
    nrows=5000,          # the first 5,000 calls, so the notebook stays fast
)

# Split the free-text "title" into its two real parts: the category before the colon
# (EMS / Fire / Traffic) and the specific emergency type after it.
df["category"] = df["title"].str.split(":").str[0]
df["emergency_type"] = df["title"].str.split(": ").str[1].str.strip()

print("Real dataset loaded from the county dispatch log.")
```

```python
# Show the defects the REAL data already contains. You do not need to create any.
print("Missing values per column (all real — these facts were never recorded):")
print(df.isna().sum())
print()
print(f"Fully duplicated rows (the same call logged twice): {df.duplicated().sum()}")
print()
print("Call categories present:")
print(df["category"].value_counts())
```
