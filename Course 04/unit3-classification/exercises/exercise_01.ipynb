```python
# Import the classification toolkit: data handling and the train/test splitter
# you will use in every task below. The data itself comes from a real CSV.
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
```

Generate sample classification data


```python
# Load the REAL Titanic passenger manifest: 891 people, with the class of ticket they
# bought, their age and fare, and whether they survived.
# WHY real data: this file arrives with 177 missing ages and two text columns that a
# model cannot read. Cleaning it is part of the exercise, exactly as it would be at work.
df = pd.read_csv("../../datasets/raw/titanic.csv")

print("Raw dataset loaded!")
print(f"Shape: {df.shape}")
print("\nMissing values (all real):")
print(df.isna().sum()[df.isna().sum() > 0])

# Keep the columns a classifier can actually use, and encode the two text ones.
FEATURES = ["Pclass", "Sex", "Age", "SibSp", "Parch", "Fare", "Embarked"]
df = df[FEATURES + ["Survived"]].copy()

# Fill the real gaps: median age, most common port of embarkation.
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

# Turn the text columns into numbers a model can use.
df["Sex"] = (df["Sex"] == "female").astype(int)             # 1 = female, 0 = male
df = pd.get_dummies(df, columns=["Embarked"], prefix="Port", dtype=int)

# Rename the target so the tasks below read naturally.
df = df.rename(columns={"Survived": "target"})

print("\nCleaned dataset ready!")
print(f"Shape: {df.shape}")
print(f"Features: {[c for c in df.columns if c != 'target']}")
print("\nReal class balance (this is not 50/50 — real outcomes rarely are):")
print(df["target"].value_counts())
print(f"Survival rate: {df['target'].mean():.1%}")
```
